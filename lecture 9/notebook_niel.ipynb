The variational formulation for elasticity is to find the $\mathbf{u} \in V $ such that

$$
\int_\Omega{\bm{\tau_u} : \bm{\epsilon_v} \ d\Omega} = \int_\Omega{\mathbf{f \cdot v} \ d\Omega}
$$
where
$$
\bm{\epsilon_v} = \frac{1}{2} \left[ \left( \nabla \mathbf{v}\right) + \left( \nabla \mathbf{v}\right)^T \right]
$$

This integral is over the defomed configuration of the body. Writing a similar expression for an integral over the reference configuration, we would obtain
$$
\int_{\Omega_0} {\bm{P_u} : \nabla \mathbf{v} \ d\Omega_0} = \int_{\Omega_0}{\mathbf{f_0 \cdot v} \ d\Omega}
$$
where $\bm{P}$ is the first Piola-Kirchoff stress.


We make the "small-displacement" assumption and assume that the obtained integral over the deformed configuration can be evaluated simply over the reference configuration itself.

Using the constitutive relation for linearized elasticity
$$
\bm{\tau_u} = \lambda \ \text{tr}(\bm{\epsilon_u})\mathbf{I} + 2\mu\bm{\epsilon_u}
$$
we obtain
$$
\int_\Omega{\{\lambda \ \text{tr}(\bm{\epsilon_u})\mathbf{I} + 2\mu \bm{\epsilon_u}\}: \bm{\epsilon_v} \ d\Omega} = \int_\Omega{\mathbf{f \cdot v} \ d\Omega}
$$

Using $\mathbf{I}:\bm{\epsilon_v} = \text{tr}(\bm{\epsilon_v})$, we can write it in the familiar abstract formulation as
$$
a(\mathbf{u},\mathbf{v}) = L(\mathbf{v})
$$
where
$$
a(\mathbf{u},\mathbf{v}) = \int_\Omega{\lambda \ \text{tr}(\bm{\epsilon_u}) \text{tr}(\bm{\epsilon_v})  + 2\mu \ (\bm{\epsilon_u}:\bm{\epsilon_v}) \ d\Omega} \ , \qquad L(\mathbf{v}) = \int_\Omega{\mathbf{f \cdot v} \ d\Omega}
$$

We know that such a formulation has a unique solution, from the Riesz representation theorem and energy minimization arguments made for the Poisson problem. The requirements for $a(. \ , .)$ required for the arguments to hold are examined now.

## Linearized Rigid Motion
We will first show that a linearized rigid motion is characterized by:
$$
\bm{\epsilon_u} = 0 \iff \mathbf{u} = \mathbf{Wx + c} \ , \quad \mathbf{W} \in \text{Skw}, \  \mathbf{c} \in \text{const}
$$

Let $\mathbf{u} = \mathbf{Wx + c}$. Then $\bm{\epsilon_u} = \frac{1}{2} \left[ \left( \nabla \mathbf{u}\right) + \left( \nabla \mathbf{u}\right)^T \right] = 0$ trivially.

Now for the converse: Let $\mathbf{W(u)} = \frac{1}{2} \left[ \left( \nabla \mathbf{u}\right) - \left( \nabla \mathbf{u}\right)^T \right] \ \in \text{Skw}$. We then have
$$
\nabla \mathbf{u} = \bm{\epsilon_u} + \mathbf{W}  \quad \overset{\bm{\epsilon_u} = 0}{=} \quad \mathbf{W(u)} \ \in \text{Skw} \quad \Rightarrow \quad \nabla_j u_i = - \nabla_i u_j
$$
$$
\therefore \nabla_k \nabla_j u_i = - \nabla_k \nabla_i u_j = - \nabla_i \nabla_k u_j = \nabla_i \nabla_j u_k = \nabla_j \nabla_i u_k = -\nabla_j \nabla_k u_i = -\nabla_k \nabla_j u_i 
$$

$$
\nabla_k \nabla_j u_i = -\nabla_k \nabla_j u_i \quad \Rightarrow \quad \nabla_k \nabla_j u_i = 0
$$

This implies that $\mathbf{u}$ is affine, and $\nabla \mathbf{u} = \mathbf{W}$ is a *constant* skew-symmetric tensor, and is termed an infinitesmal rotation. Thus we can write any linearized rigid displacement field as 
$$\mathbf{u} = \mathbf{Wx + c}$$



## V-Ellipticity of $a(. \ , \ .)$
The inequality
$$
a(\mathbf{u},\mathbf{u}) = \int_\Omega{\lambda \ \text{tr}(\bm{\epsilon_u})^2  + 2\mu \ (\bm{\epsilon_u}:\bm{\epsilon_u}) \ d\Omega} \quad \ge 0
$$
always holds, and thus $a(. \ , \ .)$ is positive-semidefinite. We now examine the positive-definiteness (V-ellipticity) of a.

$\mathbf{u} = 0:$

$$
\therefore \bm{\epsilon_u} = 0  \ (\text{linearized rigid displacement}) \quad \Rightarrow \quad a(\mathbf{u},\mathbf{u}) = 0
$$

$a(\mathbf{u},\mathbf{u}) = 0$:
$$
\int_\Omega{\lambda \ \text{tr}(\bm{\epsilon_u})^2  + 2\mu \ (\bm{\epsilon_u}:\bm{\epsilon_u}) \ d\Omega} = 0
$$
The integrand is always $\ge 0$ for all $\mathbf{u}$, and the integral $= 0$; thus the integrand must be zero at every point in the domain.
$$
\lambda \ \text{tr}(\bm{\epsilon_u})^2  + 2\mu \ (\bm{\epsilon_u}:\bm{\epsilon_u}) = 0
$$
Since both terms are non-negative, they must individually go to zero. Thus it is necessary and sufficient that:
$$
\bm{\epsilon_u} = 0 
$$
Thus 
$$
a(. \ , \ .) = 0 \quad \Rightarrow \quad \text{the motion is a linearized displacement field}
$$



We have thus shown that the bilinear $a(. \ , \ .)$ for linearized elasticity is positive-definite *modulo* linearized rigid displacements.

This ensures the existence and uniqueness of the solution to the variational formulation for all other displacement fields, using the same arguments made for the variational formulation of the Poisson PDE, which was of the same form.

## The Energy Functional
We define the potential energy functional as 
$$
J(\mathbf{u}) = \frac{1}{2}\int_\Omega{\bm{\tau_u : \epsilon_u} \ d\Omega} - \int_\Omega{\mathbf{f \cdot u} \ d\Omega}
$$
For linearized elasticity, using the above bilinear and linear forms, we obtain
$$
J(\mathbf{u}) = \frac{1}{2}a(\mathbf{u}, \mathbf{u}) - L(\mathbf{u})
$$


## Solution Existence and Uniqueness

Consider two functionals $a(. \ , .)$ and $L(.)$

 - $a(. \ , .)$ is bilinear
 - $a(. \ , .)$ is continuous
 - $a(. \ , .)$ is symmetric
 - $a(. \ , .)$ is positive-definite (or V-elliptic)
 - $L(.)$ is linear
 - $L(.)$ is continuous

Consider the two problems where $a(. \ , .)$ and $L(.)$ have the properties defined earlier.

### The Variational Problem (V):

Find $\mathbf{u}$ such that

$$
a(\mathbf{u},\mathbf{v})=L(\mathbf{v})
\qquad
\forall \mathbf{v} \in V
$$


### The Minimization Problem (M):

Find $\mathbf{u}$ such that

$$
J(\mathbf{u})=\min_{\mathbf{v}\in V}J(\mathbf{v})
$$

where

$$
J(\mathbf{v})=\frac{1}{2}a(\mathbf{v},\mathbf{v})-L(\mathbf{v}).
$$


We have shown that the two problems are equivalent, and that their solution (if it exists) is unique.


We now wish to show the equivalence between Problems (V) and (M) for the generalized elasticity problem, where $a(. \ , \ .)$ may not be symmetric.

Assume (M), i.e.,

$$
\exists\,\mathbf{u} \quad\text{s.t.}\quad J(\mathbf{u})=\min_{\mathbf{v}\in V}J(\mathbf{v}).
$$

Thus $\forall  \ \mathbf{v}\in V,\quad \varepsilon\in\mathbb{R}$ we have,
$$
J(\mathbf{u}+\varepsilon \mathbf{v})\geq J(\mathbf{u})
$$
since $\mathbf{u}$ minimises $J$.

Let
$$
g(\varepsilon)=J(\mathbf{u}+\varepsilon \mathbf{v})
\quad \Rightarrow \quad g(\varepsilon)\geq g(0) \qquad \forall\varepsilon\in\mathbb{R}
$$

Thus $g$ has a minimum at $\varepsilon=0$, so $g'(0)=0$.

$$
g(\varepsilon) = \frac{1}{2}a(\mathbf{u}+\varepsilon \mathbf{v},\mathbf{u}+\varepsilon \mathbf{v})
- L(\mathbf{u}+ \varepsilon \mathbf{v}) = \frac{1}{2} \left[ a(\mathbf{u},\mathbf{u})+ \varepsilon a(\mathbf{u},\mathbf{v})
+ \varepsilon a(\mathbf{v},\mathbf{u}) +\varepsilon^2a(\mathbf{v},\mathbf{v}) \right] - L(\mathbf{u})-\varepsilon L(\mathbf{v}).
$$

$$
\Rightarrow \quad g'(\varepsilon) =  \frac{1}{2} \left[ a(\mathbf{u}, \mathbf{v}) + a(\mathbf{v}, \mathbf{u}) \right] - L(\mathbf{v}) + \varepsilon a(\mathbf{v},\mathbf{v}) 
$$

$$
\therefore\quad g'(0)=0 \quad \Rightarrow \quad \frac{1}{2} \left[ a(\mathbf{u}, \mathbf{v}) + a(\mathbf{v}, \mathbf{u}) \right] = L(\mathbf{v})
$$



 Starting from (M), we see that
$$
J(\mathbf{u}) = \min_{\mathbf{v} \in V} J(\mathbf{v}) \quad \Rightarrow \quad \frac{1}{2} \left[ a(\mathbf{u}, \mathbf{v}) + a(\mathbf{v}, \mathbf{u}) \right] = L(\mathbf{v})
$$

Thus the symmetry of $a(\mathbf{u}, \mathbf{v})$ is a necessary condition for the existence of an energy functional $J(\mathbf{u})$, such that its minimizer (field that sets the directional derivative of $J = 0$) solves the variational formulation. 

We now show the uniqueness of the solution using the convexity of the $J(\mathbf{u}) = \frac{1}{2}a(\mathbf{u}, \mathbf{u}) - L(\mathbf{u})$ functional.

$$
\begin{align*}

& \lambda J(\mathbf{u_1}) + (1-\lambda)J(\mathbf{u_2}) - J(\lambda \mathbf{u_1} + (1-\lambda)\mathbf{u_2}) \\

&= \frac{1}{2} \left[
\lambda a(\mathbf{u_1},\mathbf{u_1}) + (1-\lambda)a(\mathbf{u_2},\mathbf{u_2})
- a(\lambda \mathbf{u_1} + (1-\lambda)\mathbf{u_2}, \lambda \mathbf{u_1} + (1-\lambda)\mathbf{u_2})
\right] 
+ \lambda L(\mathbf{u_1}) + (1-\lambda)L(\mathbf{u_2}) - L(\lambda \mathbf{u_1} + (1-\lambda)\mathbf{u_2}) \\[1em]

&= \frac{1}{2} \left[
\lambda a(\mathbf{u_1},\mathbf{u_1}) + (1-\lambda)a(\mathbf{u_2},\mathbf{u_2})
- \lambda^2 a(\mathbf{u_1},\mathbf{u_1}) - (1-\lambda)^2 a(\mathbf{u_2},\mathbf{u_2}) 
- \lambda(1-\lambda) \left\{ a(\mathbf{u_1},\mathbf{u_2}) + a(\mathbf{u_2},\mathbf{u_1}) \right\}
\right] \\[1em]

&= \frac{1}{2}\lambda(1-\lambda) \left[
a(\mathbf{u_1},\mathbf{u_1}) + a(\mathbf{u_2},\mathbf{u_2}) - a(\mathbf{u_1},\mathbf{u_2}) - a(\mathbf{u_2},\mathbf{u_1})
\right] \\[1em]

&= \frac{1}{2}\lambda(1-\lambda) \ a(\mathbf{u_1}-\mathbf{u_2},\mathbf{u_1}-\mathbf{u_2})

\end{align*}
$$

Using the V-ellipticity of $a(.,.)$, for $\lambda \in (0, 1)$ we thus have
$$ \lambda J(\mathbf{u_1}) + (1-\lambda)J(\mathbf{u_2}) - J(\lambda \mathbf{u_1} + (1-\lambda)\mathbf{u_2}) \ge 0 \quad \forall \ \mathbf{u_1, u_2} \in V, \ \text{and} \quad = 0 \iff \mathbf{u_1 = u_2 + Wx + c}$$

Hence $J$ is convex, and has a unique minimizer (*modulo* linearized rigid motion). *However*, this does not imply that this minimizer solves the variational formulation in general. We have seen that that is only true if $a(.,.)$ is symmetric.


We have thus seen that the variational formulation can be developed for any elasticity problem, but the existence of an energy functional whose minimizer solves the weak form necessitates the symmetry of the $a(. \ , .)$ operator, which pops out only once the small-strain and linearized elastic constitutive relations are invoked, and are not true in a general elasticity problem.

The existence and uniqueness of the solution to Problem (V) can still be shown when  $a(. \ , .)$ is non-symmetric, there is just no equivalent Problem (M).
