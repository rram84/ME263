## 1. Problem set-up and kinematics

An annulus (plane strain) with reference radii $A<R<B$ is deformed radially symmetrically to $a<r<b$, $r=r(R)$, $\theta=\Theta$.

Stretches:

$$
\lambda_r = r' = \frac{dr}{dR},\qquad
\lambda_\theta = \frac{r}{R},\qquad
\lambda_a=\frac{a}{A},\qquad \lambda_b=\frac{b}{B}.
$$

Deformation gradient (polar basis, $\mathbf e_r\otimes\mathbf e_R$ etc.):

$$
\mathbf F=\lambda_r\,\mathbf e_r\otimes\mathbf e_R+\lambda_\theta\,\mathbf e_\theta\otimes\mathbf e_\Theta,
\qquad J=\det\mathbf F=\lambda_r\lambda_\theta .
$$

The deformed radius in terms of $R$, $r(R)$, is the unknown.

## 2. Constitutive law and equilibrium

**Compressible neo-Hookean** (plane strain, 2D invariants):

$$
\psi=\frac{\mu}{2}\left(\operatorname{tr}\mathbf C-2-2\ln J\right)+\frac{\lambda}{2}(\ln J)^2
\quad\Rightarrow\quad
\mathbf P=\mu\left(\mathbf F-\mathbf F^{-T}\right)+\lambda\ln J\,\mathbf F^{-T}.
$$

Writing $\mathbf P=P_r\,\mathbf e_r\otimes\mathbf e_R+P_\theta\,\mathbf e_\theta\otimes\mathbf e_\Theta$:

$$
P_r=\mu\left(\lambda_r-\frac1{\lambda_r}\right)+\frac{\lambda}{\lambda_r}\ln J,\qquad
P_\theta=\mu\left(\lambda_\theta-\frac1{\lambda_\theta}\right)+\frac{\lambda}{\lambda_\theta}\ln J .
$$

### Equilibrium: $\operatorname{Div}\mathbf P=\mathbf 0$ $\Rightarrow$ $\dfrac{d}{dR}(R\,P_r)=P_\theta$

*How to interpret this?* Take a reference wedge between $R$ and $R+dR$ with opening angle $d\Theta$ and do a force balance on the **deformed** wedge:

* Traction on the inner face (normal $-\mathbf e_R$): $-P_r(R)\,\mathbf e_r\,R\,d\Theta$ (the nominal traction is force per *reference* area).
* Traction on the outer face: $+P_r(R+dR)\,\mathbf e_r\,(R+dR)\,d\Theta$.
* On the two side faces (at $\Theta\pm d\Theta/2$) the traction is $\mp P_\theta\,\mathbf e_\theta(\Theta\pm d\Theta/2)\,dR$.
  Since $\mathbf e_\theta\otimes\mathbf e_\Theta$ acting on the face normal $\mathbf e_\Theta$ gives $\mathbf e_\theta$, and $\mathbf e_R\otimes\mathbf e_r\cdot\mathbf e_\theta=0$, **only the radial component** survives after adding the two faces:

$$
2P_\theta\,dR\,\sin\!\left(\frac{d\Theta}{2}\right)\approx P_\theta\,dR\,d\Theta .
$$

Summing the radial components:

$$
-P_r(R)\,R\,d\Theta+P_r(R+dR)(R+dR)\,d\Theta-P_\theta\,dR\,d\Theta=0 .
$$

Expanding and dropping $(dR)^2$ terms: $\;(R P_r)'\,dR\,d\Theta=P_\theta\,dR\,d\Theta$, i.e.

$$
\boxed{\frac{d}{dR}\left(R\,P_r\right)=P_\theta}\qquad\text{(ODE proved).}
$$

Since $P_r=P_r(r',r,R)$ and $P_\theta=P_\theta(r',r,R)$ this is a **second-order ODE for $r(R)$**.

## 3. Boundary conditions

* Outer surface free: $P_r(B)=0$.
* Inner surface loaded by (current) pressure $p$ along the **inner radius**. The traction is $\mathbf t=-p\,\mathbf n$ in the deformed configuration; the nominal traction is $-p\,J\mathbf F^{-T}\mathbf N$. Radially, $J\mathbf F^{-T}\mathbf e_R\cdot\mathbf e_r = J/\lambda_r=\lambda_\theta$, therefore

$$
P_r(A)=-p\,\lambda_\theta\big|_{R=A}=-p\,\lambda_a .
$$

Solving the 2nd-order ODE with these two BCs gives $r(R)$.

## 4. The incompressible limit, $\nu\to\tfrac12$

For the compressible law
$\mathbf P=\mu\mathbf F-\mu\mathbf F^{-T}+\lambda\ln J\,\mathbf F^{-T}=\mu\mathbf F-q\,\mathbf F^{-T}$ with

$$
q=\mu-\lambda\ln J .
$$

As $\nu\to\tfrac12$:

$$
\frac{\lambda}{\mu}=\frac{2\nu}{1-2\nu}\to\infty,\qquad \frac{K_{\rm bulk}}{G_{\rm shear}}\to\infty .
$$

The stress stays finite, so $q$ stays finite, hence
$\ln J=\dfrac{\mu-q}{\lambda}=\dfrac{\mu-q}{\mu}\cdot\dfrac{\mu}{\lambda}\to0$, i.e. $J=1$: **the material is incompressible**.

Important conceptual point: $J=1$ is a **constraint**, not a material response. It gives one new equation but also one new unknown, $q$ — a **Lagrange multiplier** (a pressure-like field that changes with position). The stress–strain law becomes

$$
\mathbf P=\mu\mathbf F-q\,\mathbf F^{-T},\qquad
P_r=\mu\lambda_r-\frac q{\lambda_r},\quad P_\theta=\mu\lambda_\theta-\frac q{\lambda_\theta},\quad J=\lambda_r\lambda_\theta=1 .
$$

The equilibrium $\frac{d}{dR}(RP_r)=P_\theta$ does not change in form, but now includes the contribution from $q$.
Unknowns: $r(R)$ and $q(R)$; equations: $J=1$, equilibrium. BCs: $P_r(B)=0$, $P_r(A)=-p\lambda_a$.

## 5. Closed-form solution of the incompressible cylinder

### 5.1 Kinematics from $J=1$

$$
\frac{r'r}{R}=1\;\Rightarrow\; r\,r'=R\;\Rightarrow\;\frac{d}{dR}\left(\frac{r^2}{2}\right)=\frac{d}{dR}\left(\frac{R^2}{2}\right)
\;\Rightarrow\; \boxed{r^2=R^2+C}.
$$

Hence $r(R)=\sqrt{R^2+C}$ — a **closed-form** solution; we only need the constant $C$.
Physically $\pi(r^2-R^2)=\pi C$: the area of material enclosed is unchanged, $\pi(a^2-A^2)=\pi(b^2-B^2)=\pi C$.

### 5.2 Finding $q$

With $r'=R/r$, i.e. $\lambda_r=1/\lambda_\theta$:

$$
R P_r=\mu\frac{R^2}{r}-q\,r,\qquad P_\theta=\mu\frac rR-q\frac Rr .
$$

Differentiating $R P_r$ and equating to $P_\theta$:

$$
\mu\left(\frac{2R}{r}-\frac{R^3}{r^3}\right)-q'r-q\frac Rr=\mu\frac rR-q\frac Rr
\;\Rightarrow\;
q'=\frac{\mu}{r}\left(\frac{2R}{r}-\frac{R^3}{r^3}\right)-\frac{\mu}{R}.
$$

With $\lambda_\theta=r/R$:

$$
\boxed{\frac{dq}{dR}=-\frac{\mu}{R}\left(1-\frac{1}{\lambda_\theta^2}\right)^2}
$$

### 5.3 Change of variable to $\lambda_\theta$

$$
\frac{d\lambda_\theta}{dR}=\frac{r'}{R}-\frac r{R^2}=\frac1R\left(\frac1{\lambda_\theta}-\lambda_\theta\right)=\frac{1-\lambda_\theta^2}{R\lambda_\theta}
$$

(chain rule, using $r'=1/\lambda_\theta$). Therefore $q$ is **only a function of $\lambda_\theta$**:

$$
\frac{dq}{d\lambda_\theta}=\frac{dq/dR}{d\lambda_\theta/dR}
=\frac{\mu(\lambda_\theta^2-1)}{\lambda_\theta^3}
=\mu\left(\frac1{\lambda_\theta}-\frac1{\lambda_\theta^3}\right)
\;\Rightarrow\;
q(\lambda_\theta)=\mu\left(\ln\lambda_\theta+\frac1{2\lambda_\theta^2}\right)+q_0 .
$$

### 5.4 Constant from the outer BC

At $R=B$: $\lambda_\theta=\lambda_b$, $\lambda_r=1/\lambda_b$, and $P_r(B)=0$:

$$
\frac{\mu}{\lambda_b}-q(B)\lambda_b=0\;\Rightarrow\;q(B)=\frac{\mu}{\lambda_b^2}
\;\Rightarrow\; q_0=\frac{\mu}{2\lambda_b^2}-\mu\ln\lambda_b .
$$

$$
\boxed{q(\lambda_\theta)=\mu\ln\frac{\lambda_\theta}{\lambda_b}+\frac{\mu}{2}\left(\frac1{\lambda_\theta^2}-\frac1{\lambda_b^2}\right)}
$$


### 5.5 Pressure–stretch relation

Inner BC, $P_r(A)=-p\lambda_a$ with $\lambda_r=1/\lambda_a$: $\;\dfrac{\mu}{\lambda_a}-q(\lambda_a)\lambda_a=-p\lambda_a\Rightarrow p=q(\lambda_a)-\dfrac{\mu}{\lambda_a^2}$:

$$
\boxed{p(\lambda_a)=\frac{\mu}{2}\left(\ln\frac{\lambda_a^2}{\lambda_b^2}+\frac1{\lambda_b^2}-\frac1{\lambda_a^2}\right)}
$$

and since $r^2-R^2=C$ everywhere,

$$
C=(\lambda_a^2-1)A^2,\qquad \lambda_b^2=1+\frac{C}{B^2}=1+(\lambda_a^2-1)\frac{A^2}{B^2}.
$$

So $p(\lambda_a)$ is explicit; to find $\lambda_a$ for a given $p$ solve one scalar equation. Evaluate area relation $\pi(b^2-a^2)=\pi(B^2-A^2)$ as a check.

## 6. Compressible reference solution (ODE BVP)

For a general $\nu$ we need $r(R)$ from the 2nd-order ODE $R\,P_r'+P_r=P_\theta$. With $\lambda_r=r'$, $J=r r'/R$:

$$
\frac{\partial P_r}{\partial r'}=\mu\left(1+\frac1{r'^2}\right)+\frac{\lambda(1-\ln J)}{r'^2},\qquad
\frac{\partial P_r}{\partial r}=\frac{\lambda}{r\,r'},\qquad
\frac{\partial P_r}{\partial R}\bigg|_{r,r'}=-\frac{\lambda}{R\,r'}\quad(\text{since }J=rr'/R),
$$

$$
r''=\frac{P_\theta-P_r-R\left(\dfrac{\partial P_r}{\partial r}\,r'+\dfrac{\partial P_r}{\partial R}\right)}{R\,\dfrac{\partial P_r}{\partial r'}}
$$

with BCs $P_r(B)=0$, $P_r(A)+p\,r(A)/A=0$. 

## 7. Why does something go wrong as $\nu\to\tfrac12$?

### 7.1 Céa's lemma and the constant $M/\alpha$

For the linearised problem $a(\mathbf u,\mathbf v)=\int 2\mu\,\boldsymbol\varepsilon(\mathbf u):\boldsymbol\varepsilon(\mathbf v)+\lambda(\nabla\!\cdot\mathbf u)(\nabla\!\cdot\mathbf v)$,

$$
\|\mathbf u-\mathbf u_h\|\;\le\;\frac{M}{\alpha}\,\inf_{\mathbf v_h\in V_h}\|\mathbf u-\mathbf v_h\| ,
$$

where $\alpha$ is the coercivity constant and $M$ the continuity constant, and $M\sim\lambda\to\infty$ as $\nu\to\tfrac12$ (while $\alpha\sim\mu$ stays finite). The **upper bound is very high** — the error bound is not useful. The solution is poor in the energy norm, and *not that accurate in the $L^2$ norm either* (the Aubin–Nitsche argument does not rescue it when $M$ is huge).

### 7.2 What the bilinear form wants

Dividing by $\mu$: $\dfrac1\mu a(\mathbf u,\mathbf v)=\int 2\boldsymbol\varepsilon(\mathbf u):\boldsymbol\varepsilon(\mathbf v)+\dfrac{\lambda}{\mu}(\nabla\!\cdot\mathbf u)(\nabla\!\cdot\mathbf v)$. For $\dfrac1\mu a(\mathbf u,\mathbf u)$ to remain bounded as $\lambda/\mu\to\infty$ we need $\nabla\!\cdot\mathbf u\to0$.
In the discrete space this forces the discrete displacement to be (almost) **discretely divergence-free** — and the P1 space has too few such functions.

### 7.3 DOF counting with the Euler relation

* Incompressibility means the **area of each triangle must be conserved** $\Rightarrow$ $T$ scalar constraints (one per triangle, $T$ = number of triangles).
* Euler relation for a planar triangulation: $V-E+T=1$ (vertices − edges + faces; the outer face is not counted).
* Counting edges: $3T=2E-V_b$ (each interior edge is shared by two triangles, boundary edges by one), and the number of boundary edges equals the number of boundary vertices $V_b$, so $E=\dfrac{V_b}{2}+\dfrac{3}{2}T$. Substituting:

$$
V-\frac{V_b}{2}-\frac{T}{2}=1\;\Rightarrow\;\boxed{2V-V_b=T+2}.
$$

* Total displacement DOF (2 per vertex) minus Dirichlet-constrained components $V_b(x),V_b(y)$, minus the $T$ incompressibility constraints:

$$
\text{dof}=2V-V_b(x)-V_b(y)-T=V_b+2-V_b(x)-V_b(y).
$$

So the number of unconstrained DOFs left is **only of the order of the number of boundary vertices**: the only motions that preserve all areas are essentially **boundary motions** — the interior of the mesh is locked. For the coarse mesh in class this gave **48 DOF**, for the finer mesh **95 DOF** (compare the output of the cell below for your mesh).

This is why in the P1 displacement method the mesh locks(only the boundary basically moves — locking is caused by this).

### 7.4 Inf-sup condition

For the mixed method to be stable one needs the discrete inf-sup (LBB) condition and coercivity on the kernel $\{\nabla\!\cdot\mathbf u_h=0\}$:

$$
a(\mathbf u,\mathbf u)\ge\alpha\|\mathbf u\|^2 ,\qquad
\inf_{q_h}\sup_{\mathbf v_h}\frac{\int q_h\,\nabla\!\cdot\mathbf v_h}{\|q_h\|\,\|\mathbf v_h\|}\ge\beta>0 .
$$

P2–P1 (Taylor–Hood) satisfies it, so $\|\mathbf u-\mathbf u_h\|$ is independent of $\lambda/\mu$ — this is the mixed solution computed above.

### 7.5 Numerical artefact when $\nu\to\tfrac12$

In the quarter annulus, as $\nu\to\tfrac12$ the locked P1 solution shows $\lambda_\theta\to0$ near the symmetry boundaries — a **numerical artefact** of locking, not a physical effect. The exact solution stays smooth, as the closed form above shows.

## Summary

| | Compressible ($\nu<\tfrac12$) | Incompressible ($\nu=\tfrac12$) |
|---|---|---|
| Stress | $\mathbf P=\mu(\mathbf F-\mathbf F^{-T})+\lambda\ln J\,\mathbf F^{-T}$ | $\mathbf P=\mu\mathbf F-q\,\mathbf F^{-T}$, $J=1$ |
| Unknowns | $r(R)$ | $r(R)$, $q(R)$  |
| Solution | 2nd-order ODE (BVP) | $r=\sqrt{R^2+C}$, $q(\lambda_\theta)$ closed form |
| FE | displacement P1/P2: **locks** as $\nu\to\tfrac12$ | mixed P2–P1 (inf-sup stable) |
| Error bound | $\propto M/\alpha\sim\lambda/\mu$ | independent of $\lambda/\mu$ |
