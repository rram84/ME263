# Newton–Raphson Method for Nonlinear Problems

- Hat, $\hat{(\cdot)}$: nodal coefficient (DOF) vector of a function in $V_h$,  e.g. $\hat{\mathbf u}$.
- Subscript $h$: quantity associated with the finite element space $V_h$;  
  - A subscript-$h$ field such as $u_h$ is still a function defined at every point of $\Omega$. It is "discrete" only in the sense that it belongs to the finite-dimensional space $V_h$ and is fixed by finitely many numbers, $\hat{\mathbf u}$.
  - $\Pi_h$ is a function of $\hat{\mathbf u}$, not of position.
- $D\Pi(u)[v]$: directional derivative of $\Pi$ at $u$ in the direction $v$; square brackets enclose the direction.
- A semicolon separates the state from the arguments in which a form is linear: $R(u; \, v)$, $K(u; \, v, \, w)$.


## 1. Newton-Raphson Loop


### 1.1 Scalar case

Find $\xi$ such that $f(\xi) = 0$.\
Start from a guess $x_0$ with $f(x_0) \neq 0$ and seek a correction $\Delta x$ such that $x_1 = x_0 + \Delta x$:
$$
f(x_1) = f(x_0 + \Delta x) = f(x_0) + f'(x_0)\,\Delta x + \mathcal O(\Delta x^2)
$$
Set $f(x_1) = 0$ and drop the $\mathcal O(\Delta x^2)$ term:
$$
\Delta x = -\frac{f(x_0)}{f'(x_0)} \quad\Longrightarrow\quad x_1 = x_0 - \frac{f(x_0)}{f'(x_0)}
$$

Repeating this gives the iteration
$$
x_{k+1} = x_k - \frac{f(x_k)}{f'(x_k)}, \qquad x_k \to \xi \ \text{ as } k \to \infty .
$$

**Stopping criteria** (tolerance $\epsilon$) : $|f(x_k)| < \epsilon $

> **Convergence.** Suppose $f'(\xi) \neq 0$, $f'$ is Lipschitz continuous near $\xi$, and $x_0$ is sufficiently close to $\xi$. Then the iterates converge **quadratically**: $|x_{k+1} - \xi| \le C\,|x_k - \xi|^2$. Convergence from an arbitrary $x_0$ is not guaranteed.

### 1.2 Vector case

Let $\mathbf f : \mathbb R^n \to \mathbb R^n$. Find $\bm\xi$ such that $\mathbf f(\bm \xi) = \mathbf 0$. \
Start from a guess $\mathbf x_0$ with $\mathbf f(\mathbf x_0) \neq \mathbf 0$:
$$
\mathbf f(\mathbf x_0 + \Delta\mathbf x) = \mathbf f(\mathbf x_0) + \underbrace{\big(\nabla \mathbf f(\mathbf x_0)\big)}_{\mathbf J(\mathbf x_0)}\,\Delta\mathbf x + O(\|\Delta\mathbf x\|^2), \qquad J_{ij} = \frac{\partial f_i}{\partial x_j}
$$

Set the left-hand side to $\mathbf 0$ and drop the higher-order term. For the $k^{th}$ iteration, we have
$$
\Delta\mathbf x = -\mathbf J^{-1}(\mathbf x_k)\,\mathbf f(\mathbf x_k)
\quad\Longrightarrow\quad
\mathbf x_{k+1} = \mathbf x_k + \Delta\mathbf x
$$
$\mathbf J^{-1}$ is never formed explicitly. At each iteration, the linear system $\mathbf J(\mathbf x_k)\,\Delta\mathbf x = -\mathbf f(\mathbf x_k)$ is solved, and then $\mathbf x_{k+1}$ is updated. The stopping criteria uses norms: $\|\mathbf f(\mathbf x_k)\| < \epsilon$.

## 2. Variational problem and discrete equilibrium

### 2.1 Continuous Problem

Find $u$ such that $\Pi(u)$ is stationary:
$$
R(u;v) = D\Pi(u)[v] = 0 \qquad \forall\, v,
$$
where
$$
D\Pi(u)[v] = \lim_{\varepsilon\to 0}\frac{\Pi(u+\varepsilon v)-\Pi(u)}{\varepsilon} = \left.\frac{\mathrm d}{\mathrm d\varepsilon}\,\Pi(u+\varepsilon v)\right|_{\varepsilon=0}.
$$
$R(u;v)$ is linear in the perturbation direction $v$ but, in general, nonlinear in $u$.

### 2.2 Discretised Problem

The field is approximated as
$$
u_h = \phi_a \, \hat u_a = \boldsymbol\phi \cdot \hat{\mathbf u}, \qquad  \boldsymbol\phi = [\phi_1,\,\phi_2,\,\dots,\,\phi_n]^{\mathsf T} \qquad
\hat{\mathbf u} = [\hat u_1,\,\hat u_2,\,\dots,\,\hat u_n]^{\mathsf T}.
$$
The discrete problem is then: find $\hat{\mathbf u}$ such that $\Pi_h(\hat{\mathbf u}) \equiv \Pi(u_h) = \Pi(\hat u_a \phi_a)$ is stationary.

Express the test function in the same basis, $v_h = \hat v_i\,\phi_i$. Because $R$ is linear in its second argument,
$$
R(u_h;v_h) = \hat v_i\,R(u_h; \, \phi_i) = \hat v_i\,r_i(\hat{\mathbf u}) = \hat{\mathbf v} \cdot \mathbf r(\hat{\mathbf u}),
\qquad r_i(\hat{\mathbf u}) \equiv R(u_h; \, \phi_i).
$$

The arbitrariness of $\hat{\mathbf v}$ thus implies that
$$
R(u_h; \, v_h) = 0 \ \ \forall\, v_h \in V_h   \quad\Longleftrightarrow\quad
r_i(\hat{\mathbf u}) = 0 \ \ \forall\, i   \quad\Longleftrightarrow\quad    \mathbf r(\hat{\mathbf u}) = \mathbf 0 .
$$
This is a **vectorised root-finding** problem.

**Link to the FE energy.** Perturbing $\hat u_i$ by $\varepsilon$ perturbs $u_h$ by $\varepsilon\phi_i$: 
$$
\hat{\mathbf u} = \begin{bmatrix} \hat u_1 & \cdots & \hat u_i & \cdots & \hat u_n \end{bmatrix}^{\mathsf T}
\ \xrightarrow{\ \text{perturb}\ }\
\begin{bmatrix} \hat u_1 & \cdots & \hat u_i + \varepsilon & \cdots & \hat u_n \end{bmatrix}^{\mathsf T},
$$
Therefore
$$
\frac{\partial\Pi_h}{\partial\hat u_i}  =  \lim_{\varepsilon\to 0}\frac{\Pi(u_h+\varepsilon\phi_i)-\Pi(u_h)}{\varepsilon} =  D\Pi(u_h)[\phi_i] = R(u_h; \, \phi_i) = r_i(\hat{\mathbf u}).
$$
Hence $\mathbf r = \nabla_{\hat{\mathbf u}}\Pi_h$, and $\Pi_h$ is stationary if and only if $\mathbf r(\hat{\mathbf u}) = \mathbf 0$.

## 3. Obtaining the Discretised System

### 3.1 Discretise $ \ \rightarrow \ $ Linearise

Taking the discretised system, we expand the discrete residual about a guess $\hat{\mathbf u}_0$ with $\mathbf r(\hat{\mathbf u}_0) \neq \mathbf 0$:
$$
\mathbf r(\hat{\mathbf u}_0 + \Delta\hat{\mathbf u}) = \mathbf r(\hat{\mathbf u}_0) + \underbrace{\nabla_{\hat{\mathbf u}}\mathbf r(\hat{\mathbf u}_0)}_{\mathbf K}\,\Delta\hat{\mathbf u} + O(\Vert\Delta\hat{\mathbf u}\Vert^2)
$$
where the tangent stiffness matrix is
$$
\mathbf K(\hat{\mathbf u}_0) = \nabla\mathbf r(\hat{\mathbf u}_0) = \left. \frac{\partial \mathbf r}{\partial \hat{\mathbf u}} \right|_{\hat{\mathbf u} = \hat{\mathbf u}_0}, \qquad
K_{ij} = \frac{\partial r_i}{\partial \hat u_j} = \frac{\partial^2 \Pi_h}{\partial\hat u_i \ \partial\hat u_j} \qquad \text{(the 'algorithmic' or 'consistent' tangent)}
$$
Thus $\mathbf K$ is the Hessian of the discrete energy and is symmetric.

Setting $\mathbf r(\hat{\mathbf u}_0 + \Delta\hat{\mathbf u}) = \mathbf 0$ gives the linear system for the iterative correction:
$$
\mathbf K(\hat{\mathbf u}_0)\,\Delta\hat{\mathbf u} = -\mathbf r(\hat{\mathbf u}_0)
$$
$\mathbf K$ depends on the current iterate, and so is updated at every iteration.

### 3.2 Linearise $ \ \rightarrow \ $ Discretise

At the current guess $u_0$, in general $R(u_0;v) \neq 0$. We seek $\Delta u$ such that $R(u_0 + \Delta u;\,v) = 0$ for all $v$. For all $v$:
$$
R(u_0 + \Delta u; \, v) = R(u_0;v) + \underbrace{DR(u_0; \, v)[\Delta u]}_{\substack{\text{linearisation of } R \text{ at } u_0 \text{ along } \Delta u \\ \text{(a new function, linear in } \Delta u\text{)}}} + \mathcal  O(\Vert\Delta u\Vert^2)
$$

The linearisation defines the continuum tangent form, the stiffness analogue:
$$
K(u_0; \, v, \, \Delta u) \equiv DR(u_0; \, v)[\Delta u] = D^2\Pi(u_0)[v,\Delta u]
$$

Setting $R(u_0+\Delta u; \, v) = 0$ and dropping the higher-order term:
$$
K(u_0; \, v, \, \Delta u) = -R(u_0;\,v) \qquad \forall\, v
$$
This is a linear problem in $\Delta u$, to be solved at each iteration.

**Now discretise:** \
Choose $u_0 \to u_h = \hat u_a\phi_a$, $v = \phi_i$ and $\Delta u \to \Delta u_h = \Delta\hat u_j\,\phi_j$. By bilinearity of $K$ in its second and third arguments:
$$
K(u_h; \, \phi_i, \, \phi_j)\,\Delta\hat u_j = -R(u_h;\phi_i) = -r_i(\hat{\mathbf u})
$$

Perturbing the $\hat u_j$ changes the field from $u_h$ to $u_h + \varepsilon\,\phi_j$. Therefore
$$
K(u_h; \, \phi_i, \, \phi_j) = DR(u_h; \, \phi_i)[\phi_j]  = \lim_{\varepsilon\to 0}\frac{R(u_h + \varepsilon\phi_j; \, \phi_i) - R(u_h ; \, \phi_i)}{\varepsilon}
= \frac{\partial r_i(\hat{\mathbf u})}{\partial \hat u_j} = K_{ij} 
$$
gives the same $K_{ij}$ as the earlier route. Thus we obtain
$$
\mathbf K(\hat{\mathbf u}_0)\,\Delta\hat{\mathbf u} = -\mathbf r(\hat{\mathbf u}_0)
$$
once more.

## 4. Equivalence of the two routes

The two routes from the continuous residual to the discrete tangent are:
$$
\begin{array}{ccc}
R(u;v) & \xrightarrow{\ \text{discretise}\ } & \mathbf r(\hat{\mathbf u}) \\[4pt]
{\scriptstyle\ \text{Directional derivative}}\Big\downarrow  & & \Big\downarrow {\scriptstyle\ \text{differentiate / linearise}} \\[4pt]
K(u;v,w) & \xrightarrow{\ \text{discretise}\ } & \mathbf K(\hat{\mathbf u})
\end{array}
$$

Each continuous statement corresponds to a discrete counterpart by substituting the expansions in the finite element basis:
| | Continuous | Substitution | Discrete |
|---|---|---|---|
| 1 | Find $u$ such that $\Pi(u)$ is stationary | $u_h = \hat u_a\phi_a$ | Find $\hat{\mathbf u}$ such that $\Pi_h(\hat{\mathbf u}) = \Pi(\hat u_a\phi_a)$ is stationary |
| 2 | $R(u;\,v) = D\Pi(u)[v] = 0 \ \ \forall\, v$ | $u_h = \hat u_a\phi_a$, $\ v_h = \hat v_i\phi_i$ | $r_i(\hat{\mathbf u}) = R(u_h;\,\phi_i) = 0 \ \ \forall\, i \iff \mathbf r(\hat{\mathbf u}) = \mathbf 0$ |
| 3 | $K(u;\,v,\,w) = DR(u;\,v)[w]$ | $u_h = \hat u_a\phi_a$, $\ v_h = \hat v_i\phi_i$, $\ w_h = \hat w_j\phi_j$ | $K_{ij}(\hat{\mathbf u}) = DR(u_h;\,\phi_i)[\phi_j]$ |
| 4 | $R(u+\Delta u;\,v) = R(u;\,v) + DR(u;\,v)[\Delta u] + \mathcal O(\Vert\Delta u\Vert^2)$ | $u_h = \hat u_a\phi_a$, $\ \Delta u_h = \Delta\hat u_j\phi_j$ | $\mathbf r(\hat{\mathbf u}+\Delta\hat{\mathbf u}) = \mathbf r(\hat{\mathbf u}) + \mathbf K(\hat{\mathbf u})\,\Delta\hat{\mathbf u} + \mathcal O(\Vert\Delta\hat{\mathbf u}\Vert^2)$ |

**Row 4:**  Setting the discrete expansion to zero and dropping the higher-order term gives the Newton-Raphson system $\mathbf K(\hat{\mathbf u})\,\Delta\hat{\mathbf u} = -\mathbf r(\hat{\mathbf u})$ obtained earlier.

**Directional derivatives:**
$$
R(u;v) = D\Pi(u)[v] = \lim_{\varepsilon\to 0}\frac{\Pi(u+\varepsilon v) - \Pi(u)}{\varepsilon} \ , \qquad
K(u;v,w) = DR(u;v)[w] = \lim_{\varepsilon\to 0}\frac{R(u+\varepsilon w;\,v) - R(u;\,v)}{\varepsilon}
$$

**Linearity and bilinearity.** These give the steps from the continuous to the discrete column in rows 2 and 3. For $v_h = \hat v_i\phi_i$ and $w_h = \hat w_j\phi_j$:
$$
R(u_h;\,v_h) = \hat v_i\,R(u_h;\,\phi_i) = \hat v_i\,r_i(\hat{\mathbf u}) = \hat{\mathbf v} \cdot \mathbf r(\hat{\mathbf u})
$$
$$
K(u_h;\,v_h,\,w_h) = DR(u_h;\,v_h)[w_h] = \hat v_i\,\hat w_j\,DR(u_h;\,\phi_i)[\phi_j] = \hat v_i\,K_{ij}(\hat{\mathbf u})\,\hat w_j = \hat{\mathbf v} \cdot \mathbf K(\hat{\mathbf u})\,\hat{\mathbf w}
$$


The two steps commute: discretise-then-linearise and linearise-then-discretise give the same $\mathbf K$. At the fully discrete level the equality is exact provided $\mathbf r$ and $\mathbf K$ are evaluated with the same quadrature rule, because differentiation commutes with the finite quadrature sum.

## 5. Extension to vector-valued fields

For a vector-valued field $\mathbf u : \Omega \to \mathbb R^d$ (like the 3D displacement field $\mathbf u = [u \quad v \quad w]^T$), the formulation carries over unchanged once the shape-functions are modified as follows.

1. **Degrees of freedom:** Each node carries $d$ degrees of freedom, giving $dn$ in total. With node-wise ordering, component $\alpha \in \{1,\dots,d\}$ of node $a$ is DOF $i = d(a-1)+\alpha$, and $\hat{\mathbf u} \in \mathbb R^{dn}$.

2. **Scalar shape functions:** These are unchanged: $\phi_a , \ a = 1,\dots,n$.

3. **Vector basis functions:** Each $\phi_a$ generates $d$ vector-valued basis functions, one per DOF:
$$
\boldsymbol\phi_i = \phi_a\,\mathbf e_\alpha, \qquad i = d(a-1)+\alpha, \qquad i = 1,\dots,dn
$$
   where $\mathbf e_\alpha$ is the $\alpha$-th unit vector of $\mathbb R^d$. \
   For $d=2$: $\quad \boldsymbol\phi_{2a-1} = \phi_a\mathbf e_1 = [\phi_a,\,0]^{\mathsf T} \quad $ and $\quad \boldsymbol\phi_{2a} = \phi_a\mathbf e_2 = [0,\,\phi_a]^{\mathsf T}$.

4. **Shape function matrix:** The $1\times n$ row $\boldsymbol\phi^{\mathsf T}$ is replaced by the $d\times dn$ matrix $\boldsymbol\Phi$, whose $i$-th column is $\boldsymbol\phi_i$:
$$
\boldsymbol\Phi = \begin{bmatrix} \boldsymbol\phi_1 & \boldsymbol\phi_2 & \cdots & \boldsymbol\phi_{dn} \end{bmatrix}
= \begin{bmatrix} \phi_1\mathbf I & \phi_2\mathbf I & \cdots & \phi_n\mathbf I \end{bmatrix}
\overset{d=2}{=} \begin{bmatrix} \phi_1 & 0 & \phi_2 & 0 & \cdots & \phi_n & 0 \\ 0 & \phi_1 & 0 & \phi_2 & \cdots & 0 & \phi_n \end{bmatrix},
$$
$$
\mathbf u_h = \hat u_j\,\boldsymbol\phi_j = \boldsymbol\Phi\,\hat{\mathbf u} = \sum_{a=1}^{n}\phi_a\begin{bmatrix}\hat u_{2a-1}\\ \hat u_{2a}\end{bmatrix} \ (d=2)
$$

5. **Residual and tangent:** With $i = d(a-1)+\alpha$ and $j = d(b-1)+\beta$:
$$
r_i = R(\mathbf u_h; \, \boldsymbol\phi_i), \qquad K_{ij} = K(\mathbf u_h; \, \boldsymbol\phi_i, \, \boldsymbol\phi_j).
$$
   $\mathbf K$ is $dn\times dn$, and each node pair $(a,b)$ contributes a $d\times d$ block.

6. **Gradients:** The gradient of a vector basis function is a second-order tensor: $\nabla\boldsymbol\phi_i = \mathbf e_\alpha\otimes\nabla\phi_a$, with components
$$
(\phi_i)_{r,s} \equiv \frac{\partial(\boldsymbol\phi_i)_r}{\partial X_s} = \delta_{r\alpha}\,\frac{\partial\phi_a}{\partial X_s},
$$
   where $a = a(i)$ and $\alpha = \alpha(i)$ are the node and component of DOF $i$, and $\nabla$ denotes the gradient with respect to the reference coordinates $\mathbf X$.

### 5.1 Example: Compressible neo-Hookean material

$\Omega_0$ denotes the reference configuration and $\mathbf X$ the reference coordinates.

**Energy:**
$$
\mathbf F = \mathbf I + \nabla \mathbf u, \qquad J = \det\mathbf F
$$
$$
W(\mathbf F) = \frac{\mu}{2}\big(\operatorname{tr}(\mathbf F^{\mathsf T}\mathbf F) - 3\big) - \mu\ln J + \frac{\lambda}{2}(\ln J)^2, \qquad
\Pi(\mathbf u) = \int_{\Omega_0} W\big(\mathbf F(\mathbf u)\big)\,\mathrm d\Omega
$$
Only the internal energy (strain energy) is included here, i.e. no body forces or surface tractions. \
For plane strain ($d = 2$, as in the code), $\mathbf F$ is $2\times 2$ with out-of-plane $F_{33} = 1$, so $\operatorname{tr}(\mathbf F^{\mathsf T}\mathbf F)$ is replaced by $\operatorname{tr}(\mathbf F^{\mathsf T}\mathbf F) + 1$, while $\det\mathbf F$ is unchanged.

**First Piola-Kirchhoff Stress $\mathbf P$:** \
Using the relation (for 3D) $\det(\mathbf A+\mathbf B) = \det\mathbf A + \operatorname{cof}\mathbf A : \mathbf B + \mathbf A : \operatorname{cof}\mathbf B + \det\mathbf B$, setting $\mathbf A = \mathbf F$ and $\mathbf B = \varepsilon\mathbf H$, and ignoring $\mathcal O (\varepsilon^2)$ or higher terms:
$$
 DJ(\mathbf F)[\mathbf H] = \operatorname{cof}\mathbf F : \mathbf H
\quad\Longrightarrow\quad
\frac{\partial J}{\partial\mathbf F} = \operatorname{cof}\mathbf F = J\,\mathbf F^{-\mathsf T}
$$

$$
\operatorname{tr}(\mathbf F^{\mathsf T}\mathbf F) = (\mathbf F^{\mathsf T}\mathbf F)_{qq} = F_{pq}F_{pq}
\quad\Longrightarrow\quad
\frac{\partial}{\partial F_{rs}}\big(F_{pq}F_{pq}\big) = 2F_{pq}\,\frac{\partial F_{pq}}{\partial F_{rs}} = 2F_{pq}\,\delta_{pr}\delta_{qs} = 2F_{rs}
\quad\Longrightarrow\quad
\frac{\partial}{\partial\mathbf F}\operatorname{tr}(\mathbf F^{\mathsf T}\mathbf F) = 2\mathbf F 
$$

$$
\frac{\partial \ln J}{\partial\mathbf F} = \frac{1}{J}\,\frac{\partial J}{\partial\mathbf F} = \mathbf F^{-\mathsf T} \quad\Longrightarrow\quad
\frac{\partial (\ln J)^2}{\partial\mathbf F} = 2\ln J\,\frac{\partial \ln J}{\partial\mathbf F} = 2\ln J\,\mathbf F^{-\mathsf T} 
$$

$$
\therefore \mathbf P = \frac{\partial W}{\partial\mathbf F}
= \frac{\mu}{2}\,(2\mathbf F) - \mu\,\mathbf F^{-\mathsf T} + \frac{\lambda}{2}\,\big(2\ln J\,\mathbf F^{-\mathsf T}\big)
= \mu\mathbf F - \mu\mathbf F^{-\mathsf T} + \lambda\ln J\,\mathbf F^{-\mathsf T} 
$$

**Residual:**\
We have $D\mathbf F(\mathbf u)[\mathbf v] = \nabla\mathbf v$, and by the chain rule
$$
R(\mathbf u; \, \mathbf v) = \left.\frac{\mathrm d}{\mathrm d\varepsilon}\int_{\Omega_0} W\big(\mathbf F(\mathbf u) + \varepsilon\nabla\mathbf v\big)\,\mathrm d\Omega\right|_{\varepsilon=0}
= \int_{\Omega_0} \frac{\partial W}{\partial\mathbf F} : \nabla\mathbf v\,\mathrm d\Omega
= \int_{\Omega_0} \mathbf P\big(\mathbf F(\mathbf u)\big) : \nabla \mathbf v \,\mathrm d\Omega
$$

**Discretise:**\
Set $\mathbf u = \mathbf u_h$ and $\mathbf v = \boldsymbol\phi_i$, with $a = a(i)$ and $\alpha = \alpha(i)$. Then $\mathbf P : \nabla\boldsymbol\phi_i = P_{rs}\,(\phi_i)_{r,s} = P_{rs}\,\delta_{r\alpha}\,\partial\phi_a/\partial X_s = P_{\alpha s}\,\partial\phi_a/\partial X_s$, so
$$
\therefore \quad \boxed{r_i(\hat{\mathbf u}) = \int_{\Omega_0} \mathbf P\big(\mathbf F(\mathbf u_h)\big) : \nabla\boldsymbol\phi_i \,\mathrm d\Omega = \int_{\Omega_0} P_{\alpha s}\,\frac{\partial \phi_a}{\partial X_s}\,\mathrm d\Omega}
$$

**Linearise:**
$$
\mathbf F(\mathbf u_h) = \mathbf I + \nabla\mathbf u_h = \mathbf I + \hat u_k\,\nabla\boldsymbol\phi_k \quad \Rightarrow \quad
\frac{\partial\mathbf F(\mathbf u_h)}{\partial\hat u_j} = \delta_{kj}\,\nabla\boldsymbol\phi_k = \nabla\boldsymbol\phi_j \ , \qquad
\frac{\partial F_{rs}}{\partial\hat u_j} = (\phi_j)_{r,s}
$$

Using the first elasticity tensor $\boldsymbol{\mathcal A} = \partial\mathbf P/\partial\mathbf F\big|_{\mathbf F(\mathbf u_h)}$, $\mathcal A_{pqrs} = \partial P_{pq}/\partial F_{rs}$:
$$
\frac{\partial P_{pq}}{\partial\hat u_j}
= \frac{\partial P_{pq}}{\partial F_{rs}}\,\frac{\partial F_{rs}}{\partial\hat u_j}
= \mathcal A_{pqrs}\,(\phi_j)_{r,s}
= \big(\boldsymbol{\mathcal A}\,\nabla\boldsymbol\phi_j\big)_{pq},
$$
where $(\boldsymbol{\mathcal A}\,\mathbf T)_{pq} = \mathcal A_{pqrs}T_{rs}$. 

$$
K_{ij} = \frac{\partial r_i}{\partial\hat u_j}
= \int_{\Omega_0} \frac{\partial\mathbf P}{\partial\hat u_j} : \nabla\boldsymbol\phi_i\,\mathrm d\Omega
= \int_{\Omega_0} \frac{\partial P_{pq}}{\partial\hat u_j}\,(\phi_i)_{p,q}\,\mathrm d\Omega
$$
$$
\therefore \quad \boxed{K_{ij}= \int_{\Omega_0} \nabla\boldsymbol\phi_i : \left(\boldsymbol{\mathcal A}\,\nabla\boldsymbol\phi_j\right)\mathrm d\Omega
= \int_{\Omega_0} (\phi_i)_{p,q}\,\mathcal A_{pqrs}\,(\phi_j)_{r,s}\,\mathrm d\Omega}
$$

> - **Symmetry.** Since $\mathbf P = \partial W/\partial\mathbf F$, $\mathcal A_{pqrs} = \partial^2 W/\partial F_{pq}\,\partial F_{rs} \Rightarrow $ major symmetry $ \mathcal A_{pqrs} = \mathcal A_{rspq}$. Hence
>   $$
>   \nabla\boldsymbol\phi_i : \left(\boldsymbol{\mathcal A}\,\nabla\boldsymbol\phi_j\right)
>   = (\phi_i)_{p,q}\,\mathcal A_{pqrs}\,(\phi_j)_{r,s}   = (\phi_j)_{r,s}\,\mathcal A_{rspq}\,(\phi_i)_{p,q}  = \nabla\boldsymbol\phi_j : \left(\boldsymbol{\mathcal A}\,\nabla\boldsymbol\phi_i\right),
>   $$
>   so $K_{ij} = K_{ji}$. This is consistent with the formulation from earlier.

## 6. Connection to the DOLFINx tangent check

The code uses the vector-valued neo-Hookean model in 2D.
- `Res = ufl.derivative(Pi, uh, v_)` $\;\to\; R(\mathbf u_h; \, \mathbf v) = D\Pi(\mathbf u_h)[\mathbf v]$
- `Jac = ufl.derivative(Res, uh, du)` $\;\to\; K(\mathbf u_h; \, \mathbf v, \, \mathbf w) = DR(\mathbf u_h; \, \mathbf v)[\mathbf w]$
UFL performs the directional differentiation symbolically on the continuous form and then assembles it, i.e. it follows the **linearise → discretise** route of §4.

**Finite-difference check.** For random coefficient vectors $\hat{\mathbf v}$ and $\hat{\mathbf w}$, the forward-difference quotient satisfies
$$
\frac{R(\mathbf u_h + \varepsilon \, \mathbf w_h; \, \mathbf v_h) - R(\mathbf u_h;\,\mathbf v_h)}{\varepsilon} = \hat{\mathbf v}^{\mathsf T}\mathbf K\,\hat{\mathbf w} + \mathcal O(\varepsilon).
$$
- `K_vw` computes $\hat{\mathbf v}^{\mathsf T}\mathbf K\,\hat{\mathbf w}$.
- `fd` computes the difference quotient.
Agreement between the two, with an error that decreases linearly in $\varepsilon$, verifies the procedure.

```python
from mpi4py import MPI
from dolfinx import fem, mesh
from dolfinx.fem import petsc
from petsc4py import PETSc
import ufl
import numpy as np
```

```python
N = 4
msh = mesh.create_unit_square(MPI.COMM_WORLD, N, N)

V  = fem.functionspace(msh, ("Lagrange", 1, (2,)))
v_ = ufl.Argument(V, 0)   # test placeholder for the forms (arg 0 -> residual entries / rows)
du = ufl.Argument(V, 1)   # increment placeholder (arg 1 -> matrix columns)

# plane-strain neo-Hookean moduli
E, nu = 1.0, 0.3
mu    = E/(2*(1+nu))
lmbda = E*nu/((1+nu)*(1-2*nu))
```

```python
uh = fem.Function(V)                     # the state we linearize about
F  = ufl.Identity(2) + ufl.grad(uh)
J  = ufl.det(F)
Ic = ufl.tr(F.T*F) + 1                   # plane strain: +1 (z-stretch)
W  = (mu/2)*(Ic - 3) - mu*ufl.ln(J) + (lmbda/2)*ufl.ln(J)**2
Pi = W*ufl.dx                            # internal energy

Res = ufl.derivative(Pi,  uh, v_)   # first variation -> residual  R(u; .)
Jac = ufl.derivative(Res, uh, du)   # second variation -> tangent  K(u; . , .)

Res_form = fem.form(Res)
Jac_form = fem.form(Jac)
```

```python
# freeze a deliberate state (need not be equilibrium; any uh works for the tangent check)
uh.interpolate(lambda x: np.array([0.15*x[0]*x[1], 0.10*x[1]**2]))

# random test func v, and random increment du = w
rng = np.random.default_rng(0)
v = fem.Function(V); v.x.array[:] = rng.standard_normal(v.x.array.size)
w = fem.Function(V); w.x.array[:] = rng.standard_normal(w.x.array.size)
```

```python
# keep a copy of the unperturbed function uh
u0 = uh.x.array.copy()

R = petsc.assemble_vector(Res_form)   # r_i = R(uh; basis function of dof i), i = 1, ..., n_dof
K = petsc.assemble_matrix(Jac_form); K.assemble()   # K_ij = K(uh; basis functions of dofs i, j)

vvec = v.x.petsc_vec
wvec = w.x.petsc_vec

Kw  = K.createVecLeft(); K.mult(wvec, Kw)
R_v = R.dot(vvec)
K_vw = vvec.dot(Kw)
```

```python
print(f"{'h':>8}  {'K_vw':>14}  {'fd':>14}  {'|K_vw - fd|':>12}")
for eps in np.logspace(-1, -5, 5):
    uh.x.array[:] = u0 + eps*w.x.array
    R_hw = petsc.assemble_vector(Res_form)
    uh.x.array[:] = u0

    fd = (R_hw.dot(vvec) - R_v)/eps
    print(f"{eps:8.1e}  {K_vw:14.10f}  {fd:14.10f}  {abs(K_vw - fd):12.3e}")
```
