# Nonlinear Elasticity
### Kinematics / Constitutive Law

## 1. Why go from linear elasticity to nonlinear elasticity?

Linear elasticity rests on three small-deformation assumptions. The displacement gradients are small ($|\nabla\mathbf u| \ll 1$). The deformed body is close enough to the undeformed body that all calculations can be done on the undeformed (reference) body. And stress is proportional to strain, $\boldsymbol\sigma = \mathbb C:\boldsymbol\varepsilon$, for any amount of deformation. The four reasons below are the ways these assumptions break down.

| # | Reason | Why it matters |
|---|---|---|
| 1 | Strain is not measured well | $\boldsymbol\varepsilon$ drops a term that is quadratic in $\nabla\mathbf u$, so it gives the wrong length changes when $\nabla\mathbf u$ is not small. |
| 2 | $\boldsymbol\varepsilon = \operatorname{sym}(\nabla\mathbf u)$ is not a good measure (not objective) | $\boldsymbol\varepsilon$ changes under a rigid rotation, so a body that only rotates shows false strain and false stress. |
| 3 | Large deformation | Equilibrium $\operatorname{div}\boldsymbol\sigma + \mathbf f = \mathbf 0$ holds on the deformed body, which is no longer close to the undeformed one. Derivatives must be taken with respect to $\mathbf x$, so the shape-function derivatives pick up $\mathbf F$ (updated Lagrangian). |
| 4 | Material response | $\boldsymbol\sigma = \mathbb C:\boldsymbol\varepsilon$ stays linear for any deformation, so crushing to zero volume ($J\to0$) costs only finite energy. With a nonlinear material, $a(\mathbf u,\mathbf v) = \ell(\mathbf v)$ is no longer bilinear and must be solved in the deformed configuration. |

### 1.1 Strain is not measured well

Strain should tell us how much the material fibres change length. A small fibre $d\mathbf X$ in the reference body becomes $d\mathbf x = \mathbf F\,d\mathbf X$, where $\mathbf F = \mathbf I + \nabla\mathbf u$, and the exact change in squared length is
$$
|d\mathbf x|^2 - |d\mathbf X|^2 = d\mathbf X^T(\mathbf F^T\mathbf F - \mathbf I)\,d\mathbf X = 2\,d\mathbf X^T\mathbf E\,d\mathbf X,
\qquad
\mathbf E = \tfrac12(\mathbf F^T\mathbf F - \mathbf I)
= \underbrace{\tfrac12(\nabla\mathbf u + \nabla\mathbf u^T)}_{\boldsymbol\varepsilon}
+ \tfrac12\nabla\mathbf u^T\nabla\mathbf u .
$$
The linear strain $\boldsymbol\varepsilon$ keeps only the first-order part and drops $\tfrac12\nabla\mathbf u^T\nabla\mathbf u$. When $|\nabla\mathbf u| \ll 1$ the dropped term is negligible and $\boldsymbol\varepsilon$ is fine. When the displacement gradients are large, the dropped term is as big as $\boldsymbol\varepsilon$ itself, so the "strain" no longer matches the real stretching of the fibres. The clearest example is a rigid rotation $\mathbf x \to \mathbf R\mathbf x$, worked out in 1.2: no fibre changes length, yet $\boldsymbol\varepsilon$ reports about 30% compression.

### 1.2 $\boldsymbol\varepsilon = \operatorname{sym}(\nabla\mathbf u)$ is not a good measure (not objective)

A strain measure must be objective (frame-indifferent). If we superpose a rigid rotation on a deformation, the strain must not change, because a rigid motion does not deform the material. To see that $\boldsymbol\varepsilon$ fails this test, take a pure rotation $\mathbf x \to \mathbf R\mathbf x$, where $\mathbf R$ is a rotation. Then
$$
\mathbf u(\mathbf x) = \mathbf R\mathbf x - \mathbf x = (\mathbf R - \mathbf I)\,\mathbf x,
\qquad
\nabla \mathbf u = \mathbf R - \mathbf I,
$$
$$
\boldsymbol\varepsilon = \frac{(\mathbf R-\mathbf I)^T + (\mathbf R-\mathbf I)}{2}
= \frac{\mathbf R^T + \mathbf R}{2} - \mathbf I .
$$
Take a rotation by 45° (Figure 1).



*Figure 1 — Rigid rotation by 45°: the axes rotate, but no lengths or angles change.*

For this rotation,
$$
\mathbf R = \begin{bmatrix} \tfrac{1}{\sqrt2} & -\tfrac{1}{\sqrt2} \\[2pt] \tfrac{1}{\sqrt2} & \tfrac{1}{\sqrt2}\end{bmatrix},\qquad
\mathbf R^T = \begin{bmatrix} \tfrac{1}{\sqrt2} & \tfrac{1}{\sqrt2} \\[2pt] -\tfrac{1}{\sqrt2} & \tfrac{1}{\sqrt2}\end{bmatrix},
$$
$$
\frac{\mathbf R + \mathbf R^T}{2} - \mathbf I =
\begin{bmatrix} \tfrac{1}{\sqrt2}-1 & 0 \\[2pt] 0 & \tfrac{1}{\sqrt2}-1\end{bmatrix}
\approx \begin{bmatrix} -0.3 & 0 \\ 0 & -0.3\end{bmatrix},
$$
so $\boldsymbol\varepsilon$ predicts about 30% strain (exactly $\tfrac{1}{\sqrt2} - 1 = -0.293$), even though the body has only rotated and no length or angle has changed.

The Green–Lagrange strain gives the correct answer. For a rotation, the term that $\boldsymbol\varepsilon$ drops (1.1) cancels it exactly:
$$
\tfrac12\nabla\mathbf u^T\nabla\mathbf u = \tfrac12(\mathbf R-\mathbf I)^T(\mathbf R-\mathbf I) = \mathbf I - \tfrac{\mathbf R+\mathbf R^T}{2} = -\boldsymbol\varepsilon
\quad\Rightarrow\quad \mathbf E = \tfrac12(\mathbf R^T\mathbf R - \mathbf I) = \mathbf 0 .
$$
The same holds for any rotation $\mathbf Q$ superposed on any deformation. $\mathbf F$ becomes $\mathbf Q\mathbf F$, and $(\mathbf Q\mathbf F)^T(\mathbf Q\mathbf F) = \mathbf F^T\mathbf Q^T\mathbf Q\mathbf F = \mathbf F^T\mathbf F$, so $\mathbf C$ and $\mathbf E$ are unchanged. The linear strain, on the other hand, becomes $\boldsymbol\varepsilon = \operatorname{sym}(\mathbf Q\mathbf F) - \mathbf I$, which depends on $\mathbf Q$.

Linear elasticity still works when rotations are small. For a small angle $\theta$, $\mathbf R \approx \mathbf I + \mathbf W$ with $\mathbf W = \begin{bmatrix}0 & -\theta\\ \theta & 0\end{bmatrix}$ skew-symmetric, so $\boldsymbol\varepsilon = \operatorname{sym}(\mathbf W) = \mathbf 0$. The linear strain is therefore objective only to first order. The error is $1-\cos\theta \approx \theta^2/2$, which is negligible for small $\theta$ but reaches 0.29 at 45°.

Because $\boldsymbol\sigma = \mathbb C:\boldsymbol\varepsilon$, this false strain produces false stress. A structure that undergoes large rotation, for example a slender beam with a large tip deflection or a rotating part, would show stresses that are not really there.

### 1.3 Large deformation

Linear elasticity treats the deformed shape as if it were the undeformed shape, so equilibrium is written, integrated and differentiated on the reference body. But $\operatorname{div}\boldsymbol\sigma + \mathbf f = \mathbf 0$ holds in the deformed configuration: forces balance on the body as it actually is.

When the deformation is large, three things go wrong with using the reference body:
- The domain where equilibrium holds, and the boundary where tractions act, have moved and changed shape.
- Lengths, areas and volumes have changed, so integrals over the reference body no longer describe the real body.
- The derivatives in $\operatorname{div}\boldsymbol\sigma$ must be taken with respect to the current coordinates $\mathbf x$, not $\mathbf X$.

The equations must therefore be written on the deformed configuration. That configuration depends on the unknown $\mathbf u$, so the problem becomes nonlinear (geometric nonlinearity).

The last point is where the FE implementation changes. The displacement can be interpolated either on the reference configuration or on the deformed configuration,
$$
\mathbf u = \mathbf u_a\, N_a(\mathbf X) \qquad\text{or}\qquad \mathbf u = \mathbf u_a\, n_a(\mathbf x),
$$
summed over the nodes $a$. Here $N_a(\mathbf X)$ is the basis function on the reference configuration, and $n_a(\mathbf x)$ is the shape function written as a function of the deformed configuration.

The divergence $\operatorname{div}\boldsymbol\sigma$ needs the derivatives $\dfrac{\partial u_i}{\partial x_j}$, which are variables of the deformed state. What we actually compute from the reference mesh is $\dfrac{\partial u_i}{\partial X_j}$.

The two shape functions are the same function written in different variables,
$$
N(\mathbf X) = n\big(\boldsymbol\phi(\mathbf X)\big),
$$
where $\mathbf X$ is the reference (undeformed) position and $\boldsymbol\phi(\mathbf X) = \mathbf X + \mathbf u(\mathbf X) = \mathbf x$ is the deformed position (Figure 2).



*Figure 2 — The deformation map φ takes the reference body (X) to the deformed body (x = X + u(X)).*

Differentiating $N(\mathbf X) = n(\boldsymbol\phi(\mathbf X))$ with the chain rule gives
$$
\frac{\partial N}{\partial X_I} = \frac{\partial n}{\partial x_j}\cdot\underbrace{\frac{\partial \phi_j}{\partial X_I}}_{\text{this term we ignore normally}}
$$
where $\dfrac{\partial \phi_j}{\partial X_I} = F_{jI}$ is the deformation gradient.

In linear FEM the deformation is small, so $\mathbf F \approx \mathbf I$ and $\partial/\partial\mathbf x \approx \partial/\partial\mathbf X$. That is why this term is normally ignored. With large deformation $\mathbf F \neq \mathbf I$, so the term has to be kept. Taking the derivatives on the current (deformed) configuration in this way is the updated Lagrangian formulation.

### 1.4 Material response

Linear elasticity uses $\boldsymbol\sigma = \mathbb C:\boldsymbol\varepsilon$, which assumes stress stays proportional to strain however large the deformation. Real materials do not behave this way.

The local volume ratio is $J = \det\mathbf F$, with $dv = J\,dV$. Mass balance between $\rho_0$ and $\rho$ then gives $\rho\,J = \rho_0$. A real material cannot be crushed to zero volume ($J\to0$) or turned inside out ($J<0$), so resisting this should require stress and energy that go to infinity as $J\to0$ (Figure 3).

The linear law has no such barrier. It gives a finite energy $(9\lambda+6\mu)\,\text{vol}$ at $J = 0$ (Section 2). It also treats compression and tension of the same size identically.



*Figure 3 — Material response: J = det F = dv/dV is the volume (area) ratio. As J → 0 the element is crushed.*

The material response also changes the weak form $a(\mathbf u,\mathbf v) = \ell(\mathbf v)$, which we now have to solve in the deformed configuration. In linear elasticity everything is integrated over the fixed, undeformed body $\Omega_0$:
$$
a(\mathbf u,\mathbf v) = \int_{\Omega_0} \boldsymbol\varepsilon(\mathbf v):\mathbb C:\boldsymbol\varepsilon(\mathbf u)\,dV,
\qquad
\ell(\mathbf v) = \int_{\Omega_0}\mathbf f\cdot\mathbf v\,dV + \int_{\Gamma_t}\bar{\mathbf t}\cdot\mathbf v\,dA .
$$
Because $\boldsymbol\sigma = \mathbb C:\boldsymbol\varepsilon$ is linear, $a(\mathbf u,\mathbf v)$ is bilinear, which gives a single linear system $\mathbf K\mathbf u = \mathbf f$.

With a nonlinear material and large deformation, the weak form has to be written on the deformed body $\Omega$:
$$
\int_{\Omega} \boldsymbol\sigma(\mathbf u) : \nabla_{\mathbf x}\mathbf v \; dv
= \int_{\Omega}\mathbf f\cdot\mathbf v\; dv + \int_{\partial\Omega_t}\mathbf t\cdot\mathbf v\; da,
\qquad dv = J\,dV .
$$

This changes the problem in three ways:
- The stress is a nonlinear function of $\mathbf u$ (through $\mathbf F$), so the left side is no longer bilinear. It is linear in $\mathbf v$ only.
- The integration domain $\Omega$ and the volume element $dv = J\,dV$ also depend on the unknown $\mathbf u$. So we are solving in a configuration that is not known in advance, and the system has to be solved iteratively (e.g. with Newton's method).
- $J$ appears in every integral. An element with $J \to 0$ has collapsed, and one with $J < 0$ has turned inside out (Figure 3). Neither is physical, so the material law itself must keep $J > 0$.

This calls for a nonlinear constitutive law $W(\mathbf F)$ with $W\to\infty$ as $J\to0$ (material nonlinearity). That leads to the Neo-Hookean model.

## 2. Linear elasticity under extreme compression

Let
$$\mathbf u(\mathbf x) = -\alpha\,\mathbf x$$

$$
\nabla \mathbf u = -\alpha\,\mathbf I, \qquad \boldsymbol\varepsilon = -\alpha\,\mathbf I
$$

$$
\boldsymbol\sigma = \lambda\,\operatorname{tr}(\boldsymbol\varepsilon)\,\mathbf I + 2\mu\,\boldsymbol\varepsilon
= \lambda(-3\alpha)\,\mathbf I - 2\mu\alpha\,\mathbf I
= -\alpha(3\lambda + 2\mu)\,\mathbf I
$$

$$
\int \boldsymbol\sigma : \boldsymbol\varepsilon = \int \alpha^2(3\lambda+2\mu)\cdot 3
= 3\alpha^2(3\lambda + 2\mu)\,\text{vol} \;\;\rightarrow\;\; \text{this is finite}
$$

Now look at the kinematics:
$$
\mathbf F = \mathbf I + \nabla\mathbf u = (1-\alpha)\,\mathbf I, \qquad
\det\mathbf F = J = (1-\alpha)^3
$$

$$
\alpha \to 1 \;\Rightarrow\; J \to 0
\;\Rightarrow\; \int\boldsymbol\sigma:\boldsymbol\varepsilon \;\to\; (9\lambda + 6\mu)\,\text{vol} \;\rightarrow\; \text{finite value}
$$

Linear elasticity gives this finite value. In reality, $J \to 0$ (crushing the body to zero volume) should not be possible at finite energy.

> Note: the stored strain energy is $\tfrac12\int\boldsymbol\sigma:\boldsymbol\varepsilon = \tfrac32\alpha^2(3\lambda+2\mu)\,\text{vol}$. The factor ½ does not change the conclusion, because the value is still finite as $J \to 0$.

## 3. Neo-Hookean constitutive relation (1D)

Start from a linear spring:
$$F = kx$$

Stretch ratio $\lambda = \dfrac{\ell}{\ell_0}$ (current length / original length):
- $\lambda = 1$ : undeformed
- $\lambda = 0$ : must be prevented

Strain measures: the 1D linear strain is $\dfrac{\ell - \ell_0}{\ell_0}$. The finite-strain (Green–Lagrange) measure is
$$
\mathbf E = \frac{\mathbf F^T\mathbf F - \mathbf I}{2} = \frac{\mathbf C - \mathbf I}{2}, \qquad \mathbf C = \mathbf F^T\mathbf F .
$$

Energy as a function of stretch:
$$
W(\lambda) = \frac{\mu}{2}\left(\lambda^2 - 1\right) - \mu\log\lambda
$$
- $W$ : energy, $\ \lambda$ : stretch
- $-\mu\log\lambda$ : log barrier. It keeps the variable in bounds ($W\to\infty$ as $\lambda\to0$), so $\lambda = 0$ is prevented.

First derivative (force):
$$
W'(\lambda) = \frac{\mu}{2}(2\lambda) - \frac{\mu}{\lambda} = \mu\left(\lambda - \frac{1}{\lambda}\right)
$$

Second derivative:
$$
W''(\lambda) = \mu\left(1 + \frac{1}{\lambda^2}\right) \;\rightarrow\; \text{always positive}
$$



*Figure 4 — Neo-Hookean 1D (μ = 1). Energy W has its minimum at λ = 1 and blows up as λ → 0. log λ → −∞ as λ → 0. Force W′ is zero at λ = 1 and → −∞ as λ → 0.*

## 4. What happens when $\lambda = 1 + \varepsilon$, $\ \varepsilon \ll 1$?

$$
W'(\lambda) = \mu\left[(1+\varepsilon) - \frac{1}{1+\varepsilon}\right]
$$
Binomial expansion: $\dfrac{1}{1+\varepsilon} = 1 - \varepsilon + \dots$
$$
W'(\lambda) \approx \mu\big(1+\varepsilon - (1-\varepsilon)\big)
= 2\mu\varepsilon
= W''(1)\,\varepsilon
$$

For small strains, the Neo-Hookean model reduces to the linear law, with stiffness $W''(1) = 2\mu$. This is the dashed tangent in Figure 4.

## 5. 3D analog

$$
W(\mathbf F) = \frac{\mu}{2}\,(I_1 - 3) \;-\; \mu\log J \;+\; \frac{\lambda}{2}\,(\log J)^2
$$
$$
I_1 = \operatorname{tr}(\mathbf C), \qquad J = \det(\mathbf F)
$$

Correspondence with 1D: $\ I_1 - 3 \;\leftrightarrow\; \lambda^2 - 1$ and $\ \log J \;\leftrightarrow\; \log\lambda$ (log barrier as $J \to 0$).

## 6. Code: block compression, linear elasticity vs Neo-Hookean

```python
import sys, dolfinx
print(sys.executable)
print("dolfinx", dolfinx.__version__)  # 0.11.0


from mpi4py import MPI
from dolfinx import fem, mesh, io
from dolfinx.fem.petsc import LinearProblem, NonlinearProblem
from petsc4py import PETSc
import ufl
import numpy as np


# coarse structured mesh so individual elements are visible when warped
N = 8
msh = mesh.create_unit_square(MPI.COMM_WORLD, N, N)
gdim = msh.geometry.dim
tdim = msh.topology.dim
fdim = tdim - 1

# vector P1 space; plane-strain Lame constants from E, nu
V = fem.functionspace(msh,("Lagrange", 1, (2,)))
E, nu = 1.0, 0.3
mu = E / (2 * (1 + nu))
lmbda = E * nu / ((1 + nu) * (1 - 2 * nu))

# Squash the block:
# clamp the base (y=0), push the top face (y=1) straight down by a compression fraction = alpha
# (both displacement components are fixed on top --> the block barrels). top_disp is updated per load step.
bot = mesh.locate_entities_boundary(msh,fdim,lambda x: np.isclose(x[1], 0.0))
dofs_bot = fem.locate_dofs_topological(V,fdim,bot)
u_bot = fem.Function(V)       # = 0 on the base

alpha = 0.5
top = mesh.locate_entities_boundary(msh,fdim,lambda x: np.isclose(x[1], 1.0))
dofs_top = fem.locate_dofs_topological(V,fdim,top)

# top_disp = fem.Constant(msh, np.array([0.0, -alpha])) # set to (0, -alpha) below
# bc_top = fem.dirichletbc(top_disp, dofs_top, V)

top_disp = fem.Function(V)
top_disp.interpolate(lambda x: np.vstack([np.zeros_like(x[0]),np.full_like(x[0], -alpha)]))
bc_top = fem.dirichletbc(top_disp,dofs_top)
bc_bot = fem.dirichletbc(u_bot,dofs_bot)
bcs = [bc_bot, bc_top]

# ----- linear elasticity: one independent solve per load level -----

def epsilon(w):
    return ufl.sym(ufl.grad(w))

def sigma(w):
    e = epsilon(w)
    return (2 * mu * e + lmbda * ufl.tr(e) * ufl.Identity(2))

u = ufl.TrialFunction(V)
v = ufl.TestFunction(V)

a_lin = ufl.inner(sigma(u),epsilon(v)) * ufl.dx
L_lin = ufl.inner(fem.Constant(msh, np.zeros(2)),v) * ufl.dx

top_disp.interpolate(lambda x: np.vstack([np.zeros_like(x[0]),np.full_like(x[0], -alpha)]))

prob = LinearProblem(a_lin, L_lin, bcs=bcs,
    petsc_options_prefix="lin_",
    petsc_options={
        "ksp_type": "preonly",
        "pc_type": "lu"})
uh_lin = prob.solve()


uh = fem.Function(V)
v = ufl.Argument(V, 0)   # test function
w = ufl.Argument(V, 1)   # trial function

F = ufl.Identity(2) + ufl.grad(uh)
J = ufl.det(F)
Ic = ufl.tr(F.T * F) + 1
W = (mu/2) * (Ic - 3) - mu * ufl.ln(J) + (lmbda/2) * ufl.ln(J)**2
P1 = W * ufl.dx
Res = ufl.derivative(P1, uh, v)
Jac = ufl.derivative(Res, uh, w)

problem = NonlinearProblem(Res,uh,bcs=bcs,J=Jac,
    petsc_options_prefix="nh_",
    petsc_options={
        "snes_type": "newtonls",
        "snes_linesearch_type": "bt",
        "snes_rtol": 1e-8,
        "snes_atol": 1e-10,
        "snes_max_it": 25,
        "ksp_type": "preonly",
        "pc_type": "lu"})
problem.solve()

uh.name = "disp1"
with io.XDMFFile(msh.comm,"nh.xdmf","w") as file:
    file.write_mesh(msh)
    file.write_function(uh)

uh_lin.name = "disp2"
with io.XDMFFile(msh.comm,"linear.xdmf","w") as file:
    file.write_mesh(msh)
    file.write_function(uh_lin)
```

### Summary of the code

The code squashes a unit-square block (8×8 triangular mesh, $E = 1$, $\nu = 0.3$, plane strain) to half its height, with the bottom clamped and the top pushed down by 0.5. It solves this once with linear elasticity (a single linear solve of the bilinear form) and once with the Neo-Hookean energy $W = \frac{\mu}{2}(I_1 - 3) - \mu\ln J + \frac{\lambda}{2}(\ln J)^2$, solved with Newton's method.
