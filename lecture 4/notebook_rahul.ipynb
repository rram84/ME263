<span style="font-size:22px;">**Lecture 4 - Our goal is to show that $a(u,v)=L(v)$ <u>has</u> a <u>unique</u> solution $\,\forall\, v \in V$**</span>

There are different ways to do this, each giving different insights:

1. Energy minimization method
2. Representation theory

Representation theory is a bit more abstract; it is also called Riesz representation theorem. We will follow the energy minimization route to reach our goal.

Total potential energy is

$$
\boxed{
J = \frac{1}{2} \int_{\Omega} |\nabla u|^2 \, dA - \int_{\Omega} \mathbf{f} \cdot \mathbf{u} \, dA
}
$$






 

If we are given solutions with mesh sizes $u_1$ (coarser mesh), $u_2$ (fine mesh), $u_3$ (finer mesh), and so on, we expect that the sequence of solutions will approach the exact solution.
There are two questions that we should answer regarding this:
1. If we do not know the exact solution, how can we say that our numerical solution is getting closer and closer to the correct solution?
2.  As we can observe during mesh refinement:
$$J(u_1) > J(u_2) > J(u_3) > \dots > J(u_n)$$
Simply observing that the energy $J(u_h)$ decreases does not automatically guarantee that the solution $u_h$ is approaching the true solution $u$.Roughly speaking, between two states $u_1$ and $u_2$, the energy values $J(u_1)$ and $J(u_2)$ may become very close, while the solutions themselves jump significantly in the function space ($\Vert{}u_1 - u_2\Vert{}$ remains large). Therefore, energy values drawing closer together does not alone ensure that we are converging to the exact solution.



 
 <span style="font-size:22px;">**Energy route**</span>

 We have to show that:
 $$
\boxed{V = H^1_0(\Omega) \text{ is complete}} 
$$

"Complete" means that every Cauchy sequence gets closer and closer and converges to an element that lies within the space.

$a(\cdot, \cdot)$ is an inner product.$\Vert{} \cdot \Vert{}_a = \sqrt{a(\cdot, \cdot)}$ is the energy norm.

**Why should $V$ be complete?**



 As shown in the figure, $u_1$ is the solution in the subset $V_1$. When we increase the space to $V_2$, we obtain the solution $u_2$, and so on. We expect this sequence $\{u_k\}$ to be a Cauchy sequence that converges to a value in $V$. This is particularly relevant in the Finite Element Method (FEM), as it ensures that the solutions get progressively closer and converge within $V$.
 this ensures solution is getting closer and converges in $V$.

 Now, we need to show that $V$ is complete.

First, $V$ is a subset of $H^1$ (where $H^1$ also consists of functions that do not vanish on the Dirichlet boundary).

Showing that $V$ is complete is not straightforward. We will start by assuming that $H^1$ is complete and use this fact to show that $V$ is complete.
Since $V$ is a closed subspace in $H^1$, it is also complete.

 For this proof, we will require the Poincaré inequality:
 $$
\boxed{
\|v\|_{L^2(\Omega)}
\leq
C \|\nabla v\|_{L^2(\Omega)} 
}
$$
This inequality means that you can bound a function's size using the norm of its gradient.

$$
\boxed{v(x) = v(0) + \int_0^x v'(s)\,ds}
$$

If we know the derivative of a function, we can reconstruct the function, provided we know $v(0)$. That is why the condition $v = 0$ on the Dirichlet boundary is important.

Hilltop example: If we were given a device that measures the tilt of a hill relative to gravity, we could find the exact height we have climbed by integrating the derivative. However, the catch here is that we must know the initial height.

Is the converse of the Poincaré inequality true?
$$\Vert{}\nabla v\Vert{}_{L^2(\Omega)} \le \Vert{}v\Vert{}_{L^2(\Omega)}$$
No. From the hill example, we cannot tell how high we have climbed just by knowing how far we have traveled horizontally. We can reach the same height by traveling different distances.
Sawtooth example: We can construct a rapidly oscillating "sawtooth" function where the $L^2$ norm of the function, $\Vert{}v\Vert{}_{L^2(\Omega)}$, remains very small ($\mathcal{O}(\varepsilon)$), but its derivative/gradient norm, $\Vert{}\nabla v\Vert{}_{L^2(\Omega)}$, becomes arbitrarily large as the frequency of oscillation increases.


Let $\{u_n\}$ be a Cauchy sequence in $V$:
$$\Vert{}u_m - u_n\Vert{}_a \to 0 \quad \text{as } m, n \to \infty \quad \text{(energy norm)}$$
Is $\{u_n\}$ Cauchy in $H^1$?


$$
\boxed{
\Vert f \Vert_{H^1(\Omega)}^2
=
\Vert f \Vert_{L^2(\Omega)}^2
+
\Vert \nabla f \Vert_{L^2(\Omega)}^2
}
$$
We are given that $\{u_n\}$ is Cauchy in the energy norm, meaning $\Vert{}\nabla u_m - \nabla u_n\Vert{}_{L^2(\Omega)} \to 0$ as $m, n \to \infty$. To show that $\{u_n\}$ is Cauchy in $H^1(\Omega)$, we need to prove that $\Vert{}u_m - u_n\Vert{}_{H^1(\Omega)} \to 0$.

Using the Poincaré inequality $\Vert{}v\Vert{}_{L^2(\Omega)} \le C \Vert{}\nabla v\Vert{}_{L^2(\Omega)}$,
 we have:
$$\Vert{}u_m - u_n\Vert{}_{H^1(\Omega)}^2 = \Vert{}u_m - u_n\Vert{}_{L^2(\Omega)}^2 + \Vert{}\nabla u_m - \nabla u_n\Vert{}_{L^2(\Omega)}^2 \le (C^2 + 1) \Vert{}\nabla u_m - \nabla u_n\Vert{}_{L^2(\Omega)}^2$$

Since $\|\nabla u_m - \nabla u_n\|_{L^2(\Omega)} \to 0$, it follows that $\|u_m - u_n\|_{H^1(\Omega)} \to 0$. 

Therefore, $\{u_n\}$ is a Cauchy sequence in $H^1(\Omega)$. Since $H^1(\Omega)$ is complete, $u_n$ converges to some limit $u \in H^1(\Omega)$:

$$
\|u_m - u\|_{L^2(\Omega)}^2 + \|\nabla u_m - \nabla u\|_{L^2(\Omega)}^2 \to 0 \quad \text{as } m \to \infty
$$

This implies that $\|\nabla u_m - \nabla u\|_{L^2(\Omega)} \to 0$, meaning $u_m \to u$ in the energy norm. 

$V$ is a closed subspace of $H^1(\Omega)$

Since $V$ is the kernel of the trace operator in  $H^1(\Omega)$, and the trace operator is continuous, $V$ is a closed subspace of $H^1(\Omega)$.

 Thus, we have shown that $V$ is complete.




Which is the better solution: the Interpolated solution or the Galerkin solution? To answer this, we will solve a problem with a specified source term $f$ for which the exact solution is already known

```python
# weak form
u = ufl.TrialFunction(V)
v = ufl.TestFunction(V)
x = ufl.SpatialCoordinate(msh)
f = 2.0 * ufl.pi**2 *ufl.sin(ufl.pi*x[0]) * ufl.sin(ufl.pi*x[1])
a = ufl.inner(ufl.grad(u), ufl.grad(v)) * ufl.dx
L = f * v * ufl.dx
```

## 1. Source Term ($f$)
The forcing function is defined as:

$$f(x, y) = 2\pi^2 \sin(\pi x) \sin(\pi y)$$

#### 2. Exact Analytical Solution ($u_e$)
The exact solution satisfying $-\Delta u_e = f$ is:

$$u_e(x, y) = \sin(\pi x) \sin(\pi y)$$


```python
problem = LinearProblem(a, L, bcs=[bc],
                        petsc_options_prefix="poisson_",
                        petsc_options={"ksp_type": "preonly", "pc_type": "lu"})
uh = problem.solve()
# exact solution
ue = ufl.sin(ufl.pi*x[0])*ufl.sin(ufl.pi*x[1])
ui = fem.Function(V)
ui.interpolate(lambda x: np.sin(np.pi*x[0])*np.sin(np.pi*x[1]))
```

```python
#ui.interpolate(ue) #will throw error
ue((0.21,0.17)) # will returns value
#uh((0.21,0.17)) #will throw error
```

`ui.interpolate(ue)` fails

**Cause:** `ui.interpolate()` expects a NumPy-based Python callable. `ue` is just a closed-form algebraic formula built from `ufl.sin`, `ufl.pi`, and `x = ufl.SpatialCoordinate(msh)` — not a callable of that form.

`ue((0.21, 0.17))` returns a value

UFL substitutes the tuple in for `x` and evaluates the expression tree numerically. No mesh, cells, or basis functions are involved — it's pure math, so it evaluates like any ordinary function.

`uh((0.21, 0.17))` fails

`uh` is a dolfinx.fem.Function a discrete FE function represented by DOF coefficients tied to a specific mesh. To evaluate it at an arbitrary point, DOLFINx would need to:
1. Find which mesh cell contains `(0.21, 0.17)` (via a bounding-box tree / collision search), then
2. Evaluate the local basis functions on that cell at that point.

```python
dx_hi = ufl.dx(metadata={"quadrature_degree":8})
J_ue_expr = (0.5*ufl.inner(ufl.grad(ue),ufl.grad(ue))-ufl.inner(f,ue))*dx_hi
J_ue_form = dolfinx.fem.form(J_ue_expr)
J_ue_local = dolfinx.fem.assemble_scalar(J_ue_form)
J_ue = msh.comm.allreduce(J_ue_local,op=MPI.SUM)
print(f"J_ue = {J_ue:.16f}")
```

 Computing the Reference Energy $J(u_e)$ with High-Order Quadrature


**Custom quadrature measure.** `dx_hi` overrides DOLFINx's automatic quadrature-degree estimation with a fixed degree of 8. This is necessary because `ue` contains `ufl.sin`, a transcendental function UFL cannot infer an exact polynomial degree for it, so the default rule may under-integrate the expression. Since $J(u_e)$ serves as our ground-truth reference, we use a higher-order rule to keep quadrature error negligible.

**Energy functional.** `J_ue_expr` encodes the variational energy

$$
J(u_e) = \int_\Omega \left(\tfrac{1}{2}|\nabla u_e|^2 - f\,u_e\right)\,dx,
$$

integrated using the high-precision measure `dx_hi`.


**Output.** The result is printed to 16 decimal places, giving a high-precision benchmark against which $J(u_h)$ and $J(u_i)$ are later compared.

J_ue = -2.4674011002723417


```python
J_uh_expr = (0.5*ufl.inner(ufl.grad(uh),ufl.grad(uh))-ufl.inner(f,uh))*dx_hi
J_uh_form = dolfinx.fem.form(J_uh_expr)
J_uh_local = dolfinx.fem.assemble_scalar(J_uh_form)
J_uh = msh.comm.allreduce(J_uh_local,op=MPI.SUM)
print(f"J_uh = {J_uh:.16f}")
```

Computing the Galerkin Energy $J(u_h)$

**Energy functional for the FEM solution.** This mirrors the $J(u_e)$ computation, but replaces the exact solution with the **Galerkin (FEM) solution** $u_h$:

$$
J(u_h) = \int_\Omega \left(\tfrac{1}{2}|\nabla u_h|^2 - f\,u_h\right)\,dx
$$

Since $u_h$ is a piecewise-polynomial finite element function (not transcendental like $u_e$), the same high-order measure `dx_hi` is reused.

**Output.** The result is printed to 16 decimal places for direct comparison against $J(u_e)$ and $J(u_i)$.

J_uh = -2.3741762216685998

```python
J_ui_expr = (0.5*ufl.inner(ufl.grad(ui),ufl.grad(ui))-ufl.inner(f,ui))*dx_hi
J_ui_form = dolfinx.fem.form(J_ui_expr)
J_ui_local = dolfinx.fem.assemble_scalar(J_ui_form)
J_ui = msh.comm.allreduce(J_ui_local,op=MPI.SUM)
print(f"J_ui = {J_ui:.16f}")

```

Computing the Interpolant Energy $J(u_i)$

**Energy functional for the interpolant.** This repeats the same pattern, now with $u_i$ the **nodal interpolant** of the exact solution $u_e$ onto the finite element space $V_h$:

$$
J(u_i) = \int_\Omega \left(\tfrac{1}{2}|\nabla u_i|^2 - f\,u_i\right)\,dx
$$

Like $u_h$, $u_i$ is a piecewise-polynomial FE function, so `dx_hi` is reused here mainly for consistency across all three energy computations.


**Output.** The result is printed to 16 decimal places, completing the trio of energy values needed for comparison.

 $u_i$ agrees with $u_e$ exactly at mesh nodes but is *not* the energy minimizer over $V_h$ only $u_h$ has that property. So even though $u_i$ interpolates the true solution pointwise, the Ritz principle guarantees $J(u_h)$ lies closer to $J(u_e)$ than $J(u_i)$ does, which is exactly what the printed values confirm.

 J_ui = -2.3737293519143767
