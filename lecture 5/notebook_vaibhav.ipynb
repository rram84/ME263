## **1. Weak Form and Energy Minimization**

From the previous lecture, we have the weak form

$$
\boxed{
a(u,v)=L(v)
\qquad \forall v\in V
}
$$

and the corresponding energy functional

$$
\boxed{
J(u)=\frac{1}{2}a(u,u)-L(u).
}
$$

The objective of this lecture is to establish the equivalence

$$
\boxed{
a(u,v)=L(v)\quad\forall v\in V
\quad\Longleftrightarrow\quad
u=\operatorname*{arg\,min}_{w\in V}J(w).
}
$$

In other words,

$$
\boxed{
\text{Weak form}
\quad\Longleftrightarrow\quad
\text{Energy minimization problem}.
}
$$

To establish this result, we first recall some properties of the bilinear form
$a(\cdot,\cdot)$ and its associated norm.

---

# **2. The Energy Norm**

Since $a(\cdot,\cdot)$ is an inner product, we can define

$$
\boxed{
\|u\|_a=\sqrt{a(u,u)}.
}
$$

Therefore,

$$
a(u,u)=\|u\|_a^2.
$$

For the Poisson problem,

$$
a(u,u)
=
\int_\Omega |\nabla u|^2\,d\Omega,
$$

so that

$$
\|u\|_a^2
=
\int_\Omega |\nabla u|^2\,d\Omega.
$$

We will use the usual properties of this norm.

---

## **3. Cauchy--Schwarz Inequality**

For the inner product $a(\cdot,\cdot)$,

$$
\boxed{
|a(u,v)|
\leq
\|u\|_a\|v\|_a.
}
$$

To see this, consider

$$
\|u-tv\|_a^2\geq0
$$

for an arbitrary scalar $t$.

Expanding,

$$
\begin{aligned}
\|u-tv\|_a^2
&=
a(u-tv,u-tv)\\
&=
\|u\|_a^2
-2t\,a(u,v)
+t^2\|v\|_a^2.
\end{aligned}
$$

This is a quadratic polynomial in $t$ which must be non-negative for every $t$.

Therefore, its discriminant must satisfy

$$
\left[-2a(u,v)\right]^2
-
4\|u\|_a^2\|v\|_a^2
\leq0.
$$

Hence,

$$
a(u,v)^2
\leq
\|u\|_a^2\|v\|_a^2,
$$

and therefore

$$
\boxed{
|a(u,v)|
\leq
\|u\|_a\|v\|_a.
}
$$

---

# **4. Triangle Inequality**

The norm also satisfies

$$
\boxed{
\|u+v\|_a
\leq
\|u\|_a+\|v\|_a.
}
$$

To see this, start with

$$
\begin{aligned}
\|u+v\|_a^2
&=
a(u+v,u+v)\\
&=
\|u\|_a^2
+2a(u,v)
+\|v\|_a^2.
\end{aligned}
$$

Using Cauchy--Schwarz,

$$
a(u,v)
\leq
|a(u,v)|
\leq
\|u\|_a\|v\|_a.
$$

Therefore,

$$
\begin{aligned}
\|u+v\|_a^2
&\leq
\|u\|_a^2
+2\|u\|_a\|v\|_a
+\|v\|_a^2\\
&=
\left(\|u\|_a+\|v\|_a\right)^2.
\end{aligned}
$$

Taking the square root gives

$$
\boxed{
\|u+v\|_a
\leq
\|u\|_a+\|v\|_a.
}
$$

---

# **5. Weak Form $\Longrightarrow$ Energy Minimization**

We first assume that $u$ satisfies the weak form:

$$
\boxed{
a(u,v)=L(v)
\qquad\forall v\in V.
}
$$

We want to show that $u$ minimizes $J$.

Consider a perturbation of $u$ in an arbitrary direction $v$:

$$
u+\epsilon v,
$$

where $\epsilon$ is a scalar.

By definition,

$$
J(u+\epsilon v)
=
\frac12a(u+\epsilon v,u+\epsilon v)
-
L(u+\epsilon v).
$$

Using bilinearity of $a$ and linearity of $L$,

$$
\begin{aligned}
J(u+\epsilon v)
&=
\frac12
\left[
a(u,u)
+2\epsilon a(u,v)
+\epsilon^2a(v,v)
\right]\\
&\qquad
-L(u)-\epsilon L(v).
\end{aligned}
$$

Since

$$
a(u,u)=\|u\|_a^2,
\qquad
a(v,v)=\|v\|_a^2,
$$

we obtain

$$
J(u+\epsilon v)
=
J(u)
+
\epsilon[a(u,v)-L(v)]
+
\epsilon^2\|v\|_a^2.
$$

Thus,

$$
\boxed{
J(u+\epsilon v)
=
J(u)
+
\epsilon[a(u,v)-L(v)]
+
\epsilon^2\|v\|_a^2.
}
$$

Now, because $u$ satisfies the weak form,

$$
a(u,v)-L(v)=0.
$$

Therefore,

$$
\boxed{
J(u+\epsilon v)
=
J(u)+\epsilon^2\|v\|_a^2.
}
$$

Since

$$
\epsilon^2\|v\|_a^2\geq0,
$$

we have

$$
\boxed{
J(u+\epsilon v)\geq J(u).
}
$$

Hence, $u$ is a minimizer of $J$.

---

# **6. Why the Linear Term Must Vanish**

The previous result can also be understood by defining

$$
\phi(\epsilon)=J(u+\epsilon v).
$$

From the expansion,

$$
\phi(\epsilon)
=
J(u)
+
\epsilon[a(u,v)-L(v)]
+
\epsilon^2\|v\|_a^2.
$$

The graph of $\phi(\epsilon)$ is a parabola because the coefficient of $\epsilon^2$ is non-negative.

If $u$ is a minimum, then $\epsilon=0$ must be the minimum of $\phi$.

Therefore,

$$
\phi'(0)=0.
$$

But

$$
\phi'(\epsilon)
=
a(u,v)-L(v)
+
2\epsilon\|v\|_a^2.
$$

Hence,

$$
\phi'(0)
=
a(u,v)-L(v)=0.
$$

Thus,

$$
\boxed{
a(u,v)=L(v).
}
$$

This is precisely the weak form.

---

# **7. Energy Minimization $\Longrightarrow$ Weak Form**

We now prove the converse.

Suppose $u$ is a minimizer of $J$:

$$
\boxed{
J(u)\leq J(w)
\qquad\forall w\in V.
}
$$

Consider again

$$
\phi(\epsilon)=J(u+\epsilon v).
$$

Since $u$ is a minimizer, $\phi(\epsilon)$ must have a minimum at

$$
\epsilon=0.
$$

Therefore,

$$
\phi'(0)=0.
$$

From the previous calculation,

$$
\phi'(0)=a(u,v)-L(v).
$$

Hence,

$$
a(u,v)-L(v)=0.
$$

Therefore,

$$
\boxed{
a(u,v)=L(v)
\qquad\forall v\in V.
}
$$

Thus, the minimizer of $J$ satisfies the weak form.

---

# **8. Converse Using an Arbitrary $w$**

There is another useful way to see the minimization result.

Suppose $u$ satisfies the weak form, and let $w\in V$ be arbitrary.

Choose

$$
v=w-u
$$

and

$$
\epsilon=1.
$$

Then

$$
u+\epsilon v
=
u+(w-u)
=
w.
$$

From the expansion,

$$
J(w)
=
J(u)
+
[a(u,w-u)-L(w-u)]
+
\|w-u\|_a^2.
$$

Since $u$ satisfies the weak form,

$$
a(u,w-u)=L(w-u).
$$

Therefore,

$$
\boxed{
J(w)=J(u)+\|w-u\|_a^2.
}
$$

Since

$$
\|w-u\|_a^2\geq0,
$$

we obtain

$$
\boxed{
J(w)\geq J(u)
\qquad\forall w\in V.
}
$$

Therefore, $u$ minimizes $J$.

---

# **9. Uniqueness**

The previous equation also gives uniqueness of the minimizer.

Suppose $u$ and $\tilde u$ are both minimizers.

Then,

$$
J(\tilde u)
=
J(u)+\|\tilde u-u\|_a^2.
$$

Since both are minimizers,

$$
J(\tilde u)=J(u).
$$

Therefore,

$$
\|\tilde u-u\|_a^2=0.
$$

Hence,

$$
\boxed{
\tilde u=u.
}
$$

Thus, the minimizer is unique.

---

# **10. Final Result**

We have therefore established the equivalence

$$
\boxed{
a(u,v)=L(v)
\quad\forall v\in V
}
$$

if and only if

$$
\boxed{
u=\operatorname*{arg\,min}_{w\in V}
\left[
\frac12a(w,w)-L(w)
\right].
}
$$

Therefore,

$$
\boxed{
\text{Weak formulation}
\quad\Longleftrightarrow\quad
\text{Energy minimization formulation}.
}
$$

Furthermore, if $u$ is the weak-form solution, then for any $w\in V$,

$$
\boxed{
J(w)-J(u)=\|w-u\|_a^2.
}
$$

Thus, the energy functional not only has its minimum at the weak-form solution, but its difference from the minimum is exactly the squared distance from the solution in the energy norm.
