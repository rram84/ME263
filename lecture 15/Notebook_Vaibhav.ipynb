## Neo-Hookean Material: Derivation of the Nonlinear FEM Residual

In the last lecture, we saw that the equilibrium condition for a potential-energy functional can be written in terms of its directional derivative as

$$
D\Pi(u)[v]
=
\int_{\Omega_0} P(F(u)):\nabla v\,dV_0
=
0
\qquad \forall v,
$$

where $P$ is the first Piola--Kirchhoff stress tensor.

In this lecture, we obtain $P$ for a **Neo-Hookean material** and then use it to construct the nonlinear finite-element residual.


## 1. Neo-Hookean strain-energy density

For the 3D Neo-Hookean material considered here, the strain-energy density is

$$
\boxed{
W(F)
=
\frac{\mu}{2}(I_C-3)
-\mu\log J
+
\frac{\lambda}{2}(\log J)^2
}
$$

where

$$
C=F^T F,
$$

$$
I_C=\operatorname{tr}(C)
=\operatorname{tr}(F^TF),
$$

and

$$
J=\det(F).
$$

Here, $F$ is the deformation gradient and $\lambda,\mu$ are material parameters.

We now differentiate each term in $W$ with respect to $F$ separately.


## 2. Derivative of $I_C$

We first consider

$$
I_C=\operatorname{tr}(F^TF).
$$

To calculate its derivative, perturb $F$ in the direction $H$:

$$
F\rightarrow F+\varepsilon H.
$$

Then

$$
(F+\varepsilon H)^T(F+\varepsilon H)
=
F^TF
+
\varepsilon(F^TH+H^TF)
+
\varepsilon^2H^TH.
$$

Therefore,

$$
\begin{aligned}
\frac{d}{d\varepsilon}
\operatorname{tr}\left[
(F+\varepsilon H)^T(F+\varepsilon H)
\right]
\bigg|_{\varepsilon=0}
&=
\operatorname{tr}(F^TH+H^TF)\\
&=2F:H.
\end{aligned}
$$

Hence,

$$
\boxed{
\frac{\partial I_C}{\partial F}=2F
}
$$


## 3. Derivative of the first term in $W$

Consider

$$
W_1(F)=\frac{\mu}{2}(I_C-3).
$$

Using

$$
\frac{\partial I_C}{\partial F}=2F,
$$

we obtain

$$
\frac{\partial W_1}{\partial F}
=
\frac{\mu}{2}(2F).
$$

Thus,

$$
\boxed{
\frac{\partial W_1}{\partial F}=\mu F
}
$$


## 4. Derivative of $J=\det(F)$

Next, consider

$$
J=\det(F).
$$

Again perturb $F$:

$$
F\rightarrow F+\varepsilon H.
$$

We write

$$
F+\varepsilon H
=
F\left(I+\varepsilon F^{-1}H\right).
$$

Taking the determinant,

$$
\det(F+\varepsilon H)
=
J\det\left(I+\varepsilon F^{-1}H\right).
$$

For small $\varepsilon$,

$$
\det(I+\varepsilon A)
=
1+\varepsilon\operatorname{tr}(A)
+O(\varepsilon^2).
$$

Therefore,

$$
\det(F+\varepsilon H)
=
J
\left[
1+\varepsilon\operatorname{tr}(F^{-1}H)
+O(\varepsilon^2)
\right].
$$

Hence,

$$
\left.
\frac{d}{d\varepsilon}\det(F+\varepsilon H)
\right|_{\varepsilon=0}
=
J\operatorname{tr}(F^{-1}H).
$$

Using

$$
\operatorname{tr}(F^{-1}H)
=
F^{-T}:H,
$$

we get

$$
\boxed{
\frac{\partial J}{\partial F}
=
JF^{-T}
}
$$


## 5. Derivative of $\log J$

Using the result above,

$$
\frac{\partial\log J}{\partial F}
=
\frac{1}{J}
\frac{\partial J}{\partial F}.
$$

Since

$$
\frac{\partial J}{\partial F}=JF^{-T},
$$

we obtain

$$
\boxed{
\frac{\partial\log J}{\partial F}
=
F^{-T}
}
$$


## 6. Derivative of the second term in $W$

The second term in the Neo-Hookean energy is

$$
W_2(F)=-\mu\log J.
$$

Therefore,

$$
\frac{\partial W_2}{\partial F}
=
-\mu
\frac{\partial\log J}{\partial F}.
$$

Using the result from above,

$$
\boxed{
\frac{\partial W_2}{\partial F}
=
-\mu F^{-T}
}
$$


## 7. Derivative of the volumetric term

The final term is

$$
W_3(F)
=
\frac{\lambda}{2}(\log J)^2.
$$

Applying the chain rule,

$$
\frac{\partial W_3}{\partial F}
=
\lambda\log J
\frac{\partial\log J}{\partial F}.
$$

Since

$$
\frac{\partial\log J}{\partial F}=F^{-T},
$$

we obtain

$$
\boxed{
\frac{\partial W_3}{\partial F}
=
\lambda\log J\,F^{-T}
}
$$


## 8. First Piola--Kirchhoff stress

The first Piola--Kirchhoff stress tensor is obtained from the strain-energy density as

$$
\boxed{
P=\frac{\partial W}{\partial F}.
}
$$

Since

$$
W=W_1+W_2+W_3,
$$

we have

$$
P
=
\frac{\partial W_1}{\partial F}
+
\frac{\partial W_2}{\partial F}
+
\frac{\partial W_3}{\partial F}.
$$

Substituting the three results,

$$
P
=
\mu F
-\mu F^{-T}
+\lambda\log J\,F^{-T}.
$$

Therefore,

$$
\boxed{
P
=
\mu F
+
(\lambda\log J-\mu)F^{-T}
}
$$

This is the constitutive relation for the Neo-Hookean material used in the weak form.


## 9. Finite-element equations

Let $\{\phi_i\}_{i=1}^N$ denote the finite-element basis functions.

Since the residual is linear in its second argument, it is sufficient to impose

$$
R(u_h;\phi_i)=0,
\qquad i=1,\ldots,N.
$$

Define

$$
\boxed{
R_i(u_h)=R(u_h;\phi_i).
}
$$

Then the nonlinear finite-element problem becomes

$$
\boxed{
R_i(u_h)=0,
\qquad i=1,\ldots,N.
}
$$

or, collectively,

$$
\boxed{
\mathbf R(\mathbf u)=\mathbf 0.
}
$$

This is the nonlinear algebraic system that must be solved for the unknown nodal displacement degrees of freedom.


## 10. Residual in FEniCSx

In FEniCSx, the residual can be obtained directly by differentiating the potential energy with respect to the displacement in the direction of the test function.

Schematically,

```python
Pi = ...
Res = ufl.derivative(Pi, u_h, v)
```

where:

- `u_h` is the finite-element displacement,
- `v` is the test function,
- `Pi` is the total potential energy.

Thus,

$$
\boxed{
\texttt{Res}=D\Pi(u_h)[v].
}
$$

For the Neo-Hookean model, the energy is constructed from

$$
F=I+\nabla u,
\qquad
J=\det F,
\qquad
I_C=\operatorname{tr}(F^TF).
$$


## 11. Taylor test

The directional derivative can be verified using a **Taylor test**. The Taylor test is not required when the residual is obtained directly using `ufl.derivative`, since UFL performs the differentiation automatically. Instead, it is useful as an independent check when the residual or directional derivative is derived and implemented manually.

For a small perturbation $\varepsilon v$, the Taylor expansion of the potential energy is

$$
\Pi(u+\varepsilon v)
=
\Pi(u)
+
\varepsilon D\Pi(u)[v]
+
O(\varepsilon^2).
$$

Since

$$
R(u;v)=D\Pi(u)[v],
$$

we have

$$
\Pi(u+\varepsilon v)-\Pi(u)
=
\varepsilon R(u;v)
+
O(\varepsilon^2).
$$

Therefore,

$$
\boxed{
\frac{\Pi(u+\varepsilon v)-\Pi(u)}{\varepsilon}
=
R(u;v)+O(\varepsilon).
}
$$

Thus, for a randomly chosen admissible direction $v$, the finite-difference approximation

$$
\frac{\Pi(u+\varepsilon v)-\Pi(u)}{\varepsilon}
$$

should approach the residual

$$
R(u;v)
$$

as $\varepsilon$ becomes small.

The Taylor test therefore checks the **consistency between the implemented potential energy and its directional derivative**. It is particularly useful when the residual is coded independently rather than generated automatically by UFL.

## 12. Taylor test implementation in FEniCSx

We now implement the Taylor test for the Neo-Hookean energy using FEniCSx.

The procedure is:

1. Define a finite-element space and a state $u_h$.
2. Define the Neo-Hookean energy $\Pi(u_h)$.
3. Obtain the residual using `ufl.derivative`.
4. Choose a random perturbation $v$.
5. Evaluate the energy at $u_h$ and at $u_h+h v$.
6. Compare the actual change in energy with the first-order prediction
   $hR(u_h;v)$.

The Taylor remainder is

$$
\boxed{
e(h)
=
\left|
\Pi(u_h+h v)
-
\Pi(u_h)
-
hR(u_h;v)
\right|.
}
$$


````python
from mpi4py import MPI
from dolfinx import fem, mesh
from dolfinx.fem import petsc
from petsc4py import PETSc
import ufl
import numpy as np
````

````python
N = 4
msh = mesh.create_unit_square(MPI.COMM_WORLD, N, N)
````

````python

V = fem.functionspace(msh, ("Lagrange", 1, (2,)))

v = ufl.Argument(V,0)
du = ufl.Argument(V,1)

# Plane-strain Neo-Hookean material
E, nu = 1.0, 0.3
mu = E / (2 * (1 + nu))
lmbda = E * nu / ((1 + nu) * (1 - 2 * nu))
````

````python
uh = fem.Function(V)                    # the state we linearise about

F = ufl.Identity(2) + ufl.grad(uh)
J = ufl.det(F)
Ic = ufl.tr(F.T * F) + 1                #plane strain: +1 (z-stretch)

W = (
    (mu / 2) * (Ic - 3)
    - mu * ufl.ln(J)
    + (lmbda / 2) * ufl.ln(J)**2
)

Pi = W * ufl.dx                         #internal energy (a conservative load drops out of the checks)

Res = ufl.derivative(Pi, uh, v)
Jac = ufl.derivative(Res, uh, du)
````

````python
Pi_form = fem.form(Pi)
Res_form = fem.form(Res)
````

### 12.5 Choose a specific state

To test the derivative, we first assign a known, nonzero displacement field to $u_h$.

Here we use

$$
u_1=0.15x^2y,
\qquad
u_2=0.1y^2.
$$

This gives a nontrivial deformation state about which the derivative is tested.


````python
uh.interpolate(
    lambda x: np.vstack((
        0.15 * x[0]**2 * x[1],
        0.1 * x[1]**2
    ))
)
````

````python
rng = np.random.default_rng(0)

v = fem.Function(V)
v.x.array[:] = rng.standard_normal(v.x.array.size)
````

### 12.7 Assemble the energy and residual

We first evaluate the potential energy at the unperturbed state,

$$
\Pi_0=\Pi(u_h).
$$

The residual vector is then assembled from the residual form. To obtain the directional derivative in the particular random direction $v$, we take its inner product with the assembled residual:

$$
R(u_h;v)
=
\mathbf R(u_h)\cdot\mathbf v.
$$

This is the finite-dimensional equivalent of evaluating the residual functional with the chosen test direction $v$.


````python
def energy():
    local_energy = fem.assemble_scalar(Pi_form)
    return msh.comm.allreduce(local_energy, op=MPI.SUM)

Pi_0 = energy()
````

````python

R = petsc.assemble_vector(Res_form)
# R(uh; v)
vvec = v.x.petsc_vec
R_v = R.dot(v.x.petsc_vec)
u0 = uh.x.array.copy()

h = 1e-2                            #small perturbation

uh.x.array[:] = u0 + h * v.x.array
uh.x.scatter_forward()

Pi_pert = energy()

diff = abs(Pi_pert - Pi_0 - h * R_v)
print(diff)
````

````python
h = 1e-3                            #small perturbation

uh.x.array[:] = u0 + h * v.x.array
uh.x.scatter_forward()

Pi_pert = energy()

diff = abs(Pi_pert - Pi_0 - h * R_v)
print(diff)
````

### Conclusion

When $h$ was reduced by a factor of about $10$, the Taylor-test residual decreased by a factor of approximately $100$.

This is the expected behaviour for this test, since the directional derivative was obtained directly using the UFL implementation:

Res = ufl.derivative(Pi, uh, v)
