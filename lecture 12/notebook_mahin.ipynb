## Post processing solution

Previous class discussed about how to get back the reaction force on the deflection boundaries. For finding reaction force on dirichlet boundary we earlier used
$$F_{naive} = \int \sigma(u_h).n ds$$
where uh is the finite element solution.
But the FE solution of reaction force (Fnaive) has big error even though error in displacement is quite small.
Naive way of calculating moment is given as
$$M_{naive} = \int r\times(\sigma(u_h).n)ds$$
Even though displacements can be very accurate (ie $u_h \approx u$), the stresses involve derivatives of $u_h$ and strain contains displacement gradients $ \nabla u_h$.  
In energy norm 
$$||u - u_h||_E \sim O(h)$$
$$||u - u_h||_E \le Ch||u||_E$$
$$||u - u_h||_E \le C(u) h$$
energy norm converges linearly.

the displacements converges faster in the order of $O(h^2)$
$$||u - u_h||_{L2} \le Ch^2 ||u||_{L2}$$
$$||u - u_h||_{L2} \le C(u)h^2$$
$||u - u_h||_E$ converges only by h comes from best approximation
$$
||u -u_h||_E \le ||u -v_h||_E \; \forall\; v_h\;\in \mathcal V_h
$$ 
$v_h = I_h u$ $\to$ interpolant of u  
$||u - I_h u||_E \to$ basically is a taylor series 

$$F - F_{naive} = \int ((\sigma(u) - \sigma(u_h))n)ds \sim O(h)$$
Force converges only as h.

consider the equilibrium equation 
$$div(\sigma) + f = 0$$
multiplying by a test function $v$ and integrating by parts we get
$$ \int_{\Omega} \sigma(u) : \varepsilon(v) = \int_{\Omega} f \cdot v + \int_{\Gamma_N} t \cdot v + \int_{\Gamma_D} (\sigma(u)n) \cdot v \quad   \forall v \in \mathcal V_h $$
We need to compute the part $\int_{\Gamma_D} (\sigma(u)n) \cdot v \quad$ from the above expression.  
If we take the test function $v$ as $e_x$ and $e_y$ we will get the correponding forces in x and y directions. (ie $F_x$ and $F_y$)  
The source of absolute error being very high is that $u_h$ is never tested on the dirichlet boundary $\Gamma_D$ part since the test function vanishes on the $\Gamma_D$. Equilibrium is satisfied only in the neumann part $\Gamma_N$ of boundary while in the dirichlet part $\Gamma_D$ equilibrium is not satisfied. Dirichlet part just satisfies a constraint. The other reason is $u_h$ and its derivative is piece wise smooth therefore some amount of error is expected.  
- On $\Gamma_N$ (Neumann part): equilibrium is satisfied, because the traction there enters the
  weak form directly and test functions are nonzero on $\Gamma_N$.
- On $\Gamma_D$ (Dirichlet part): equilibrium is not satisfied by construction -- the boundary is
  only satisfying a constraint (the prescribed displacement), not a force balance.


## Recovering reaction force components
To extract the reaction force in a fixed direction (e.g. $e_x$), the trick is to plug in $v = e_x$ into the weak form even though $e_x \notin V_h$. Thus $e_x$ does not vanish on $\Gamma_D$.  
- consider $v = e_x$  
A rigid-body translation $v = \mathbf e_x$ produces zero strain for the exact stress field:

$$
\int_\Omega \sigma(u_h):\varepsilon(\mathbf e_x)\, d\Omega = 0
\qquad\text{(any rigid-body mode has } \varepsilon(\mathbf e_x) = 0\text{).}
$$  
$$
\int_\Omega \sigma(u_h):\varepsilon(\mathbf e_x)\, d\Omega
= \int_\Omega R\cdot \mathbf e_x\, d\Omega + \int_{\Gamma_N} \bar t\cdot \mathbf e_x \, ds
+ \int_{\Gamma_D} \big(\sigma(u_h) n\big)\cdot \mathbf e_x\, ds .
$$

Since the left-hand side is $0$, rearrange to isolate the unknown boundary
traction on $\Gamma_D$:

$$
\int_{\Gamma_D} \big(\sigma(u_h) n\big)\cdot \mathbf e_x \, ds
= -\int_\Omega f\cdot \mathbf e_x\, d\Omega \;-\; \int_{\Gamma_N} \bar t \cdot \mathbf e_x\, ds \;=\;l(e_x)
$$  
Thus in general the force can be evaluated as 
$$
\boxed{\;F_n \;=\; -\int_\Omega f\cdot \mathbf e_n \, d\Omega \;-\; \int_{\Gamma_N} \bar t\cdot \mathbf e_n\, ds\;}
$$

**Physical interpretation:** $F_n$ is exactly whatever remaining force is needed to satisfy global
equilibrium -- i.e. it balances the applied body force and the applied Neumann traction.

By choosing $v = \mathbf e_x$ we get $F_x$; by the identical argument with $v=\mathbf e_y$ we get
$F_y$.

## Recovering the reaction moment
The reaction moment about the origin, from the traction distribution on $\Gamma_D$, is

$$
M = \int_{\Gamma_D} (x,y)\times(\sigma\cdot n)\, ds
= \int_{\Gamma_D}\big(-y\,t_D^x + x\,t_D^y\big)\, ds ,
$$

where $t_D = \sigma(u_h)\,n$ is the (recovered) traction on the Dirichlet boundary.

Using exactly the same argument as for the force components, choose the test function to be the
rigid-body rotation mode

$$
v = (-y,\, x),
$$

which again produces zero strain energy ($\varepsilon(v) = 0$ for a rigid rotation), so

$$
\boxed{\;M \;=\; -\int_\Omega f\cdot v\,d\Omega \;-\; \int_{\Gamma_N}\bar t\cdot v\,d\Gamma\;}
\qquad v=(-y,x).
$$

So $F_x$, $F_y$, and $M$ are all obtained the *same* way -- plug a rigid-body mode (translation or rotation) into the discrete weak form as if it were an admissible test function and use the fact that a rigid-body mode has zero strain energy.

## Why strain energy converges faster?
Energy converges in the order of $h^2$
$$ |E(u) - E(u_h)| \sim h^2 $$
For kth order polynomial   $|E(u) - E(u_h)| \sim h^2k$
- How come energy is so much accurate?  
Define the strain energy functional (for simplicity not considering the 1/2 term):
$$
E(v) = \tfrac12\int_\Omega \sigma(v):\varepsilon(v)\, d\Omega \;=\; a(v,v)
$$  
$$ 
E(v) \sim a(v,v)
$$
so for both the exact solution $u$ and the FEM solution $u_h$:
$$
E(u) = a(u,u), \qquad E(u_h) = a(u_h,u_h).
$$
Since $u$ is the exact solution, it satisfies
$a(u,v) = \ell(v)$ for every admissible $v$.

$$
E(u) - E(u_h) = a(u,u) - a(u_h,u_h) = \ell(u) - \ell(u_h) 
$$
because $u$ is the exact solution, $a(u,u)=\ell(u)$ and $a(u_h,u_h)=\ell(u_h)$ by the discrete equilibrium.
$$
E(u) - E(u_h) = \ell(u-u_h).
$$

use equilibrium to swap $\ell$ for $a$ again. Because of equilibrium we can go from
$a(u,u)\to \ell(u)$ and back, $\ell(u-u_h) \to a(u-u_h,\,u-u_h)$:

$$
E(u)-E(u_h) = a(u-u_h,\, u-u_h) = \|u-u_h\|_E^2 ,
$$

where $\|\cdot\|_E$ is the energy norm, $\|w\|_E^2 = a(w,w)$.

For $k$-th order Lagrange finite elements, the standard a priori estimate for the energy-norm error is

$$
\|u-u_h\|_E \sim h^{k}
$$

Squaring gives the result:

$$
\boxed{\;|E(u)-E(u_h)| = \|u-u_h\|_E^2 \sim h^{2k}\;}
$$

Why this matters -- the displacement error itself only decays as $h^k$ in the energy norm, but because the energy error is exactly the square of the energy norm of the error, it inherits an automatic squaring -- it converges at twice the rate of the displacement error. This is a direct consequence of Galerkin orthogonality (the FEM solution is
the energy-orthogonal projection of $u$ onto $\mathcal V_h$), not an extra numerical coincidence.

## Convergence of linear functionals
linear functionals also converges with $2k$.  
Consider the functinal $J(v) \rightarrow R$  
$$
|J(u) - J(u_h)| \sim h^{2k}
$$  
- Example :  $J(u) = \int_\Omega u.e_x d\Omega \;=\;\int_\Omega u_x d\Omega $  
$J(v)$ can be reepresented as taking a dot product of $v$ with some $j$ in the energy norm  
that is $J(v) = a(j,v)$  
$$
\begin{aligned}
J(u) - J(u_h) &= a(j,u) - a(j, u_h) \\
&= a(j, (u-u_h)) \\ 
&\le ||j||_E||u-u_h||_E \\
&= a(j-j_h, u-u_h) + a(j_h, u-u_h) \\
&= a(j-j_h, u-u_h) \\
J(u) - J(u_h) &\le ||j - j_h||_E||u - u_h||_E
\end{aligned}
$$
$$
||j - j_h||_E \sim h^k \;and \;||u - u_h||_E \sim h^k
$$  
$$
J(u) - J(u_h) \sim h^{2k}  
$$
