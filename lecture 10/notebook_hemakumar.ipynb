### Korn's 1st Inequality and Rigid Body Modes

#### Energy Functional

Consider the functional

$$
J(v) = \frac{1}{2}a(v,v) - L(v)
$$

where

- $a(v,v)$ represents the **strain energy**,
- $L(v)$ represents the **external work done**.

For the problem to have a unique solution, we require appropriate properties of the bilinear form $a(\cdot,\cdot)$.

In particular,

$$
a(v,v) \geq 0
$$

and

$$
a(v,v)=0 \iff v=0
$$

However, for elasticity problems, this condition requires some care because **rigid body motions** can have zero strain energy.

---

#### Cea's Lemma

Let's consider Cea's Lemma for Laplace problem

$$\|{}u-u_{h}\|{}_{H^{1}(\Omega )}\le \frac{M}{\alpha }\inf _{v_{h}\in V_{h}}\|{}u-v_{h}\|{}_{H^{1}}$$

where M is the boundedness constant, α is the coercivity constant of the bilinear form, and H¹ is the energy norm.

This tells us that the FEA solution minimizes the strain energy error. The error we get is bounded by the interpolation error of the mesh geometry.

We need to show that 

$$a(v,v)\geq \alpha\|v\|_{H^1}^2$$

and prove that any reasonable displacement is related to reasonable amount of energy.

---
#### Linear Strain Tensor

$a(v,v)$ knows only the symmetric part of the strain tensor $(\operatorname{sym}(\varepsilon)$ or $\operatorname{sym}(\nabla v)
)$, which means we only know 3 independent components.

Therefore, for a displacement field $v$, the infinitesimal strain tensor is

$$
\varepsilon(v)
=
\operatorname{sym}(\nabla v)
=
\frac{1}{2}\left(\nabla v+\nabla v^T\right)
$$

In index notation,

$$
\varepsilon_{ij}
=
\frac{1}{2}\left(v_{i,j}+v_{j,i}\right)
$$

For linear elasticity, the bilinear form can be written as

$$
a(v,v)
=
\int_{\Omega}
\left[
\lambda(\operatorname{tr}(\varepsilon))^2
+
2\mu\,\varepsilon:\varepsilon
\right]\,d\Omega
$$

but we know that, trace of strain tensor is the divergence of that strain tensor,

$$
\operatorname{tr}(\varepsilon) = (\nabla\cdot v)

$$

therefore,

$$
a(v,v)
=
\int_{\Omega}
\left[
\lambda(\nabla\cdot v)^2
+
2\mu\,\varepsilon:\varepsilon
\right]\,d\Omega
$$

Since

$$
\varepsilon:\varepsilon = |\varepsilon|^2
$$

we have

$$
a(v,v)
=
\int_{\Omega}
\left[
\lambda(\nabla\cdot v)^2
+
2\mu|\varepsilon|^2
\right]\,d\Omega
$$

---

#### Korn's 1st Inequality

To establish a lower bound on the energy, we would like to relate the strain to the displacement gradient.

In particular, we want an inequality of the form

$$
\int_{\Omega}|\varepsilon|^2\,d\Omega
\geq
c\int_{\Omega}|\nabla v|^2\,d\Omega
$$

for some constant $c>0$.

This is related to **Korn's inequality**.

For the case where the displacement vanishes on the boundary,

$$
v=0
\qquad \text{on } \partial\Omega
$$

we can derive the following result.

---

#### Derivation of Korn's Inequality

Starting with

$$
|\varepsilon|^2
=
\varepsilon_{ij}\varepsilon_{ij}
$$

and using

$$
\varepsilon_{ij}
=
\frac{1}{2}(v_{i,j}+v_{j,i})
$$

we obtain

$$
|\varepsilon|^2
=
\frac{1}{4}
(v_{i,j}+v_{j,i})
(v_{i,j}+v_{j,i})
$$

Expanding,

$$
|\varepsilon|^2
=
\frac{1}{4}
\left(
v_{i,j}v_{i,j}
+
v_{i,j}v_{j,i}
+
v_{j,i}v_{i,j}
+
v_{j,i}v_{j,i}
\right)
$$

Since

$$
v_{i,j}v_{j,i}
=
v_{j,i}v_{i,j}
$$

and

$$
v_{j,i}v_{j,i}
=
v_{i,j}v_{i,j}
$$

we obtain

$$
|\varepsilon|^2
=
\frac{1}{2}
\left(
v_{i,j}v_{i,j}
+
v_{i,j}v_{j,i}
\right)
$$

Therefore,

$$
|\varepsilon|^2
=
\frac{1}{2}
\left(
|\nabla v|^2
+
v_{i,j}v_{j,i}
\right)
$$

---

#### Integration by Parts

Consider the second term:

$$
\int_{\Omega}v_{i,j}v_{j,i}\,d\Omega
$$

Using integration by parts,

$$
\int_{\Omega}v_{i,j}v_{j,i}\,d\Omega
=
\int_{\Omega}
(v_{j,i} v_i)_{,j}\,d\Omega
-
\int_{\Omega}v_i v_{j,ij}\,d\Omega
$$

using Gauss Divergence Theorem,

$$
\int_{\Omega}v_{i,j}v_{j,i}\,d\Omega
=
\int_{\partial\Omega}
v_i v_{j,i}n_j\,d\Gamma
-
\int_{\Omega}v_i v_{j,ij}\,d\Omega
$$

Since

$$
v=0
\qquad\text{on }\partial\Omega
$$

the boundary term vanishes.

Hence,

$$
\int_{\Omega}v_{i,j}v_{j,i}\,d\Omega = - \int_{\Omega}v_i v_{j,ij}\,d\Omega
$$

Integrating by parts once again,

$$
-\int_{\Omega}v_i v_{j,ij}\,d\Omega
=
-\int_{\Omega}(v_{i,i}v_j)_{,j}\,d\Omega
+\int_{\Omega}v_{i,i}v_{j,j}\,d\Omega
$$

Since, 
$$v_{i,i} = \nabla\cdot v$$

we get 

$$
-\int_{\Omega}v_i v_{j,ij}\,d\Omega
=
-\int_{\Omega}((\nabla\cdot v) v_j)_{,j}\,d\Omega
+\int_{\Omega}(\nabla\cdot v) (\nabla\cdot v)\,d\Omega
$$

using Gauss Divergence Theorem,

$$
-\int_{\Omega}v_i v_{j,ij}\,d\Omega
=
-\int_{\partial\Omega}(\nabla\cdot v) (v\cdot n)_{,j}\,d\Gamma
+\int_{\Omega}(\nabla\cdot v)^2\,d\Omega
$$

Since

$$
v=0
\qquad\text{on }\partial\Omega
$$

the boundary term vanishes.

Therefore,

$$
-\int_{\Omega}v_i v_{j,ij}\,d\Omega
=
\int_{\Omega}
(\nabla\cdot v)^2\,d\Omega
$$

Since

$$
\int_{\Omega}v_{i,j}v_{j,i}\,d\Omega = - \int_{\Omega}v_i v_{j,ij}\,d\Omega
$$

we get

$$
\boxed{
\int_{\Omega}v_{i,j}v_{j,i}\,d\Omega
=
\int_{\Omega}(\nabla\cdot v)^2\,d\Omega
}
$$

---

#### Final Result

Substituting the above result into the expression for $|\varepsilon|^2$ gives

$$
\int_{\Omega}|\varepsilon|^2\,d\Omega
=
\frac{1}{2}
\left[
\int_{\Omega}|\nabla v|^2\,d\Omega
+
\int_{\Omega}(\nabla\cdot v)^2\,d\Omega
\right]
$$

Here, $(\nabla \cdot v)$ indicates us that strain can see the volumetric part.

Therefore,

$$
\boxed{
\int_{\Omega}|\varepsilon|^2\,d\Omega
\geq
\frac{1}{2}
\int_{\Omega}|\nabla v|^2\,d\Omega
}
$$

or, equivalently,

$$
\boxed{
\|\varepsilon\|_{L^2(\Omega)}^2
\geq
\frac{1}{2}
\|\nabla v\|_{L^2(\Omega)}^2
}
$$

This provides the required lower bound for the strain energy.

**Note:** The strain tensor contains only the symmetric part of the displacement gradient. Korn's inequality allows us to control the full displacement gradient using the strain.

---

#### Coercivity of the Bilinear Form

Recall

$$
a(v,v)
=
\int_{\Omega}
\left[
\lambda(\nabla\cdot v)^2
+
2\mu|\varepsilon|^2
\right]\,d\Omega
$$

Using the previous inequality,

$$
\int_{\Omega}|\varepsilon|^2\,d\Omega
\geq
\frac{1}{2}
\int_{\Omega}|\nabla v|^2\,d\Omega
$$

Therefore,

$$
a(v,v)
\geq
\mu
\int_{\Omega}|\nabla v|^2\,d\Omega
$$

when the remaining terms are non-negative.

Hence, the bilinear form satisfies a coercivity estimate of the form

$$
\boxed{
a(v,v)\geq \alpha\|v\|_{H^1(\Omega)}^2
}
$$

for some $\alpha>0$, under the appropriate boundary conditions.

The physical interpretation is:

Any non-zero displacement that produces strain must require a corresponding amount of strain energy.

---

#### Necessity of the Boundary Conditions 

In the above derivation we used the boudary condition

$$
v=0
\qquad\text{on }\partial\Omega.
$$

This condition was important because it caused the boundary terms arising from integration by parts to vanish.

Without sufficient boundary constraints, there can be non-zero displacement fields for which

$$
\varepsilon(v)=0.
$$

These correspond to **rigid body motions**.

```python
# Let's hand in all the necessary libraries

from mpi4py import MPI
from petsc4py.PETSc import ScalarType
import numpy as np
import ufl

from dolfinx import fem, io, mesh
from dolfinx.fem.petsc import LinearProblem
# 1. CREATE MESH, Function Space, Coordinates

msh = mesh.create_unit_square(
    MPI.COMM_WORLD,
    10,
    10
)

V = fem.functionspace(
    msh,
    ("Lagrange", 1, (2,))
)

u = ufl.TrialFunction(V)
v = ufl.TestFunction(V)

# Spatial coordinates
x = ufl.SpatialCoordinate(msh)
# 4. MATERIAL PROPERTIES
# We define some arbitrary material properties for the problem and derive the Lame parameters from them. 

E = 1.0
nu = 0.3

mu = E / (2.0 * (1.0 + nu))

lmbda = (
    E * nu
    / ((1.0 + nu) * (1.0 - 2.0 * nu))
)
# Let's define a fucntion to compure strain and stress tensors

def epsilon(w):
    return ufl.sym(ufl.grad(w)) # Yeah, you can return ufl functions directly

def sigma(w):
    eps = epsilon(w)
    return (
        lmbda * ufl.tr(eps) * ufl.Identity(2)
        + 2.0 * mu * eps
    )
# Exact Solution so we can manufacture a force and compare the numerical solution to it.


ue = ufl.as_vector((
    ufl.cos(ufl.pi * x[0]) * ufl.cos(ufl.pi * x[1]),
    ufl.sin(ufl.pi * x[0]) * ufl.sin(ufl.pi * x[1])
))
# So we have a body force like so:

f = -ufl.div(sigma(ue))
# Now let's make a weak form

a = (
    ufl.inner(
        sigma(u),
        epsilon(v)
    )
    * ufl.dx
)

L = (
    ufl.dot(f, v)
    * ufl.dx
)
# Let's apply Boundary COnditions

tdim = msh.topology.dim # Chat GPT added this on
print(tdim) # Should come to 2 for a 2D mesh
fdim = tdim - 1

facets = mesh.locate_entities_boundary(
    msh,
    msh.topology.dim - 1,
    marker=lambda x: np.full(x.shape[1], True)
)


dofs = fem.locate_dofs_topological(
    V,
    msh.topology.dim - 1,
    facets
)
```

```python
# Interpolate the exact solution to the function space 

ue_expr = fem.Expression(
    ue,
    V.element.interpolation_points
)

u_bc = fem.Function(V)

u_bc.interpolate(ue_expr)

bc = fem.dirichletbc(u_bc, dofs)
# Solve

problem = LinearProblem(
    a,
    L,
    bcs=[bc],
    petsc_options={
        "ksp_type": "preonly",
        "pc_type": "lu"
    },
    petsc_options_prefix="elasticity"

)

uh = problem.solve()

uh.name = "displacement" #For Paraview stuff

#Quick Check
print("uh name:", uh.name)
print("uh size:", uh.x.array.size)
print("uh min:", uh.x.array.min())
print("uh max:", uh.x.array.max())
```

```python
a_form = fem.form(a)
L_form = fem.form(L)

# No boundary condition version
A_free = fem.petsc.assemble_matrix(a_form)
A_free.assemble()

# With Boundary Condition version
A_bc = fem.petsc.assemble_matrix(a_form, bcs=[bc])
A_bc.assemble()

#print("A_free is symmetric: ", A_free.isSymmetric(1.e-10))
```

```python
# Check the translation in x is a Rigid Body Motion
def make_mode(field):
    r = fem.Function(V)
    r.interpolate(field)
    return r

transx = make_mode(lambda x: np.vstack((np.ones_like(x[0]), np.zeros_like(x[0]))))
#print(type(transx))
rot = make_mode(lambda x: np.vstack([x[1], -x[0]]))

# A_free x transx
res_x_free = A_free.createVecLeft()
A_free.mult(transx.x.petsc_vec, res_x_free)
print("res_x norm: ", res_x_free.norm())

res_x_bc = A_bc.createVecLeft()
A_bc.mult(rot.x.petsc_vec, res_x_bc)
print("res_x norm: ", res_x_bc.norm())
```

```python
k_formed = ufl.action(a, rot)
Arot = fem.petsc.assemble_vector(fem.form(k_formed))
print(Arot.norm())
```
