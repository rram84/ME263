You want to minimize the average temperature of the electronic component:$$\min_{\theta(x),\, \phi(x)} J = T_{\text{hot}} = \frac{1}{\vert{}\omega\vert{}} \int_\omega u(x) \, d\Omega$$ find optimal ($\theta$,$\phi$)

### governing equation
$\mu$ = ($\theta$ , $\phi$)

$$-\nabla \cdot \big(K(\mu) \nabla u\big) = S \quad \text{in } \Omega$$
Dirichlet Boundary (Heat Sink):$$u = 0 \quad \text{on } \Gamma_D$$
Neumann Boundary (Insulated Walls):$$- \big(K(\mu) \nabla u\big) \cdot \mathbf{n}_{\partial\Omega} = 0 \quad$$

### Weak Form
$$a(u, v; \mu) = L(v)$$
$$a(u, v; \mu) = \int_\Omega \nabla v^T K(\mu) \nabla u \, d\Omega$$
$$L(v) = \int_\omega S v \, d\Omega$$
### Discrete form
$$A(\boldsymbol{\mu}) \mathbf{U} = \mathbf{F}$$
Stiffness / Conduction Matrix $A(\boldsymbol{\mu}) \in \mathbb{R}^{N \times N}$ <br>
State / Solution Vector $\mathbf{U} \in \mathbb{R}^N$<br>
Load / Source Vector $\mathbf{F} \in \mathbb{R}^N$<br>

### Objective 
$$Z(\boldsymbol{\mu}) = \int_{\omega} u_h(x) \, d\Omega = \mathbf{l}^T \mathbf{U}$$
$$l_j = \int_\omega \psi_j(x) \, d\Omega$$

$Z(\boldsymbol{\mu})$ is mathematically identical to minimizing the average temperature $T_{\text{hot}}$ 

The computational cost of solving the finite element system $$A(\boldsymbol{\mu}) \mathbf{U} = \mathbf{F}$$increases significantly as the number of degrees of freedom \(N\) grows. For very large systems, such as those with billions of degrees of freedom, assembling and solving the system can require substantial computational time and memory.

For every new parameter $\mu$ we have to the expansive solution Again. Reduced-order modeling (ROM) addresses this problem by constructing a low-dimensional approximation of the solution space using a small number of representative solutions, called snapshots. The reduced model can then provide approximate solutions for different parameter values at a much lower computational cost, while maintaining good accuracy.

### why is it necessary to compute solutions across a large set of parameter values $\boldsymbol{\mu}$

- Real manufactured components rarely match nominal datasheet values due to fabrication defects, boundary contact resistances, or layer misalignments.
- Sensitivity Analysis. Designers must understand how small variations in input parameters
- Quality Control and Uncertainty Quantification

### Reduced-Order Model (ROM) Plan
-   Offline Phase: Solve the full-order problem for selected parameter values and construct the reduced basis, where r << N.
-   Online Phase: For each new parameter value, solve the reduced system to obtain a fast approximate solution.

### Three main goals of Reduced-Order Modeling (ROM)
- Computational efficiency (Cheap)
    - The reduced-order model should be computationally inexpensive to evaluate compared with the original high-fidelity model, particularly when solutions are required for many parameter values $\boldsymbol{\mu}$
- High accuracy (Small error) || $\boldsymbol{\mu}$ - $\boldsymbol{\bar{\mu}}$ || << 1.
- Error estimation and improvement.  || $\boldsymbol{\mu}$ - $\boldsymbol{\bar{\mu}}$ || < $\Delta \boldsymbol{\mu}$

### general plan 
#### online stage
the approximate solution is constructed as a linear combination of \(m\) precomputed basis solutions.
$$\bar{\mathbf{u}}(\boldsymbol{\mu}) = y_1 \mathbf{U}_1 + y_2 \mathbf{U}_2 + \dots + y_m \mathbf{U}_m = \sum_{i=1}^m y_i(\boldsymbol{\mu}) \mathbf{U}_i$$

Here m << n.

If raw snapshots $\mathbf{U}_i$ are used directly without filtering or orthogonalization, they are almost always nearly linearly dependent. This makes the reduced stiffness matrix $A_r$ severely ill-conditioned, leading to numerical instability or complete solver failure.

$$\bar{\mathbf{u}}(\boldsymbol{\mu}) = \sum_{j=1}^m y_j(\boldsymbol{\mu}) \mathbf{V}_j$$

- $\mathbf{V}_j$ (Spatial Basis Vectors / Modes):Parameter-independent: They do not depend on $\boldsymbol{\mu}$.They live in the high-dimensional space $\mathbb{R}^N$
- $y_j(\boldsymbol{\mu})$ (coefficient):Parameter-dependent: They depend only on $\boldsymbol{\mu}$
