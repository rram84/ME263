### 2. Galerkin Approximation

$
\tilde{U}\text{ is a Galerkin approximation of }U
$

Find: $U\in\mathbb{R}^n$

Define a bilinear form:

$$
a(U,W)=l(W),\qquad \forall W\in\mathbb{R}^n
$$

$$
W^TAU=W^TF
$$

#### Claim: $\tilde{U} \text{ is the Galerkin approximation of }U$

Find:

$$
\tilde{U}\in W_r
$$

such that:

$$
a(\tilde{U},W)=l(W),
\qquad \forall W\in W_r
$$

Since:

$$
\tilde{U}\in W_r
$$

$$
\Rightarrow \tilde{U}=Vy;
\qquad \text{for some y.}
$$



Therefore:

$$
a(Vy,W)=l(W),
\qquad \forall W\in W_r
$$

Choose: $W=V_i$

Then:

$$
a(Vy,V_i)=l(V_i)
$$

$$
\Longleftrightarrow
V_i^TAVy=V_i^TF
$$

By definition:

$$
V^TAVy=V^TF
$$

$$
\Rightarrow V^TA\tilde{U}=V^TF
$$

$$
\Rightarrow \tilde{U}
\text{ is the Galerkin approximation of }U
$$

#### Consequence: Orthogonality

We have:

$$
a(U,W)=l(W),
\qquad \forall W\in\mathbb{R}^n
\qquad (\text{also, }   \forall W\in W_r.)

$$



And:

$$
a(\tilde{U},W)=l(W),
\qquad \forall W\in W_r
$$

Subtracting:

$$
a(U-\tilde{U},W)=0,
\qquad \forall W\in W_r
$$

Let:

$$
e=U-\tilde{U}
$$

Then:

$$
\boxed{a(e,W)=0,\qquad \forall W\in W_r}
$$

For the matrix representation:

$$
W^TAe=0
$$

$$
\Rightarrow V^TAe=0
$$

$$\Rightarrow \text{Error e is perpendicular to the basis.}$$

$$
\Rightarrow U-\tilde{U}
\text{ is } \perp_A \text{ to all vectors in }W_r
$$

### 3. Optimality

We want to establish:

$$
a(U-W,U-W)=\|U-W\|_A^2
$$

Consider:

$$
\begin{aligned}
a(U-W,U-W) = a(U-\tilde{U}+\tilde{U}-W, U-\tilde{U}+\tilde{U}-W)
\end{aligned}
$$

Expanding:

$$
\begin{aligned}
&=a(U-\tilde{U},U-\tilde{U})\\
&\quad+a(\tilde{U}-W,\tilde{U}-W)\\
&\quad+2a(U-\tilde{U},\tilde{U}-W)
\end{aligned}
$$

From Galerkin orthogonality:

$$
a(U-\tilde{U},\tilde{U}-W)=0
$$

because:

$$
\tilde{U}-W\in W_r
$$

Therefore:

$$
\begin{aligned}
\|U-W\|_A^2
&=\|U-\tilde{U}\|_A^2\\
&\quad+\|\tilde{U}-W\|_A^2
\end{aligned}
$$

Hence:

$$
\boxed{
\|U-\tilde{U}\|_A^2
\leq
\|U-W\|_A^2
}
$$

$$
\boxed{
\|U-\tilde{U}\|_A
\leq
\|U-W\|_A
}
$$

$$
\therefore\ \tilde{U}
\text{ is the best we can do in a given approximation space.}
$$

#### Projection

$U\rightarrow \tilde{U}$ is a projection.

What is a projection?

If $P$ satisfies $P^2=P$; then $P$ is a Projection.

Closest point operation in one such operation.

From the Galerkin equations:

$$
V^TAVy=V^TF
$$

$$
\Rightarrow y=(V^TAV)^{-1}V^TF
$$

Since:

$$
F=AU
$$

$$
y=(V^TAV)^{-1}V^TAU
$$

Therefore:

$$
\tilde{U}=Vy
$$

$$
\tilde{U}=V(V^TAV)^{-1}V^TAU
$$

Hence:

$$
\boxed{
\tilde{U}
=
\underbrace{V(V^TAV)^{-1}V^TA}_{P_A}
U
}
$$

where:

$$
P_A=V(V^TAV)^{-1}V^TA
$$

$$
P_A:\text{ projection of high fidelity solution onto }W_r
$$

$$
\tilde{U}_{n\times 1}
=
P_A U_{n\times 1}
$$

Transformation of high fidelity solution $U$ into reduced order solution $\tilde{U}$.

It doesn't matter what basis we choose because it depends on the subspace we choose.

### 6. Projection Operator: Idempotency

We should show that:

$$
P^2=P
$$

Consider:

$$
P_A=V(V^TAV)^{-1}V^TA
$$

Then:

$$
\begin{aligned}
P_A^2
={}&
V(V^TAV)^{-1}V^TA\\
&\times V(V^TAV)^{-1}V^TA
\end{aligned}
$$

Rearranging:

$$
\begin{aligned}
P_A^2
={}&V(V^TAV)^{-1}\\
&\times(V^TAV)(V^TAV)^{-1}V^TA
\end{aligned}
$$

Therefore:

$$
\boxed{
P_A^2=V(V^TAV)^{-1}V^TA=P_A
}
$$

### 10. Range Space and Null Space of the Galerkin Projection Operator
Given the projection operator:
$$
P_A=V(V^TAV)^{-1}V^TA
$$
where:
- $V\in\mathbb{R}^{m\times n}$ has full column rank.
- $A\in\mathbb{R}^{m\times m}$ is symmetric positive definite.
- $V^TAV$ is therefore invertible.
#### 1. Range Space of $P_A$
By definition:
$$
R(P_A)=\{P_Ax:x\in\mathbb{R}^m\}
$$
Substituting the expression for $P_A$:
$$
P_Ax=V(V^TAV)^{-1}V^TAx
$$
The result is always a linear combination of the columns of $V$. Therefore:
$$
R(P_A)\subseteq\operatorname{Range}(V)
$$
To establish the reverse inclusion, take any vector in $\operatorname{Range}(V)$, written as $Vy$:
$$
\begin{aligned}
P_A(Vy)
&=V(V^TAV)^{-1}V^TAVy\\
&=V(V^TAV)^{-1}(V^TAV)y\\
&=Vy
\end{aligned}
$$
Thus every vector in $\operatorname{Range}(V)$ is also in the range of $P_A$.
Hence:
$$
\boxed{R(P_A)=\operatorname{Range}(V)}
$$
or equivalently:
$$
\boxed{
R(P_A)=\operatorname{span}\{v_1,v_2,\ldots,v_n\}
}
$$
Physical interpretation: The range space is the reduced-order approximation space onto which the high-fidelity solution is projected.
#### 2. Null Space of $P_A$
By definition:
$$
N(P_A)=\{x\in\mathbb{R}^m:P_Ax=0\}
$$
Substituting:
$$
V(V^TAV)^{-1}V^TAx=0
$$
Since $V$ has full column rank and $V^TAV$ is invertible, the equation implies:
$$
\boxed{V^TAx=0}
$$
Therefore:
$$
\boxed{
N(P_A)=\{x\in\mathbb{R}^m:V^TAx=0\}
}
$$
This can also be written as:
$$
\boxed{N(P_A)=\operatorname{Null}(V^TA)}
$$
Physical interpretation: The null space contains vectors that are completely eliminated by the projection operator.

### 11. Showing $P_A e = 0$
We want to show that:
$$
\boxed{P_Ae=0}
$$
where \(P_A\) is the Galerkin projection operator:
$$
P_A=V(V^TAV)^{-1}V^TA
$$

and the error is:
$$
e=U-\tilde U
$$

Step 1: Use the Galerkin orthogonality condition
From Galerkin approximation, we have:
$$
V^TA(U-\tilde U)=0
$$

Since:
$$
e=U-\tilde U
$$

we get:
$$
\boxed{V^TAe=0}
$$

Step 2: Substitute into the projection operator
$$
P_Ae=V(V^TAV)^{-1}V^TAe
$$

Using $V^TAe=0$:
$$
P_Ae=V(V^TAV)^{-1}(0)
$$

Therefore:
$$
\boxed{P_Ae=0}
$$
