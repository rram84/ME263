
# Lecture 3 — Why Does the Weak Formulation Have a Solution?

## Motivation from energy, membrane deflection, function spaces and Hilbert spaces

### Main question


> **Why should the weak formulation have a solution?**




# 1. The Poisson problem

For the homogeneous Dirichlet case,

$$
-\Delta u=f \qquad \text{in }\Omega,
$$

with

$$
u=0 \qquad \text{on }\Gamma_D.
$$

Here:

- $\Omega$ is the domain.
- $u(x,y)$ is the unknown field.
- $f(x,y)$ is the source/load.
- $\Gamma_D$ is the Dirichlet boundary.

The important point is that $u=0$ is **one particular boundary condition**, not a requirement of the Poisson equation itself.

For the general Dirichlet condition,

$$
u=g \qquad \text{on }\Gamma_D.
$$



## 2. Weak form

Take a test/variation function $v$ that satisfies the homogeneous essential boundary condition,

$$
v=0 \qquad \text{on }\Gamma_D.
$$

Multiply the strong equation by $v$ and integrate over $\Omega$:

$$
\int_\Omega (-\Delta u)v\,d\Omega
=
\int_\Omega fv\,d\Omega.
$$

Using integration by parts,

$$
\int_\Omega \nabla u\cdot\nabla v\,d\Omega
=
\int_\Omega fv\,d\Omega
$$

for the homogeneous Dirichlet case.

Thus the weak form is

$$
\boxed{
\int_\Omega \nabla u\cdot\nabla v\,d\Omega
=
\int_\Omega fv\,d\Omega
\qquad\forall v\in V.
}
$$



## 3. Define $a(u,v)$ and $L(v)$

Define

$$
\boxed{
a(u,v)
=
\int_\Omega \nabla u\cdot\nabla v\,d\Omega
}
$$

and

$$
\boxed{
L(v)
=
\int_\Omega fv\,d\Omega.
}
$$

Then the weak formulation becomes

$$
\boxed{
a(u,v)=L(v)
\qquad\forall v\in V.
}
$$

### What do these symbols mean?

$$
a:V\times V\rightarrow\mathbb R
$$

is a **bilinear form**: it takes two functions, $u$ and $v$, and returns a real number.

$$
L:V\rightarrow\mathbb R
$$

is a **linear functional**: it takes one function, $v$, and returns a real number.

The notation $V\times V$ means the Cartesian product of the space with itself; it is **not** the vector cross product.



# 3.1. Requirements on $V$ and $a$

Now comes the main question of Lecture 3:

> **What mathematical requirements are needed so that**
> $$
> a(u,v)=L(v)
> $$
> **has a solution?**

We need suitable properties of the function space $V$, the bilinear form $a$, and later the functional $L$.


## 3.1.1. Requirements on $V$

The weak formulation is

$$
a(u,v)=L(v)\qquad\forall v\in V.
$$

So before discussing existence, we must understand what kind of space $V$ should be.

### 1. $V$ must be a vector space

If $u,v\in V$ and $\alpha,\beta\in\mathbb R$, then

$$
\boxed{\alpha u+\beta v\in V.}
$$

This allows us to add functions, scale them, form variations, and construct approximations.

### 2. $V$ needs an inner product

An inner product lets us compare functions and measure their size.

For example,

$$
\boxed{
(u,v)_{L^2}
=
\int_\Omega uv\,d\Omega
}
$$

with corresponding norm

$$
\boxed{
\|u\|_{L^2}
=
\sqrt{(u,u)_{L^2}}
=
\left(\int_\Omega u^2\,d\Omega\right)^{1/2}.
}
$$

So the inner product gives us a notion of distance and size for functions.

### 3. Why $H^1(\Omega)$?

The strong Poisson equation contains second derivatives:

$$
-\Delta u=f.
$$

But the weak formulation contains only first derivatives:

$$
\int_\Omega\nabla u\cdot\nabla v\,d\Omega.
$$

Therefore we need functions whose first weak derivatives are square-integrable. This is the role of the Sobolev space

$$
\boxed{H^1(\Omega)}.
$$

Informally,

$$
\boxed{
H^1(\Omega)
=
\{u:\ u\text{ and its first weak derivatives belong to }L^2(\Omega)\}.
}
$$

For a two-dimensional domain,

$$
u,\quad u_x,\quad u_y\in L^2(\Omega).
$$

A standard $H^1$ norm is

$$
\boxed{
\|u\|_{H^1}^2
=
\int_\Omega
\left(u^2+|\nabla u|^2\right)d\Omega.
}
$$

### 4. Incorporating the boundary condition

For the homogeneous Dirichlet problem,

$$
u=0\qquad\text{on }\Gamma_D,
$$

the allowable variations/test functions satisfy

$$
v=0\qquad\text{on }\Gamma_D.
$$

Therefore,

$$
\boxed{
V=H_0^1(\Omega)
=
\{v\in H^1(\Omega):v=0\text{ on }\Gamma_D\}.
}
$$

So $H_0^1(\Omega)$ is the $H^1$ space with the required zero Dirichlet boundary condition built in.

For a nonzero Dirichlet condition $u=g$,

$$
u\in U_g,
\qquad
v\in V.
$$

### 5. Completeness

Suppose

$$
u_1,u_2,u_3,\ldots
$$

is a sequence of functions that becomes arbitrarily close to itself:

$$
\|u_n-u_m\|_V\rightarrow0
\qquad\text{as }n,m\rightarrow\infty.
$$

Such a sequence is called a **Cauchy sequence**.

A space is **complete** if every Cauchy sequence converges to a limit that is still inside the same space.

Therefore,

$$
\boxed{
\text{complete space}
\Rightarrow
\text{limits of valid approximations remain in the space}.
}
$$

This matters because existence proofs and approximation procedures involve taking limits.

### 6. Therefore $V$ is a Hilbert space

A Hilbert space is a

$$
\boxed{\text{complete inner-product vector space}.}
$$

Thus,

$$
\boxed{
V=H_0^1(\Omega)
}
$$

provides the required Hilbert-space setting for the homogeneous Dirichlet Poisson problem.

### Summary

$$
\boxed{
\text{vector space}
\rightarrow
\text{inner product}
\rightarrow
\text{norm}
\rightarrow
\text{completeness}
\rightarrow
\text{Hilbert space }V=H_0^1(\Omega)
}
$$

Now we ask:

> **What properties must the bilinear form $a(u,v)$ have?**



## 3.1.2. Requirements on $a(u,v)$

For the Poisson problem,

$$
a(u,v)=
\int_\Omega\nabla u\cdot\nabla v\,d\Omega.
$$

The important properties are:

### 1. Bilinearity

$$
a(\alpha u_1+\beta u_2,v)
=
\alpha a(u_1,v)+\beta a(u_2,v),
$$

and similarly in the second argument.

### 2. Symmetry

$$
\boxed{
a(u,v)=a(v,u).
}
$$

### 3. Continuity

There must be a constant $C>0$ such that

$$
\boxed{
|a(u,v)|
\le C\|u\|_V\|v\|_V.
}
$$

For the Poisson form this follows from the Cauchy–Schwarz inequality.

### 4. Coercivity

There must be an $\alpha>0$ such that

$$
\boxed{
a(v,v)\ge\alpha\|v\|_V^2.
}
$$

For the Poisson problem,

$$
a(v,v)=
\int_\Omega|\nabla v|^2\,d\Omega.
$$

With $V=H_0^1(\Omega)$, Poincaré's inequality provides the required control of the $H^1$-norm.

These properties are precisely what allow the existence/uniqueness result to be applied.



## 3.1.3. Requirements on $L(v)$

The load functional is

$$
L(v)=\int_\Omega fv\,d\Omega.
$$

It must be

- **linear**, and
- **continuous**, meaning there is a constant $C_L>0$ such that

$$
\boxed{
|L(v)|\le C_L\|v\|_V.
}
$$

Once the required properties of $V$, $a$, and $L$ are established, the Lax–Milgram theorem gives existence and uniqueness.


# Geometry of the deformed membrane

Represent the surface by

$$
\mathbf r(x,y)=
\begin{bmatrix}
x\\
y\\
u(x,y)
\end{bmatrix}.
$$

The two tangent vectors are obtained by differentiating with respect to $x$ and $y$:

$$
\boxed{
t_1=\frac{\partial\mathbf r}{\partial x}
=
(1,0,u_x)
}
$$

and

$$
\boxed{
t_2=\frac{\partial\mathbf r}{\partial y}
=
(0,1,u_y).
}
$$

These two vectors span a small tangent parallelogram on the deformed membrane.



# Deformed area element

The area of the tangent parallelogram is the magnitude of the cross product:

$$
dA_{\text{def}}
=
|t_1\times t_2|\,dA.
$$

Calculate

$$
t_1\times t_2
=
(-u_x,-u_y,1).
$$

Therefore,

$$
\begin{aligned}
|t_1\times t_2|
&=
\sqrt{u_x^2+u_y^2+1}\\
&=
\sqrt{1+u_x^2+u_y^2}.
\end{aligned}
$$

Since

$$
|\nabla u|^2=u_x^2+u_y^2,
$$

we obtain

$$
\boxed{
|t_1\times t_2|
=
\sqrt{1+|\nabla u|^2}.
}
$$

This is the key geometric step: **the gradient of the deflection controls the increase in membrane area.**



# Change in area

Let the original flat area be

$$
A_0=\int_\Omega dA.
$$

The deformed area is

$$
A_{\text{def}}
=
\int_\Omega
\sqrt{1+|\nabla u|^2}\,dA.
$$

Therefore,

$$
\boxed{
\Delta A
=
\int_\Omega
\sqrt{1+|\nabla u|^2}\,dA
-
\int_\Omega dA.
}
$$



# Small-deflection / small-slope approximation

For a small slope,

$$
|\nabla u|^2\ll1.
$$

Use the Taylor approximation

$$
\sqrt{1+s}
\approx
1+\frac{s}{2}
\qquad (|s|\ll1).
$$

Taking

$$
s=|\nabla u|^2,
$$

gives

$$
\boxed{
\sqrt{1+|\nabla u|^2}
\approx
1+\frac12|\nabla u|^2.
}
$$

Substitute into the change in area:

$$
\begin{aligned}
\Delta A
&\approx
\int_\Omega
\left(
1+\frac12|\nabla u|^2
\right)dA
-\int_\Omega dA\\
&=
\boxed{
\frac12\int_\Omega|\nabla u|^2\,dA
}.
\end{aligned}
$$

This is the key approximation used to obtain the quadratic membrane energy.



# From change in area to membrane energy

For a membrane under constant tension $T$, the strain energy is proportional to the increase in area:

$$
U=T\,\Delta A.
$$

Using the small-slope result,

$$
\boxed{
U[u]
=
\frac{T}{2}
\int_\Omega|\nabla u|^2\,dA.
}
$$

If $T=1$ is used for normalization, this becomes

$$
\boxed{
U[u]
=
\frac12
\int_\Omega|\nabla u|^2\,dA.
}
$$

This explains physically where the term

$$
\int_\Omega|\nabla u|^2\,dA
$$

comes from.



# External work and total potential energy

The transverse load $f$ does work as the membrane deflects.

With the sign convention used for the Poisson problem, the potential-energy contribution of the load is

$$
-\int_\Omega fu\,dA.
$$

Therefore the total potential energy is

$$
\boxed{
J[u]
=
\frac{T}{2}
\int_\Omega|\nabla u|^2\,dA
-
\int_\Omega fu\,dA.
}
$$

For $T=1$,

$$
\boxed{
J[u]
=
\frac12
\int_\Omega|\nabla u|^2\,dA
-
\int_\Omega fu\,dA.
}
$$

The three methods are three ways of calculating the same quantity, a(uh, wh), for the finite-element solution uh. In the first method, we directly write the integral ∫Ω |∇uh|² dx and assemble it. In the second method, we write the same integral but use the scalar_eval() function, which packages the form creation, assembly, and MPI summation into one reusable function. In the third method, instead of writing the integral again, we use the bilinear form a(u, v) that was already defined and substitute uh into both arguments, giving a(uh, uh). Therefore, all three methods should give the same numerical value.
