# 3D Form of the Neo-Hookean Constitutive Model

In this lecture, we will justify the 3D form of the Neo-Hookean constitutive model based on its form in 1D.

The 1D form of the Neo-Hookean constitutive model is given as, 
$$
W(\lambda) = \frac{\mu}{2}(\lambda^2-1) - \mu \log \lambda
$$
Now, based on our intuition, we will expect the 3D form to be the sum of the energy due to stretch along the 3 directions, i.e., 

$$
W = W(\lambda_1) + W(\lambda_2) + W(\lambda_3) = \frac{\mu}{2}(\lambda_1^2 + \lambda_2^2 + \lambda_3^2 - 3) - \mu(\log \lambda_1 + \log \lambda_2 + \log \lambda_3)
$$

which can be further simplified as 

$$
W
=
\frac{\mu}{2}(I_C-3)
-
\mu \log J,
$$

where, 
$$
I_C = \operatorname{tr}(C), C = F^T F
$$
and 


$$
J = \det(F).
$$


The issue with the above form is that it contains only 1 material parameter $\mu$. So, accounting for just the length changes along the 3 directions is not sufficient for accounting volume changes. In order to account for volume changes, there should be some coupling. Moreoever, the neo-hookean model should get back to the linear one in the small strain limit

Now, we check the small strain limit of the terms obtained above
We know that,

$$
F = I + \nabla u.
$$

So,
$$
I_C
=
\operatorname{tr}(F^T F)
=\operatorname{tr}{(I + \nabla u)^T (I + \nabla u)}
+ \operatorname{tr}{}
=
\operatorname{tr}{I}
+
2\,\operatorname{tr}(\epsilon)
+\operatorname{tr}\!\left((\nabla u)^T\nabla u\right)
=
3
+
2\,\operatorname{tr}(\epsilon)
+
(\nabla u : \nabla u),
$$


where

$$
\epsilon
=
\frac12
\left(
\nabla u + \nabla u^T
\right).
$$

So, 
$$
I_C - 3 = 2\,\operatorname{tr}(\epsilon)
+
(\nabla u : \nabla u)
$$

Similarly,
$$
\log J
=
\log(\det(I+\nabla u))
=
\operatorname{tr}(\nabla u)
-
\frac12 \operatorname{tr}\!\left((\nabla u)^2\right)
+
O(\|\nabla u\|^3).
$$

the above expression is based on the Matrix analogue of the identity for the expansion $$\log(1 + x) = x - \frac12 x^2 + ...$$
Knowing that $\operatorname{tr}{(\nabla u)} = \operatorname{tr}{(\epsilon)} + \operatorname{tr}{(W)} = \operatorname{tr}{(\epsilon)} + 0$, we get 
$$
\log J
=
\operatorname{tr}(\epsilon)
-
\frac12 \operatorname{tr}\!\left((\nabla u)^2\right)
+
O(\|\nabla u\|^3).
$$

Substituting into the above results for $I_C - 3$ and $\log J$ into the expected expression for W gives 
$$
W
=
\frac{\mu}{2}(I_C-3)
-
\mu \log J
=
\frac{\mu}{2}((\nabla u : \nabla u) + \operatorname{tr}\!\left((\nabla u)^2\right)
=
\frac{\mu}{2}[(\nabla u : \nabla u) + (\nabla u : \nabla u^T)]
= \mu (\epsilon : \epsilon)
$$


---
Note that the above result for combining the two terms into a tensor inner product of the strain tensors is based on the following trick -  
We express $\nabla u$ as an additive decomposition of a symmetric and skew symmetric tensor.

$$
\nabla u = \epsilon + W,
$$

where

$$
W
=
\frac12
\left(
\nabla u-\nabla u^T
\right)
$$

is the skew-symmetric part.

Then

$$
(\nabla u)^T
=
\epsilon - W.
$$

Hence

$$
(\nabla u) : (\nabla u)^T
=
(\epsilon + W):(\epsilon - W)
=
\epsilon:\epsilon - W:W,
$$

and

$$
(\nabla u) : (\nabla u)
=
(\epsilon + W):(\epsilon + W)
=
\epsilon:\epsilon + W:W.
$$

Therefore,

$$
\frac{\mu}{2}((\nabla u : \nabla u) + \operatorname{tr}\!\left((\nabla u)^2\right)
=
\frac{\mu}{2}[(\nabla u : \nabla u) + (\nabla u : \nabla u^T)]
= \mu (\epsilon : \epsilon)
$$

Thus

$$
\frac{\mu}{2}(I_C-3)-\mu\log J
\approx
\mu (\epsilon : \epsilon)
$$

---

So, we can see that the assumed form of the Neo-Hookean Model in 3D gets to the linear constitutive model in the small strain limit. However, one of the terms is still missing, as the complete form of the linear elastic constitutive model is given as 

$$
W = \frac {\lambda}{2} (\operatorname{tr}{(\epsilon)})^2 + \mu (\epsilon : \epsilon)

$$

So, we must consider an additional term in the 3D form of the Neo-Hookean constitutive model.

Now, we want something to account for the volume changes.
A reasonable measure of volumetric strain should satisfy:

1. Additivity:
$$
vol-strain(J_3)
=
vol-strain(J_2)
+
vol-strain(J_1).
$$

2. Scale invariance.

These properties are satisfied by

$$
vol-strain(J)=\log J.
$$



Hence we augment the energy by a function \(g(J)\):

$$
W
=
g(J)
+
\frac{\mu}{2}(I_C-3)
-
\mu\log J.
$$

There are some conditions that the function $g(J)$ must satisfy
$$
g(1)=0,
$$

$$
g'(1)=0,
$$

and

$$
g''(1)=\lambda.
$$

There requirements are satisfied by considering $g(J)$ of the form

$$
g(J)
=
\frac{\lambda}{2}
(\log J)^2.
$$

Therefore the 3D Neo-Hookean model becomes

$$
W
=
\frac{\mu}{2}(I_C-3)
-\mu\log J
+
\frac{\lambda}{2}
(\log J)^2.
$$

A final check is the term $g(J)$ should go to $\frac{\lambda}{2}(\operatorname{tr}{\epsilon})^2$ within the small strain limit.

For small strains,

$$
J
=
\det(F)
=
\det(I+\nabla u)
=
1+\operatorname{tr}(\epsilon)
+
O(\|\nabla u\|^2).
$$

Therefore

$$
\log J
=
\log(1+\operatorname{tr}(\epsilon))
\approx
\operatorname{tr}(\epsilon).
$$

Hence

$$
\frac{\lambda}{2}
(\log J)^2
\approx
\frac{\lambda}{2}
\bigl(\operatorname{tr}(\epsilon)\bigr)^2.
$$

---

A point to note is that the linear elastic constitutive law is not objective. Objectivity restricts what a good measure of strain is.

# Directional Derivative
We know that

$$
\Pi(u)
=
\int_\Omega
W(F(u))
\, dV.
$$

The displacement corresponding to equilibrium is the one that extrimizes $\Pi$
So, can we directly write that the equilibrium displacement is the one that satisfies

$$
\frac {d\Pi}{du} = 0
$$

Note that one must be careful about these ideas, as a mere straightforward extrapolation of the 1D ideas can be misleading when dealing with functions with vector arguments, or functionals.

To make things clear, we define the directional derivative of $\Pi$ at $(u)$ along $(v)$ as

$$
D\Pi(u)[v]
=
\lim_{\varepsilon\to0}
\frac{
\Pi(u+\varepsilon v)-\Pi(u)
}{\varepsilon}.
$$

At equilibrium,

$$
D\Pi(u)[v]
=
0
\qquad
\forall\,v.
$$

which denotes the stationarity of energy, which means that energy doesn't change irrespective of which direction we move in.


---

Example: Directional Derivative of \($W(F)=F^T F$\)

We define

$$
W(F)=F^T F.
$$

Then

$$
DW(F)[H]
=
\lim_{\varepsilon\to0}
\frac{
W(F+\varepsilon H)-W(F)
}{\varepsilon}.
$$

Using the definition of $W(F)$,

$$
(F+\varepsilon H)^T(F+\varepsilon H)
=
F^TF
+
\varepsilon(H^TF+F^TH)
+
\varepsilon^2H^TH,
$$

and substituting this in the derivative formulation, we obtain

$$
DW(F)[H]
=
H^TF+F^TH.
$$

Note that the derivative is linear in $H$.

---








A point to remember is that $D\Pi(u)[v]$ is linear in $v$.
But, why and how does it matter to us?

$$ \Pi(u)=\int_{\Omega_0} W(F(u))\,d\Omega_0 $$ For equilibrium, the first variation vanishes: $$ D\Pi(u)[v]=0 \qquad \forall v $$ or equivalently, $$ \left.\frac{d}{d\varepsilon}\Pi(u+\varepsilon v)\right|_{\varepsilon=0}=0, \qquad \forall v. $$ Substituting the expression for $\Pi$, $$ \left.\frac{d}{d\varepsilon}\int_{\Omega_0} W\!\big(F(u+\varepsilon v)\big)\,d\Omega_0 \right|_{\varepsilon=0} =0. $$ Interchanging differentiation and integration, $$ \int_{\Omega_0} \left. \frac{d}{d\varepsilon} W\!\big(F(u+\varepsilon v)\big) \right|_{\varepsilon=0} \,d\Omega_0 =0. $$ Applying the chain rule, $$ \int_{\Omega_0} \frac{\partial W}{\partial F} : \left. \frac{d}{d\varepsilon} F(u+\varepsilon v) \right|_{\varepsilon=0} \,d\Omega_0 =0. $$ Defining $$ P=\frac{\partial W}{\partial F}, \qquad \left. \frac{d}{d\varepsilon} F(u+\varepsilon v) \right|_{\varepsilon=0} =\nabla v, $$ gives $$ \int_{\Omega_0} P(F(u)):\nabla v \,d\Omega_0 =0, \qquad \forall v. $$

Here, $P$ denotes the "First Piola-Kirchoff Stress Tensor"

So, the virtual work of the internal stresses vanishes for every admissible virtual displacement.

When solving non-linear problems, we define the residual as 

$$
Res = ufl.derivative(\Pi, u_h, v)
$$

This residual is nothing but 

$$
Res = D\Pi(u_h)[v]
$$

We seek $u_h$ such that 
$$ R(u_h; v) = 0 \qquad \forall v \in V. 
$$

It is sufficient to require $$ R(u_h; \phi_i) = 0, \qquad i = 1,\ldots,N. $$ 
Where $\phi_i$ denotes the basis functions. Since any test function $v$ in the finite-dimensional space can be written as (assume summation over repeated indices) $$ v = v_i \phi_i, $$ we have, by linearity in the second argument, $$ R(u_h; v_i \phi_i) = v_i\,R(u_h;\phi_i). $$ If $$ R(u_h;\phi_i)=0, \qquad \forall i, $$ then $$ R(u_h;v)=0, \qquad \forall v. $$ The finite-element approximation is written as $$ u_h= u_a \phi_a, $$ or equivalently, $$ \mathbf{u}_h= \begin{bmatrix} u_1\\ u_2\\ \vdots\\ u_N \end{bmatrix}. $$
