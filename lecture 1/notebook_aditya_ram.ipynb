The Poisson equation acts as a fundamental mathematical prototype for modeling numerous physical phenomena. Because it elegantly describes systems driven by gradients such as heat conduction, electrostatics, mass diffusion, and fluid flow.

Before setting up the equations, we must define the physical space where our problem exists.

<p align="center">
  <img src="images/1.png" width="400">
</p>


*   **$\Omega$ (The Domain):** Represents the continuous interior volume.
*   **$\partial\Omega$ (The Boundary):** Represents the outer boundary enclosing the domain.

The boundary value problem for the Poisson equation can be expressed in its primary mathematical form, known as the strong form. The strong form represents the exact governing partial differential equation and its boundary conditions evaluated at every single point in the domain.

$$-\text{div}(q) = f \quad \text{in} \quad \Omega$$

$$u = 0 \quad \text{on} \quad \partial\Omega$$

where:
*   **$q$**: is the flux term.
*   **$f$**: is the source term.
*   **$u$**: is the primary variable.


The Poisson equation fundamentally consists of two distinct parts. 

The first part represents the physics, which is expressed as an equation representing some form of balance. This balance equation is given by $\text{div}(q) = f$ in $\Omega$, where $q$ represents the flux. Depending on the specific physics we are solving, the definitions of the flux ($q$),source ($f$),variable ($u$) terms will change accordingly.

To understand this balance, consider $\omega$ as any arbitrary subset within our main domain $\Omega$.

<p align="center">
  <img src="images/2.png" width="400">
</p>

Considering the example of heat transfer, this balance dictates that the amount of rate of heat energy produced within the subset is exactly equal to the amount of heat flux flowing through its boundary, $\partial\omega$. This can be expressed as:

$$\int_{\partial \omega} q \cdot n \, ds = \int_{\omega} f \, d\omega \tag{1}$$

where $n$ is the outward normal vector for $\omega$.

Next, by applying the divergence theorem, we can relate the surface integral of the flux over the boundary $\partial\omega$ to the volume integral of the divergence over the subset $\omega$:

$$\int_{\partial \omega} q \cdot n \, ds = \int_{\omega} \text{div}(q) \, d\omega \tag{2}$$

Combining equations (1) and (2),

$$\int_{\omega} \text{div}(q) \, d\omega = \int_{\partial \omega} q \cdot n \, ds = \int_{\omega} f \, d\omega$$


$$\int_{\omega} \text{div}(q) \, d\omega = \int_{\omega} f \, d\omega$$

Because we can choose this subset $\omega$ completely arbitrarily, the balance holds true for any such region. By the arbitrariness of this subset domain, the governing differential equation is guaranteed to be valid throughout the entire domain $\Omega$:

$$\int_{\omega} \text{div}(q) = f$$


The second part is the constitutive law, which takes into account the actual specific problem being modeled. A primary example is heat conduction, which is one of the specific cases of the Poisson equation. For heat transfer, the constitutive relationship is Fourier's law, defined as 

$$q = -k\nabla u$$

It is critically important to maintain the negative sign in this constitutive equation; if we ignore the negative sign here, the resulting stiffness matrix  will become negative definite.

### Examples of the Poisson Equation in Different Physics

**1. Heat Conduction**

In the context of heat transfer, the Poisson equation models the steady-state temperature distribution within a material. 

*   **Constitutive Law :** The behavior of the flux is governed by Fourier's Law of heat conduction,can be expressed as:
    $$q = -k\nabla u$$
*   **$f$ :** Represents the internal volumetric heat generation rate, **W/m³**.
*   **$q$ :** Represents the heat flux (the rate of heat transfer per unit cross-sectional area), **W/m²**.
*   **$k$ :** Represents the thermal conductivity of the material, **W/(m·K)** or **W/(m·°C)**.
*   **$u$ :** Represents the temperature field variable, **K** or **°C**.

**2. Mass Diffusion (Fick's Law)**

In the context of mass transfer, the Poisson equation models the diffusion of a substance within a medium. This mathematical framework works because there is a gradient in a specific quantity (concentration) that physically drives the flow.
*   **Constitutive Law :** The behavior of the mass flux is governed by Fick's Law of Diffusion, 
    $$q = -D\nabla u$$
*   **$f$ :** Represents the internal mass generation or depletion rate per unit volume (e.g., due to a chemical reaction), **kg/(m³·s)**.
*   **$q$ :** Represents the mass flux (the rate of mass transfer per unit cross-sectional area), **kg/(m²·s)**.
*   **$D$ :** Represents the mass diffusion coefficient, **m²/s**.
*   **$u$ :** Represents the mass concentration field variable, **kg/m³**.

**3. Fluid Flow in Porous Media (Darcy's Law)**

In the context of fluid mechanics, specifically for flow through porous media (such as groundwater flowing through soil or fluid passing through a filter), the Poisson equation models the steady-state pressure distribution. Here, the flow of the fluid is physically driven by a pressure gradient across the medium.

*   **Constitutive Law :** The behavior of the fluid flux is governed by Darcy's Law,
    $$q = -K\nabla u$$
*   **$f$ :** Represents the volumetric fluid injection or extraction rate per unit volume , **1/s** (which is m³/(m³·s)).
*   **$q$ :** Represents the specific discharge or Darcy flux (the volumetric flow rate per unit cross-sectional area), **m/s**.
*   **$K$ :** Represents the mobility coefficient, which depends on the permeability of the porous medium and the viscosity of the fluid, **m²/(Pa·s)** or **m⁴/(N·s)**.
*   **$u$ :** Represents the fluid pressure field variable, **Pa** or **N/m²**.

**4. Membrane Deflection**

Poisson equation can also model membrane deflection, such as a deflection of stretched thin cloth. It is derived from the balance of force in the vertical direction.Here *initial* (undeformed) surface area which is represented by dashed line in the above figure and *end* position after deflection is represented by solid line and the yellow dots at the two ends are supports. This model relies on the small deflection approximation of $u$.

<p align="center">
  <img src="images/3.png" width="400">
</p>

*   Governing Equation of membrane deflection can be expressed as,
    $$\nabla \cdot (\nabla u) = -p/T$$
    *(Note: In this specific context, $\nabla u$ acts as the equivalent of our flux term $q$, and $-p/T$ acts as our source term $f$.)*
*   **$u$ :** Represents the vertical displacement of the membrane, **m**.
*   **$p$ :** Represents the pressure. Because we are assuming a small displacement case, we approximate that this pressure is acting normal to the *initial* (undeformed) surface area, **Pa** or **N/m²**. 
*   **$T$ :** Represents the tension $T$ in the membrane (force per unit length). This tension is assumed to be constant and acts in the tangent direction of the membrane, **N/m**.

Consider a small differential element with side lengths $dx$ and $dy$.

<p align="center">
  <img src="images/4.png" width="400">
</p>


To derive the governing equation, we perform a force balance  in the vertical direction on our small differential element. Let's first analyze the forces acting along the membrane by seeing from the $y$ plane, from position $x$ to $x + \Delta x$.

<p align="center">
  <img src="images/5.png" width="400">
</p>

The tension $T$ acts tangentially to the deflected membrane. The vertical component of this tension at any point is $T \sin\theta$ where $\theta$ is measured from the positive $x$-axis. Utilizing our small deflection approximation, we can state that $\sin\theta \approx \theta \approx \tan\theta$. From calculus, we know that the tangent of the angle (the slope) is equal to the spatial derivative of the displacement:

$$\tan\theta = \frac{\partial u}{\partial x}$$

Therefore, the net vertical force acting on the element's edges along the $x$-axis (spanning a width of $\Delta y$) is the difference between the vertical tension components at $x + \Delta x$ and $x$:

$$T \Delta y \left[ \left. \frac{\partial u}{\partial x} \right|_{x+\Delta x} - \left. \frac{\partial u}{\partial x} \right|_x \right]$$

By dividing and multiplying this expression by $\Delta x$ and taking the limit as $\Delta x \to 0$, we can express it as a second order partial derivative:

$$T \Delta x \Delta y \frac{\partial^2 u}{\partial x^2}$$

Applying the exact same logic on the other edges (along the $y$-direction) we get,

$$T \Delta x \Delta y \frac{\partial^2 u}{\partial y^2}$$

Now, we consider both of these net boundary forces and balance them against the external pressure $p$. Again, due to our small displacement assumption, the pressure $p$ is assumed to act over the *initial* undeformed horizontal area ($\Delta x \Delta y$). The restoring forces from the membrane tension must balance this applied downward it can result as ,

$$T \Delta x \Delta y \left( \frac{\partial^2 u}{\partial x^2} + \frac{\partial^2 u}{\partial y^2} \right) + p \Delta x \Delta y = 0$$

Dividing out the common area term ($\Delta x \Delta y$) and rearranging the equation, we arrive at the strong form of the Poisson equation for membrane deflection (which can intuitively be thought of as the 2D analog version of strings):

$$\frac{\partial^2 u}{\partial x^2} + \frac{\partial^2 u}{\partial y^2} = -\frac{p}{T}$$


$$\nabla \cdot (\nabla u) = -\frac{p}{T}$$

**5. Electrostatics**

In the context of electromagnetism, the Poisson equation is used to model the electrostatic potential field generated by a specific charge distribution across a domain.

*   **Constitutive Law :** The electric field (acting as our flux equivalent) is driven by the negative gradient of the voltage,expressed as:
    $$q = -\nabla u$$
*   **$f$:** Represents the ratio of charge density to the permittivity of the medium. It is defined by the formula $f = \frac{\rho}{\varepsilon}$, where $\rho$ represents the volumetric charge density (C/m³) and $\varepsilon$ represents the electric permittivity (F/m). unit  is **V/m²**.
*   **$q$ :** Represents the electric field, **V/m** .
*   **$u$ :** Represents the voltage (electric potential) **V**.

**6. Data Extending from Boundary to Interior**

Beyond modeling physical phenomena, the Poisson equation specifically in its homogeneous form, known as Laplace's equation (where the source term $f = 0$) serves as a powerful mathematical tool for Extending from Boundary to Interior. If we have a domain $\Omega$ where the data is only known along the outer edge $\partial\Omega$, we can use this governing equation to extend that data throughout the entire interior. You can think of this conceptually as having color data at the boundary and mathematically propagating it inward to fill the interior space.


Using this specific differential equation is an optimal choice for data filling because it yields a perfectly smooth function for all internal points, inherently satisfying two critical mathematical properties: the **Mean Value Theorem** and the **Maximum Principle**.

If we instead attempted to fill the interior space simply by applying the global average of the boundary points everywhere, it would result in a constant value throughout the interior. This creates a sudden, discontinuous jump in the function exactly at the boundary. Solving Laplace's equation naturally avoids this problem.


### The Mean Value Theorem for Harmonic Functions

Functions that solve Laplace's equation ($\Delta u = 0$) are known as **harmonic functions**. The Mean Value Theorem states that if $u$ is a harmonic function within a given domain $\Omega$, then the exact value of $u$ at any specific point $x$ inside that domain is perfectly equal to the average value of $u$ over the boundary of any circle ,which lies completly inside domain $\Omega$, centered at point $x$.

<p align="center">
  <img src="images/6.png" width="400">
</p>


To prove this, we first define a function $\phi(r)$ representing the average value of the function $u$ over a circle of radius $r$ centered at point $x$:

$$\phi(r) = \frac{1}{2\pi r} \int_{\text{circle}} u \, ds$$

In this equation, $2\pi r$ is the circumference, and the line integral sums the values around the boundary. Dividing the integral by the circumference yields the average value.

Next, we convert this line integral into polar coordinates, where $ds$ becomes $r \, d\theta$:

$$\phi(r) = \frac{1}{2\pi r} \int_{0}^{2\pi} u(r, \theta) \, r \, d\theta$$

Because we are finding the average over one specific circle of a fixed size, our distance from the center ($r$) remains exactly the same as we sweep around the perimeter. Therefore, with respect to the integration variable $\theta$, the radius $r$ is a constant.

pull that constant $r$ completely outside the integral sign,

$$\phi(r) = \frac{1}{2\pi r} \cdot r \int_{0}^{2\pi} u(r, \theta) \, d\theta$$

The $r$ in the numerator and denominator cancel each other out, leaving us with a  equation ,

$$\phi(r) = \frac{1}{2\pi} \int_{0}^{2\pi} u(r, \theta) \, d\theta$$

The core of the Mean Value Theorem relies on proving that this average value never changes, regardless of how large or small the circle is. We do this by proving its derivative with respect to r is zero.

We take the derivative of above equation with respect to $r$. Because the integration limits are constants, we move the derivative directly inside the integral:

$$\phi'(r) = \frac{1}{2\pi} \int_{0}^{2\pi} \frac{\partial u}{\partial r} \, d\theta$$

We convert back from ($d\theta$) to ($ds$). Since $d\theta = \frac{ds}{r}$, we substitute this in. Because $r$ is constant for this specific circle, we factor $\frac{1}{r}$ outside the integral,

$$\phi'(r) = \frac{1}{2\pi r} \int_{\text{circle}} \frac{\partial u}{\partial r} \, ds$$

Physically, the partial derivative $\frac{\partial u}{\partial r}$ represents the rate of change pointing straight outward from the center. This is exactly equal to the gradient of $u$ dotted with the outward-pointing unit normal vector $n$:

$$\frac{\partial u}{\partial r} = \nabla u \cdot n$$

by Substituting,

$$\phi'(r) = \frac{1}{2\pi r} \int_{\text{circle}} \nabla u \cdot n \, ds$$

The Divergence Theorem states that the flux of a vector field across a boundary (the circle) equals the integral of its divergence over the interior area (the disk):

$$\int_{\text{circle}} (\nabla u) \cdot n \, ds = \int_{\text{disk}} \nabla \cdot (\nabla u) \, dA$$

since $\nabla \cdot (\nabla u)$ is the Laplacian ($\Delta u$),above equation becomes,

$$\phi'(r) = \frac{1}{2\pi r} \int_{\text{disk}} \Delta u \, dA$$

By definition, our function is harmonic, meaning there is no internal source generation:
$$\Delta u = 0 \quad \text{everywhere inside the domain}$$

Because $\Delta u$ is exactly zero, the volume integral over the disk is the integral of zero. Therefore:

$$\phi'(r) = 0$$

Because the derivative is zero, we have proven that the average value $\phi(r)$ is completely constant and independent of the radius $r$.So the average value on the boundary cannot change as the circle grows or shrinks.

Since $\phi(r)$ is constant for all valid radii, we can find its true value by taking the mathematical limit as the circle shrinks down to the center point $x$ ($r \to 0$). As the circle becomes infinitely small, the average value simply converges to the exact value of the function at that center point:

$$\lim_{r \to 0} \phi(r) = u(x)$$

Because $\phi(r)$ is constant, it must be equal to $u(x)$ for *every* radius $r$:

$$u(x) = \frac{1}{2\pi r} \int_{\text{circle}} u \, ds$$

## Implementation

> **Note:** This code will work when this `ME263_Lecture1_Notes.ipynb` file is opened in the `fenicsx-env` environment.

FEniCSx is designed to mirror the actual mathematical formulation of the Finite Element Method as closely as possible. Rather than manually looping over elements to build local stiffness matrices, we use the Unified Form Language (UFL) to write the mathematical weak form directly into the script. 

Here we will go through the code and explain what it is going to do and not by explainig each and every function.

**1. Imports and Domain Discretization**
Before solving, the continuous physical domain $\Omega$ must be discretized into a finite number of elements, forming the computational mesh $\Omega_h$. Here, we import the necessary FEniCSx modules and define our geometry as a 10x10 unit square mesh.

```python
from mpi4py import MPI
from dolfinx import mesh, fem, io
from dolfinx.fem.petsc import LinearProblem
import ufl
import numpy as np
#mesh
msh = mesh.create_unit_square(MPI.COMM_WORLD, 10, 10)
```

**2. Function Space ($V$ and $V_h$)**
In FEM, we seek a solution $u$ from a specific mathematical function space $V$ (the trial space) and test the equation against variations $v$ from a test space. 

we approximate this with a discrete, finite-dimensional subspace $V_h \subset V$. We define `V` as a space of first-order ("Lagrange", degree 1) continuous basis functions. This means the solution will be interpolated using linear shape functions across each element.

```python
#function space
V = fem.functionspace(msh, ("Lagrange", 1))
```

**3. Boundary Conditions ($\partial\Omega$)**
The strong form requires $u = 0$ on the boundary $\partial\Omega$. In this block, FEniCSx mathematically locates the outer boundary by looking for entities with a topological dimension one less than the mesh (i.e., the 1D exterior lines of our 2D square). It extracts the exact Degrees of Freedom (DOFs) located on these boundaries and explicitly binds them to $0.0$ using `dirichletbc`.

```python
# boundary condition
facets = mesh.locate_entities_boundary(msh, msh.topology.dim - 1, lambda x: np.full(x.shape[1], True))
dofs = fem.locate_dofs_topological(V, msh.topology.dim - 1, facets)
bc = fem.dirichletbc(0.0, dofs, V)
```

**4. Variational (Weak) Form**

Weak form is the core of the finite element formulation. We start with the strong form of the Poisson equation:
$$-\nabla \cdot (\nabla u) = f$$

To get the weak form, we multiply by our test function $v$ and integrate over the domain $\Omega$,
$$-\int_{\Omega} \nabla \cdot (\nabla u) v \, dx = \int_{\Omega} f v \, dx$$

Applying integration by parts  transfers one derivative from $u$ to $v$:
$$\int_{\Omega} \nabla u \cdot \nabla v \, dx - \int_{\partial\Omega} (\nabla u \cdot n) v \, ds = \int_{\Omega} f v \, dx$$

Because we are enforcing $u=0$ on the boundary(u is specified on the boundary), the test function $v$ must also be exactly $0$ on $\partial\Omega$ to maintain admissibility. Thus, the boundary integral completely vanishes, leaving us with the canonical weak form. We seek $u \in V$ such that:
$$a(u,v) = L(v) \quad \forall v \in V$$

Where the bilinear form $a(u,v)$ represents the stiffness matrix, and the linear form $L(v)$ represents the load vector:
$$a(u,v) = \int_{\Omega} \nabla u \cdot \nabla v \, dx$$
$$L(v) = \int_{\Omega} f v \, dx$$

Fenicsx also takes the weak form as input so here we are giving the weak form as input 

```python
# weak form
u = ufl.TrialFunction(V)
v = ufl.TestFunction(V)
f = fem.Constant(msh, 1.0)
a = ufl.inner(ufl.grad(u), ufl.grad(v)) * ufl.dx
L = f * v * ufl.dx  
```

**5. Matrix Assembly and Solve ($K U = F$)**
The forms $a$ and $L$ are passed to the `LinearProblem` solver. FEniCSx evaluates the shape functions and numerically integrates to assemble the global stiffness matrix $K$ and force vector $F$. The solver applies the boundary condition arrays to modify $K$ and $F$, and then solves the resulting algebraic system of linear equations $K U = F$ using a direct LU factorization (`"pc_type": "lu"`).

```python
# assemble and solve
problem = LinearProblem(a, L, bcs=[bc],
                        petsc_options_prefix="poisson_",
                        petsc_options={"ksp_type": "preonly", "pc_type": "lu"})
uh = problem.solve()
```

**6. Result Verification**
The variable `uh` now holds the discrete solution. we are checkingg max value of it

```python
print("max u:", uh.x.array.max())
```

**7. Post-Processing **
We use the XDMF file format for visualization of results in the paraview.

```python
with io.XDMFFile(msh.comm, "poisson.xdmf", "w") as xdmf:
    xdmf.write_mesh(msh)
    xdmf.write_function(uh)
```
