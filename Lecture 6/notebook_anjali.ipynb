## Why Does $J$ Have a Minimum?

From the previous section, the Poisson problem can be written in weak form as

$$
a(u,v)=L(v)
\qquad \forall v\in V
$$

and the corresponding energy functional is

$$
J(v)=\frac{1}{2}a(v,v)-L(v).
$$

We have already shown that if $u$ solves the weak problem, then $u$ minimizes $J$.



We can consider different functions and calculate

$$
J(u_1),J(u_2),J(u_3),\ldots
$$

and try to approach the minimum.

But first we need to make sure that $J$ cannot decrease indefinitely.

If J has no bottom value then, 

$$
J(v)\rightarrow-\infty,
$$

We need to find a number J can't fall below.

Therefore, we need to show that $J$ is **bounded below**.

## Showing That $J$ Is Bounded Below

Recall

$$
J(u)=\frac12a(u,u)-L(u).
$$

Since

$$
a(u,u) = \int \nabla u \cdot \nabla u \
= \|u\|_a^2
$$

we have

$$
J(u)
=
\frac12\|u\|_a^2-\int_\Omega fu\,dx.
$$

Using the **Cauchy--Schwarz inequality**,

$$
\left|\int_\Omega fu\,dx\right|
\leq
\left(\int f^2\right)^{1/2}
\left(\int g^2\right)^{1/2}
=\|f\|_{L^2}\|u\|_{L^2}.
$$

Therefore,

$$
J(u)
\geq
\frac12\|u\|_a^2
-
\|f\|_{L^2}\|u\|_{L^2}.
$$

In this equation, $f$ is a given function so let $\|f\|_{L^2}$ be $\lambda$.

So,
$$
J(u)
\geq
\frac12\|u\|_a^2
-
\lambda\|u\|_{L^2}.
$$

with the only unknown being $u$ in 2 norms.


The **Poincaré inequality** gives

$$
\|u\|_{L^2}\leq C\|u\|_a.
$$

Therefore,

$$
J(u)
\geq
\frac12\|u\|_a^2
-
C\lambda\|u\|_a.
$$

Define

$$
\alpha=C\lambda.
$$

Then

$$
J(u)
\geq
\frac12\|u\|_a^2-\alpha\|u\|_a.
$$

Completing the square,

$$
\frac12\|u\|_a^2-\alpha\|u\|_a
=
\frac12(\|u\|_a-\alpha)^2
-
\frac{\alpha^2}{2}.
$$

Since

$$
(\|u\|_a-\alpha)^2\geq0,
$$

we obtain

$$
\boxed{
J(u)\geq-\frac{\alpha^2}{2}
}
$$

Therefore, $J$ is bounded below.

So $J$ cannot decrease to $-\infty$ and is **bounded by $-\frac{\alpha^2}{2}$** irrespective of the function. It does depend on the domain.

## Minimizing Sequence

The energy values approach the minimum:

$$
J(u_1),J(u_2),J(u_3),\ldots\rightarrow J_{\min}.
$$

If J converges, it need not be that $u_n$ converge as J is an integral. 

But if the displacements converge then energy surely converges.

Hence we claim that 
$u_n \rightarrow u\quad \text{in } V$

A useful result for the quadratic functional is

$$
\boxed{
J(u)+J(w)
=
2J\left(\frac{u+w}{2}\right)
+
\frac14\|u-w\|_a^2
}
$$

### Proof:

$$
J(u)+J(w)-2J\left(\frac{u+w}{2}\right)
= \frac{1}{4}\|u-w\|_a^2
$$

We know 
$$
2J\left(\frac{u+w}{2}\right)
=
2\left[
\frac{1}{2}
a\left(\frac{u+w}{2},\frac{u+w}{2}\right)
-
L\left(\frac{u+w}{2}\right)
\right]
$$

Using the linearity of \(a\) and \(L\),

$$
=
\frac{1}{4}\|u+w\|_a^2
-L(u)-L(w)
$$

Therefore,

$$
2J\left(\frac{u+w}{2}\right)
=
\frac{1}{4}\|u+w\|_a^2
-\frac{1}{2}L(u)
-\frac{1}{2}L(w)
$$

Add $\frac{1}{4}\|u-w\|_a^2$ to both sides:

$$
2J\left(\frac{u+w}{2}\right)
+\frac{1}{4}\|u-w\|_a^2
=
\frac{1}{4}\|u+w\|_a^2
+\frac{1}{4}\|u-w\|_a^2
-\frac{1}{2}L(u)
-\frac{1}{2}L(w)
$$

Using the **parallelogram identity**,

$$
\frac{1}{4}
\left[
\|u+w\|_a^2+\|u-w\|_a^2
\right]
=
\frac{1}{2}\|u\|_a^2
+\frac{1}{2}\|w\|_a^2
$$

Hence,

$$
2J\left(\frac{u+w}{2}\right)
+\frac{1}{4}\|u-w\|_a^2
=
\left[
\frac{1}{2}\|u\|_a^2-L(u)
\right]
+
\left[
\frac{1}{2}\|w\|_a^2-L(w)
\right]
$$

$$
\therefore\quad
2J\left(\frac{u+w}{2}\right)
+\frac{1}{4}\|u-w\|_a^2
=
J(u)+J(w)
$$

$$
\boxed{
J(u)+J(w)-2J\left(\frac{u+w}{2}\right)
=
\frac{1}{4}\|u-w\|_a^2
}
$$
or

$$
\frac14\|u-w\|_a^2
=
J(u)+J(w)
-
2J\left(\frac{u+w}{2}\right).
$$

### 1. Using this result, we claim that $u_n $ is Cauchy in V.

We claim that

$$
\|u_n-u_m\|_a \rightarrow 0
$$
Proof:

We have,
$$
2J\left(\frac{u_n+u_m}{2}\right)
+\frac{1}{4}\|u_n-u_m\|_a^2
=
J(u_n)+J(u_m).
$$

Therefore,

$$
\frac{1}{4}\|u_n-u_m\|_a^2
=
J(u_n)+J(u_m)
-2J\left(\frac{u_n+u_m}{2}\right).
$$

Since

$$
J\left(\frac{u_n+u_m}{2}\right)\geq J_{\min},
$$

we obtain

$$
\frac{1}{4}\|u_n-u_m\|_a^2
\leq
J(u_n)+J(u_m)-2J_{\min}.
$$

Hence,

$$
\frac{1}{4}\|u_n-u_m\|_a^2
\leq
\left[J(u_n)-J_{\min}\right]
+
\left[J(u_m)-J_{\min}\right].
$$

Since
$$
J(u_n)-J_{\min}\rightarrow 0,
\qquad
J(u_m)-J_{\min}\rightarrow 0,
$$

we have

$$
\|u_n-u_m\|_a^2\rightarrow 0.
$$

Thus,

$$
\boxed{\|u_n-u_m\|_a\rightarrow 0}
$$

and therefore  **$u_n $ is Cauchy Sequence in V**.

$$
\boxed{u_n\rightarrow u\quad\text{in }V}
$$
 
### 2.

$J(u_n)\rightarrow J_0 \Rightarrow \{u_n\}$ is Cauchy in $V$.

Since $V$ is complete,

$$
u_n\rightarrow u \quad \text{in } V
$$

Therefore,

$$
J(u_n)\rightarrow J(u)=J_{\min}
$$

since $J$ is a continuous function in $u$.

Reason: If $f$ is a function, by continuity of $f$,

$$
u_n\rightarrow u
\quad\Rightarrow\quad
f(u_n)\rightarrow f(u).
$$

We have shown that the variational problem $(V)$ is equal to the weak problem:

$$
\boxed{\text{solution }u \Longleftrightarrow \text{weak solution}}
$$

$$
J(w)=J(u)+\frac{1}{2}\|u-w\|_a^2
$$

This is true only when $u$ is the converged solution.

With $u'$ as another discrete solution,

$$
J(u')=J(u)+\frac{1}{2}\|u-u'\|_a^2
$$

so

$$
J(u')\geq J(u).
$$

As $u'\rightarrow u$,

$$
\|u-u'\|_a^2\rightarrow0.
$$

Hence,

$$
J(u') = J(u).
$$

Therefore if energy converges, the function converges


### Finally

We have shown that

$$
\boxed{
u\text{ solves the weak problem}
\iff
u\text{ minimizes }J
}
$$

where

$$
J(v)=\frac12a(v,v)-L(v).
$$

**Thus, solving the Poisson weak problem is equivalent to finding the minimum of the energy functional**.

## Riesz Representation Theorem

Let $H$ be a Hilbert space with an inner product

$$
(\cdot,\cdot).
$$

The Riesz representation theorem states that if H is a Hilbert space with inner product $(\cdot,\cdot)$ then every bounded linear functional is of the form
$$
\boxed{
L(v)=(u_L,v)
}
$$

for every

$$
v\in H.
$$
Thus, a linear functional can be represented by an element of the Hilbert space.



### For the weak problem,

$$
a(u,v)=L(v).
$$

If we define the energy inner product as

$$
(u,v)_a=a(u,v),
$$

then the weak problem becomes

$$
\boxed{
(u,v)_a=L(v)
}
$$

for every $v\in V$.

In other words, when $H=V$ $\Rightarrow$ $(\cdot,\cdot)=a(\cdot,\cdot)$ and $L(v)=\int fv$.
The weak solution can therefore be interpreted as the **Riesz representative** of the linear functional $L$ with respect to the energy inner product.

### Checking the Residual Against Different Test Functions

Recall $\text{res}(w,v) = a(w,v) - L(v)$ measures how far $w$ is from satisfying the weak form for a given test function $v$.

We now check this residual using different combinations of trial function ($u_{exact}$, $u_h$, $u_i$) and test function ($v_i$, $v_{func}$):

- $u_{exact}$ — the exact (analytical) solution
- $u_h$ — the finite element solution (solves the discrete weak form)
- $u_i$ — the interpolant of $u_{exact}$ into $V_h$ (i.e. $u_{exact}$ evaluated at the nodes, not the FE solution itself)
- $v_i$ — a test function in the discrete space $V_h$
- $v_{func}$ — a smooth test function ($x(1-x)y(1-y)$) that lives in $V$ but **not** in $V_h$

**Residuals:**

- $\text{res}(u_{exact}, v_i) \approx 0$ and $\text{res}(u_{exact}, v_{func}) \approx 0$: the exact solution satisfies $a(u,v) = L(v)$ for *any* $v \in V$.
- $\text{res}(u_h, v_i) \approx 0$: by construction, $u_h$ only satisfies the weak form for test functions in $V_h$.
- $\text{res}(u_h, v_{func}) \neq 0$: $u_h$ does **not** satisfy the weak form when tested against a function outside $V_h$.
- $\text{res}(u_i, v_i) \neq 0$ and $\text{res}(u_i, v_{func}) \neq 0$: the interpolant $u_i$ is *not* a solution of the discrete weak form.
- $\text{res}(u_{exact}-u_h, u_h) \approx 0$: this is the **Galerkin orthogonality** result — the error $u_{exact}-u_h$ is orthogonal (in the energy inner product) to the finite element solution.

This distinguishes three related but different solutions: the FE solution $u_h$ (solves the discrete problem), the interpolant $u_i$ (matches nodal values only), and the exact solution $u_{exact}$ (solves the full continuous problem).
